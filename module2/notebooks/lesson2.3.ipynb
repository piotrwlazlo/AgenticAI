Edukacja to nie sprint. To **maraton**. Tutaj wchodzimy na głęboką wodę.

Gdy budujesz AI Tutora do nauki języka (lub czegokolwiek innego), nie interesuje Cię jedna transakcja. Interesuje Cię proces, który trwa tygodnie lub miesiące. Tutaj pojawia się brutalna prawda: architektura, która świetnie sprawdziła się przy taksówkach, tutaj polegnie w pierwszej godzinie.

Dlaczego? Ponieważ w edukacji (i procesach długotrwałych) sukces nie zależy od "dowiezienia odpowiedzi", ale od zmiany stanu w głowie użytkownika.

## Inżynieria kontekstu w procesach ciągłych

W poprzednich lekcjach tego modułu zbudowaliśmy solidny fundament oparty na pamięci semantycznej (faktach). Ta warstwa jest kluczowa i nigdzie nie znika. Jednak, aby wejść na wyższy poziom inżynierii systemów edukacyjnych, musimy ten fundament wzbogacić o pamięć proceduralną. O ile ta pierwsza przechowuje informacje (np. definicje, zasady), o tyle ta druga mapuje konkretne umiejętności (*skills*). Dzięki temu system nie tylko "wie", o czym rozmawialiśmy, ale zaczyna "czuć" dynamikę progresu – rozumie, że tydzień temu czas *Past Simple* był barierą, wczoraj poszło nieźle, a dzisiaj warto zrobić szybki test utrwalający ścieżkę neuronową.


To, co tutaj zbudujemy, to uniwersalny wzorzec *Agenta AI długotrwałego procesu*. Choć skupiamy się na nauce języka (Language Tutor), ta sama architektura obsłuży każdy z poniższych, popularnych problemów biznesowych:

1.  Doradca finansowy. Nie wystarczy powiedzieć "Kup akcje X". Trzeba znać historię portfela, tolerancję ryzyka i cele emerytalne klienta, które zmieniają się w czasie.
2.  Trener personalny (Fitness). To nie jest jednorazowy plan treningowy. To ciągła adaptacja do progresu (lub kontuzji) podopiecznego tydzień po tygodniu.
3.  Onboarding pracownika. Proces wdrażania w korporacji to nie jednorazowy e-mail, ale sekwencja zadań, sprawdzania wiedzy i odblokowywania kolejnych dostępów.
4.  Mentoring techniczny (Code Reviewer). Agent, który pamięta, że junior developer ma tendencję do nieoptymalnych zapytań SQL i zwraca na to uwagę w kolejnych projektach.
5.  Coaching kariery. Długofalowe planowanie ścieżki zawodowej, a nie jednorazowa korekta CV.

W tej lekcji nie budujemy "gadającego bota". Budujemy system operacyjny dla kompetencji. Zamiast "Answer Machine", stworzymy "Socratic Tutor". Co o tym myślisz? Brzmi jak wyzwanie? I o to chodzi. Zaufaj mi, że po tej lekcji zupełnie inaczej spojrzysz na to, co znaczy "pamięć" w systemach AI.

### Zmiana paradygmatu: Od podręcznika do realnego życia

Naszym celem inżynierskim NIE jest stworzenie bota, który karze za brak przecinka.
Naszym celem jest stworzenie *communication coacha*.

Chcemy, aby użytkownik był w stanie:
1.  **Rozumieć** native speakerów (którzy rzadko mówią jak lektor z płyty CD).
2.  **Być zrozumianym**, używając naturalnych fraz, a nie sztywnych formułek, które brzmią nienaturalnie (tzw. "robot language").

Z inżynierskiego punktu widzenia, uczymy model rozpoznawania i sugerowania **wzorców komunikacyjnych**, a nie tylko reguł składniowych.

### 🔬 Case Study: podręcznik vs rzeczywistość 

Spójrzmy na różnicę. Nasz Agent musi znać obie te warstwy, ale dążyć do tej prawej (Real Life). To jest ta wartość, której szukamy.

| Sytuacja | Wersja "podręcznikowa" 🤖 | Wersja "Real Life" (Native Speaker) 🗣️ | Dlaczego to ważne? |
| :--- | :--- | :--- | :--- |
| **Rezerwacja** | "Hello. I would like to reserve a table for two people for 7 PM." | "Hi, can I get a table for two at seven?" | Native speaker stawia na szybkość i płynność ("flow"). Wersja pierwsza jest poprawna, ale sztywna. |
| **Powitanie** | "How are you doing today?" <br> "I am fine, thank you." | "How's it going?" <br> "Good, you?" | Jeśli użytkownik usłyszy szybkie "How's it going?", może zgłupieć, jeśli zna tylko wersję szkolną. |
| **Niezrozumienie** | "I do not understand what you are saying." | "Sorry, I didn't catch that." | Użycie frazy "catch that" jest sygnałem: "Znam ten język, mów do mnie normalnie". |

Nasz system ma zasypywać tę przepaść (gap). Ma uczyć "języka ludzi", a nie "języka książek".


### Fundamenty Strategiczne: Punkt A i Punkt B (analiza luk)

Skoro wiemy, *czego* uczymy (komunikacji), musimy ustalić ramy procesu. W nauce języka kluczowe są dwa stany. Bez ich zdefiniowania Agent błądzi we mgle.

Spójrz na poniższy diagram, który mapuje logikę "dużego obrazka" (Big Picture):


![](../images/edu_diagram_1.jpg)


**Co tu się dzieje?**
1.  **Analiza (lewa strona):** Agent pobiera stan obecny (punkt A - np. "rozumiem, ale nie mówię") i cel użytkownika (punkt B - np. "swobodna kolacja z klientem").
2.  **Gap Analysis (środek):** System oblicza różnicę (Delta). To jest kluczowy moment inżynierski. Jeśli delta jest za duża ("Od zera do native speakera"), planowanie się załamie. Musimy znaleźć realistyczną ścieżkę.
3.  **Plan (prawa strona):** Delta zamieniana jest na konkretne kamienie milowe (Milestones). To one są realizowane w trakcie kolejnych sesji.


### Taktyka: anatomia sesji (*sequence*)

Skoro nauka to maraton, musimy go podzielić na etapy. Wprowadzamy pojęcie **Sesji**. Jest to nasza atomarna jednostka interakcji. Nieważne, czy sesja odbywa się raz dziennie czy raz w tygodniu – struktura pozostaje ta sama.

Poniższy diagram sekwencji pokazuje "przepływ" pojedynczego spotkania z Agentem:

![](../images/edu_diagram_2.jpg)

**Analiza przepływu:**
1.  Inicjacja. Zanim zaczniemy nową naukę, Agent *musi* sprawdzić pracę domową z poprzedniej sesji. To buduje ciągłość procesu (pamięć).
2.  Faza niebieska (*core*):** To jest właściwa nauka. Tutaj odbywa się roleplay, symulacja rozmowy czy ćwiczenie wymowy.
3.  Faza zielona (*wrap-up*): Meta-kognicja. Agent podsumowuje, co poszło dobrze ("Wins") i co wymaga poprawy ("Gaps"), a następnie zadaje pracę domową na kolejną sesję.



### Operacja: pętla uczenia 

Na koniec musimy zrozumieć, że ten proces nigdy nie jest linią prostą. To pętla. Wynik jednej sesji staje się wejściem (Input) dla kolejnej.

### Mapa Architektury

Wyobraź sobie nasz stan (`TutorAgentState`) jako drzewo. Korzeń jest jeden, ale ma trzy główne gałęzie, z których każda odpowiada za inny aspekt "mózgu" agenta AI:

```text
TutorAgentState (ROOT)
│
├── 1. CORE (Tożsamość - DNA)
│   └── "Kim jestem?" (Stałe instrukcje, których nie wolno złamać)
│
├── 2. PROFILE (pamięć semantyczna - student)
│   ├── Kim jest Marek? (język, poziom)
│   ├── Jaki jest cel? (LearningGoal - SMART)
│   └── Jak go uczyć? (Psychometria: Styl + Polityka Błędów)
│
└── 3. SESSION (pamięć krótkotrwała - tu i teraz)
    ├── Co robimy w tej sekundzie? (Faza: sprawdzanie czy nauka?)
    └── Jaki jest temat? (Context Window)
```

Widzisz to? Każda linijka kodu, którą za chwilę napiszesz, wpadnie do jednego z tych trzech pudełek.

W praktyce to oznacza, że chronimy się przed *wyciekiem danych między użytkownikami*. Gdybyś przypisał tu zwykłą pustą listę `[]` lub słownik `{}`, Python stworzyłby ten obiekt tylko **raz** w momencie startu programu. Efekt? Wszyscy Twoi studenci dzieliliby tę samą pamięć! Jeśli Marek nauczyłby się słówka, system uznałby, że Kasia też już je umie. Użycie `default_factory` wymusza stworzenie *nowego, czystego obiektu* dla każdej nowej instancji klasy.

## Pamięć epizodyczna - kompresja rzeczywistości

Wiem, że powtarzam to jak zdarta płyta. Mówiłem o tym wcześniej. Ale powtórzę to raz jeszcze, bo to najważniejsza zasada inżynierii Software 3.0:

> **Kontekst to Twoje najcenniejsze zasoby. Szanuj go.**

W poprzednich lekcjach, omawiając warstwy pamięci, zdefiniowaliśmy zasadę **🟢 Compress**. Tam pamięć epizodyczna służyła do tego, by Agent nie uczył się w kółko na błędach (*"Uwaga, oni wymagają przedpłaty"*). Bez tego, przy każdym telefonie system byłby tak samo zaskoczony.

Tutaj stawka jest jeszcze wyższa.
W edukacji (proces długotrwały) historia rozmów rośnie lawinowo. Jeśli będziesz wrzucać do prompta surowy zapis wszystkich lekcji z ostatniego miesiąca, to:
1.  **Zbankrutujesz** (płacisz za każdy token wejściowy).
2.  **Zabijesz latency** (użytkownik uśnie, zanim model przetworzy 50 stron tekstu).
3.  **Otumanisz model** (zjawisko *Lost in the Middle* – model gubi się w szumie informacyjnym).

Dlatego *pamięć epizodyczna* w naszym systemie nie jest "magazynem logów". Jest "inteligentną niszczarką". Brzmi paradoksalnie? Budujemy pamięć poprzez usuwanie danych? A jednak – czy właśnie tak nie działa Twój własny mózg? Nie pamiętasz każdej klatki obrazu z drogi do pracy ani koloru butów mijanego przechodnia. Twój mózg agresywnie filtruje 99% szumu, aby zachować jedynie 1% (a czasem nawet 0,00001%) sygnału. Wyobraź sobie, że byłoby inaczej – że pamiętałbyś wszystko, każdy milimetr, każdą sekundę... Twój wewnętrzny system operacyjny (Twój "context window" byłby przepełniony) natychmiast by się zawiesił. Tutaj robimy dokładnie to samo: zadaniem tej warstwy jest wziąć tysiąc słów rozmowy i pozostawić z nich pięćdziesiąt słów czystej esencji.

To jest proces **ETL (Extract, Transform, Load)** dla AI. Sprawdźmy to w praktyce.

### Symulacja "brudnego" kawałka dialogu (Raw Stream)

Wyobraź sobie, że właśnie zakończyła się 15-minutowa sesja z Markiem. Zobacz, co wydarzyło się w logach. Zauważ, że ten tekst jest pełen "śmieci": powitań, dygresji o kocie i błędów gramatycznych. To jest nasz surowiec.

````python
raw_conversation_log = """
Tutor: Hi Marek! Ready to practice your presentation?
Marek: Yes, hello. I have stress because boss say I go to New York next month.
Tutor: Oh, New York! That's exciting. Don't worry about stress. What exactly will you do there?
Marek: I must to speak about Q3 results. I thinked about logic structure, but words is hard.
Tutor: I see. Remember, we say "I thought", not "I thinked". And "words are hard".
Marek: Ah, yes. Thought. The words are hard. My cat, Luna, jumped on keyboard yesterday and I lost my notes.
Tutor: Haha, classic Luna. Okay, let's focus on the opening. How will you start?
Marek: I want say: "Welcome everybody on the meeting".
"""
````

### Definicja struktury 

Czego potrzebujemy z tego szumu? Pamiętasz diagram? Interesują nas:
1.  **Podsumowanie:** O czym to było? (dla kontekstu)
2.  **Fakty osobiste:** Żeby za tydzień zapytać "Jak tam przygotowania do NY?" (budowanie relacji)
3.  **Błędy (Knowledge Tracing):** Żeby system wiedział, co ćwiczyć w przyszłości (pamięć proceduralna)

Model nie ma prawa nam tu "opowiadać". Ma wypełnić formularz.

Zróbmy dwie struktury:
- `MistakeLog` - definiujemy strukturę pojedynczego błędu
- `EpisodeMemory` - definiujemy strukturę notatki epizodycznej

````python
class MistakeLog(BaseModel):
    user_phrase: str = Field(description="The incorrect phrase used by user")
    correction: str = Field(description="The correct form")
    grammar_rule: str = Field(description="Name of the grammar rule broken (e.g. Past Simple Irregular)")

class EpisodeMemory(BaseModel):
    summary: str = Field(description="One sentence summary of the session focus")
    personal_facts: List[str] = Field(description="Facts about user's life (non-grammar related, e.g. names, plans)")
    mistakes: List[MistakeLog] = Field(description="List of linguistic errors detected")
    next_step_recommendation: str = Field(description="What should we practice next based on this?")
````

To jest nasza 'foremka' (struktura), do której wlejemy płynną inteligencję modelu.

### Bilans zysków i strat (sztuka kompresji)

Zatrzymajmy się na chwilę. Zrozumienie tego mechanizmu to różnica między juniorem a seniorem w świecie AI.

Z jednej strony mamy **ZYSKI**:
1.  **Redukcja o 90-99%:** Z kilkuset tokenów "brudnego" tekstu zrobiliśmy mały obiekt JSON. Przy następnej sesji nie wczytamy całego dialogu ("Cześć, jak leci..."), tylko sam fakt: `cat_name: Luna`.
2.  **Relacyjność na autopilocie:** Dzięki polu `personal_facts`, system "pamięta" o kocie czy podróży do NY, zużywając na to grosze.
3.  **Analityka:** Lista `mistakes` to twarde dane. Po miesiącu możesz wygenerować wykres: "Marek, 80% Twoich błędów to czasowniki nieregularne".

Ale tutaj wchodzi **RYZYKO** (i brutalna prawda): to jest **kompresja stratna**. Liczba informacji na wyjściu jest (znacznie) mniejsza niż na wejściu.

Jeśli Twój `system_prompt` lub struktura JSON (Schema) nie przewidziała pola na "nastrój użytkownika", a Marek był wczoraj wściekły – ta informacja **zniknie bezpowrotnie**. Przepadnie. Model ją odrzuci jako "szum".

#### Co to oznacza w praktyce?
Że projektowanie struktury `EpisodeMemory` to sztuka przewidywania przyszłości.
- Zbyt ogólna struktura? Zgubisz kluczowe detale (np. datę wylotu do NY).
- Zbyt szczegółowa? Zapchasz kontekst śmieciami, których nie potrzebujesz.

W zaawansowanych systemach (Software 3.0) stajemy tu przed dylematem architektonicznym: **Jeden "Super-Call" czy kilka precyzyjnych strzałów?**

Teoretycznie możesz rozbudować strukturę JSON (Schema) tak, aby wyciągnąć wszystko za jednym razem: daty, emocje i gramatykę. To tańsze i szybsze. Ale pamiętaj o **inżynierii uwagi (Attention Engineering)**. Model ma ograniczoną "pojemność skupienia". Jeśli każesz mu analizować wszystko naraz, jakość detekcji w każdym z obszarów może spaść.

Dlatego w systemach produkcyjnych czasem celowo puszcza się ten sam log przez **osobne filtry**, sterowane różnymi System Promptami:
- **Filtr Analityczny:** "Ignoruj emocje, ekstrahuj tylko twarde daty i miejsca"
- **Filtr Psychologiczny:** "Nie interesują mnie fakty. Skup się na tonie, frustracji i motywacji"
- **Filtr Dydaktyczny:** "Ignoruj treść. Wypisz tylko błędy składniowe"

**"Ale to kosztuje 3x więcej tokenów!"** – powiesz.
Tak. Ale w inżynierii kontekstu musisz patrzeć na ROI (zwrot z inwestycji):
1.  **Latency nie mają tak dużego znaczenia:** Te procesy dzieją się zazwyczaj **asynchronicznie** (w tle, np. raz na dobę), więc użytkownik tego nie czuje.
2.  Koszt kilku centów za API jest pomijalny, jeśli "filtr psychologiczny" wykryje, że Twój kluczowy klient jest sfrustrowany i chce odejść. Tego nie chcesz "kompresować stratnie".


Pamiętaj: to, że zmniejszyliśmy objętość, jest super. Ale w inżynierii zawsze jest coś za coś. Twoim zadaniem jest znalezienie złotego środka między jednym tanim strzałem, a drogą, ale precyzyjną analityką.

## Podsumowanie

W tej lekcji wykonaliśmy kawał solidnej roboty inżynierskiej. Zamiast bawić się w "gadanie z botem", zeszliśmy do poziomu architektury systemowej. Oto twarde fakty, które zabierasz ze sobą:

1.  Stan to hierarchia, a nie płaska lista. Zrozumieliśmy, że profesjonalny Agent potrzebuje podziału na `Core` (niezmienne zasady), `Profile` (dane długoterminowe) i `Session` (tu i teraz). Użycie `default_factory` w *Pydantic* uratowało nas przed wyciekiem danych między użytkownikami.
2.  Prompt to kod. Zamiast ręcznego sklejania stringów na system szablonów Jinja2. Dzięki temu logika warunkowa (np. "jeśli uczeń jest wrażliwy, bądź delikatny") siedzi w warstwie prezentacji, a nie w kodzie Pythona.
3.  Pamięć to sztuka zapominania. Zamiast wrzucać do kontekstu całą historię czatu, zbudowaliśmy mechanizm ETL, który ekstrahuje tylko to, co ważne (fakty i błędy), a resztę bezlitośnie odrzuca. To jest klucz do skalowalności.
4.  Struktura ponad treścią. Zobaczyliśmy, że LLM radzi sobie świetnie jako parser wyciągający dane do formatu JSON (`EpisodeMemory`). To pozwala nam traktować wyjście z modelu jako bezpieczny wsad do bazy danych.
5.  Symulacja przed wdrożeniem. Zanim wysłaliśmy zapytanie do modelu, wyrenderowaliśmy prompt lokalnie. To fundamentalna praktyka Software 3.0 – musisz widzieć, co dokładnie "widzi" model, zanim zaczniesz debugować jego odpowiedzi.
6.  Pętla sprzężenia zwrotnego. Zamknęliśmy obieg: wynik poprzedniej sesji (JSON z błędami) stał się integralną częścią instrukcji systemowej dla nowej sesji. To jest moment, w którym "chatbot" staje się "tutorem".
7.  Modelowanie domeny. Zdefiniowaliśmy edukację nie jako serię pytań i odpowiedzi, ale jako proces zamykania luki między językiem podręcznikowym a żywą komunikacją.
