## Mapa terenu: 7 poziomów lub spektrum autonomii AI

Jednym z największych wyzwań, jakie dziś widzimy w projektach AI, jest gigantyczny chaos komunikacyjny. Aby go rozwiązać i dać Ci realne, inżynierskie narzędzie, w DataWorkshop Lab opracowaliśmy **7-poziomowe Spektrum Autonomii AI**.

Pewnie myślisz: "Zaraz, zaraz, przecież istnieje już wiele klasyfikacji poziomów AI". I to prawda. Widzieliśmy, jak na ten problem patrzą giganci tacy jak Google czy Anthropic, a także wielu naukowców. W naszym odczuciu ich podejścia, choć wartościowe, nie rozwiązywały fundamentalnego problemu, z którym mierzymy się w praktyce. Potrzebowaliśmy mapy, która byłaby prosta, pragmatyczna i zrozumiała dla każdego – od osoby z biznesu, przez architekta, po inżyniera AI, który będzie pisał kod.


### Dlaczego ta mapa powstała?
Pozwól, że opowiem Ci krótką historię z DataWorkshop Lab.

W ostatnich latach o AI robi się tyle szumu, że coraz trudniej się dogadać. Pojawiają się mity, buzzwordy, hype... i każdy mówi o czymś innym, choć używa tych samych słów.

Najprostszy przykład:
- CEO: "Potrzebujemy agenta AI do X (np. obsługi klientów)."
- CTO: "Spoko, mamy LangChain, RAG i bazę wektorową (lub nawet przypomina, że może być jeszcze grafowa) — jesteśmy prawie gotowi."
- Inżynier (patrzy spod oka): "Ekhm… coś tu nie pasuje." Wie, że to, co nazywają "agentem AI", w praktyce rozkleja się po pięciu minutach na produkcji"

I choć wszyscy mówili o agencie AI, każdy miał w głowie zupełnie inną rzecz.
- CEO – wizję Jarvisa z Iron Mana
- CTO – zestaw bibliotek, które brzmią bardzo "pro" (zgodnie z trendami buzzwordów), więc pewnie wystarczą
- Inżynier – czuje, że wjeżdżają w strefę magii, bo sam widzi, że to nadal bardziej chatbot z wyszukiwarką niż jakikolwiek "agent AI"

W pewnym momencie zrozumiałem: brakuje nam wspólnego języka.
I dokładnie ten problem rozwiązuje ta klasyfikacja.


Dlatego nasza mapa powstała z jednej potrzeby: aby ludzie **zaczęli się dogadywać** (tak, tak — z definicji to trudne, ale przynajmniej w tym kierunku dążymy). Pozwala precyzyjnie określić trzy kluczowe rzeczy:
1.  **Jak złożony jest problem**, który próbujemy rozwiązać?
2.  **Jakich kompetencji potrzebujemy**, aby go dowieźć?
3.  **Czego realnie możemy oczekiwać** od budowanego systemu?

## Jak to robiliśmy?

Wzięliśmy dwa skrajne przypadki:
- Poziom 1: Piszesz prompt w ChatGPT lub podobne UI
- Poziom 7: AI nie potrzebuje już ludzi (mam nadzieję, że nigdy tam nie dojdziemy)

Między nimi zmapowaliśmy wszystko, co faktycznie istnieje lub jest w zasięgu technologii w najbliższych latach.
Kluczowy wyróżnik: adaptowalność (autonomiczność), nie *features*.

- Poziom 1-2: Człowiek musi sprawdzać każdy output ("human in the loop")
- Poziom 3-4: System może działać semi-autonomicznie w znanych scenariuszach
- Poziom 5-6: System sam decyduje i adaptuje się do nowych sytuacji
- Poziom 7: System ewoluuje bez człowieka (w DataWorkshop nie budujemy tego!)


### Iluzja kompetencji

Zanim przejdziemy do samych poziomów, musimy porozmawiać o poważnej pułapce, w którą wpada dziś ogromna część rynku. Nazwijmy ją **iluzją frameworków**.

Wygląda to tak:
*   **Biznes słyszy "Agent AI"** i od razu oczekuje systemu o zdolnościach z poziomu 5 (co najmniej) – inteligentnego, strategicznego partnera
*   **Inżynier, który dopiero zaczyna**, sięga po popularne frameworki, takie jak LangChain, CrewAI czy AutoGen. Widzi w dokumentacji hasła takie jak "pamięć agenta", "narzędzia" czy "planowanie"
*   **Pojawia się iluzja.** Inżynier buduje demo, które działa dla jednego, prostego przypadku. Wygląda magicznie. Wszyscy myślą, że są już na Poziomie 4 lub 5.

**Brutalna prawda jest taka: nie są!** Istnieje gigantyczna przepaść między demo, które działa przez pięć minut dla jednego użytkownika, a produkcyjnym systemem, który ma działać stabilnie przez lata, na dużą skalę i radzić sobie z tysiącami nieprzewidywalnych scenariuszy. Te frameworki dają pozór posiadania pewnych zdolności, ale nie dają inżynierskich fundamentów do budowy solidnych, adaptacyjnych systemów. To najgorsza ze "złych uliczek", bo daje fałszywe poczucie bezpieczeństwa, które kończy się porażką projektu.

### Spektrum w praktyce od pełnej kontroli do pełnej autonomii

Pomyśl o tym spektrum jako o suwaku, który określa, jak bardzo potrzebny jest człowiek w pętli (*Human-in-the-Loop*). To jest kluczowy wyróżnik: *zdolność systemu do adaptacji*.

- Na poziomie 1, `Human-in-the-Loop` jest obowiązkowy w 100%. Człowiek manualnie musi weryfikować każdy wynik. System nie ma żadnej zdolności adaptacji.
- Na poziomie 7, `Human-in-the-Loop` nie istnieje. System jest w pełni autonomiczny i sam podejmuje decyzje.
- Wszystko pomiędzy to odcienie szarości, które precyzyjnie określają, jak bardzo system potrafi samodzielnie adaptować swoje działanie do nowych okoliczności.

A teraz zobaczmy, jak to wygląda w podziale na fazy:

#### FAZA 1: FUNDAMENTY (Poziomy 1-2) – Zero adaptacji
- **Poziom 1: Konwersacyjne AI** – To po prostu interakcja z modelem językowym, np. przez interfejs typu ChatGPT. Zero automatyzacji, zero narzędzi, zero pamięci.
- **Poziom 2: Programowalne AI** – Inteligentna automatyzacja (RPA 2.0). System potrafi generować ustrukturyzowane odpowiedzi (np. JSON) i wywoływać inne systemy, ale podąża sztywno zdefiniowaną ścieżką. Nadal brak adaptacji.

#### FAZA 2: POŁĄCZONE AI (Poziomy 3-4) – Początki adaptacji
- **Poziom 3: Zintegrowane AI** – System potrafi inteligentnie wybrać jedną z kilku predefiniowanych ścieżek (routing). To tutaj znajduje się większość systemów RAG. System integruje się z zewnętrzną wiedzą, ale nie "działa" w świecie.
- **Poziom 4: Kontekstowe AI** – **Pierwszy prawdziwie agentowy poziom.** System zyskuje dwie supermoce: **Pamięć** (pamięta poprzednie interakcje) i **Narzędzia** (potrafi aktywnie działać w zewnętrznym świecie, np. wywoływać API). Tutaj zaczyna się prawdziwa adaptacja.

#### FAZA 3: AUTONOMICZNE AI (Poziomy 5-7) – Pełna adaptacja
- **Poziom 5: Strategiczne AI** – System potrafi nie tylko używać narzędzi, ale także **adaptacyjnie planować** ich użycie. Zaczyna "myśleć" w ustrukturyzowany sposób, tworzyć strategie i modyfikować je w trakcie działania.
- **Poziom 6: Meta-autonomiczne AI** – System, który potrafi sam siebie modyfikować i ewoluować – np. tworzyć własne, nowe schematy rozumowania. To jest poziom, na którym agent "zaczyna budować własne tory".
- **Poziom 7: Pełna Autonomia** – Teoretyczny punkt skrajny. System działający w pełni niezależnie od człowieka. Mam nadzieję, że ten poziom nigdy nie nastąpi, bo jest niebezpieczny, i **nie uczymy, jak go budować**. Umieszczamy go na mapie jako punkt odniesienia, który pozwala nam nawigować i rozumieć, jak daleko jeszcze jesteśmy.

## Mity

### Mit #1: "Mam agenta AI = jestem na poziomie 4+"

Często zakłada się, że jeśli system nazywa się "agentem", to musi być zaawansowany. 
Nie! Nic bardziej mylnego. Agent może być na każdym poziomie:
- `L1` Agent AI: ChatGPT z custom instructions (to też  może być "agent AI"!)
- `L2` Agent AI: Bot z fixed workflow
- `L3` Agent AI: Router do innych systemów
- `L4` Agent AI: Agent AI z pamięcią i narzędziami (dopiero tu zaczyna się prawdziwa agencja)
- `L5` Agent AI: Agent AI ze strukturalnym rozumowaniem

☝️ Zwróć uwagę, że `L1` oznacza `Level 1` (czyli poziom 1) itd.


Dlatego według naszej klasyfikacji, **agenta AI można zbudować na każdym z siedmiu poziomów**. Możesz mieć bardzo prostego agenta z poziomu 1, który jedynie opakowuje prompt, jak i super-zaawansowanego z poziomu 5. Dlatego słowo "agent AI" nie definiuje poziomu zaawansowania.



### Mit #2: "Mam multi-agentów = jestem na poziomie 6"

To kolejna popularna pułapka. "Mamy wiele agentów, które ze sobą rozmawiają, więc to na pewno poziom 6!". Niekoniecznie. 


Możesz mieć 100 głupich agentów na poziomie 1. To jak mieć 100 kalkulatorów zamiast jednego komputera.
Multi-agent to architektura, nie poziom zaawansowania. Możesz mieć multi-agentów na L2 (każdy robi swoją część sztywnego workflow) albo na L5 (koordynują się adaptacyjnie).

Dlatego liczba agentów nie wpływa na poziom w spektrum AI. Kluczowa jest **zdolność do adaptacji** pojedynczego agenta i całego systemu.


### Mit #3: "Każdy problem wymaga poziomu 5+"
NIE. I to jest być może najważniejsza, praktyczna lekcja płynąca z tej mapy.

Cała sztuka polega na tym, by poprawnie zdiagnozować złożoność problemu i dobrać do niego odpowiedni poziom kompetencji. Próba rozwiązania prostego zadania metodami z Poziomu 5 byłaby jak strzelanie z armaty do komara – nieefektywna, kosztowna i niepotrzebnie skomplikowana.

Spójrz na to w ten sposób:
- Chatbot dla ad-hoc zapytań? L1 wystarcza (i działa dla milionów użytkowników!)
- Przypisanie kategorii do artykułu (komentarza)? L2 załatwi sprawę
- System wsparcia z bazą wiedzy? L3 jest optymalny
- Personal assistant z pamięcią? L4 jest tym, czego szukasz
- Analityk strategiczny? Tu potrzebujesz L5

A teraz dodajmy do tego pewną życiową obserwację, która przygotuje Cię na przyszłość. Gdy tylko kompetencje do budowania produkcyjnych systemów na poziomach 4 i 5 staną się na rynku powszechne – a dzięki takim programom jak ten tak stanie się to możliwe – zobaczysz ciekawą zmianę. Biznes, widząc nowe, realne możliwości, naturalnie zacznie chcieć "upgrade'ować" swoje dotychczasowe problemy. Zadania, które dziś rozwiązuje się na Poziomach 2 i 3, będą redefiniowane tak, aby wykorzystać potencjał pamięci, narzędzi i planowania z Poziomów 4 i 5.

Dlatego Twoim zadaniem jako architekta AI (możesz pisać kod lub zarządzać zespołem, który pisze kod) jest nie tylko budować, ale przede wszystkim trafnie diagnozować. To jest esencja dojrzałości inżynierskiej w świecie AI.

Teraz czas sprawdzić, na ile dobrze rozumiesz, o czym mówimy. Spokojnie — to może być jeszcze trochę niejednoznaczne, ale wierzę, że szybko złapiesz tę ideę. Gdy ją tworzyliśmy, bardzo mocno stawialiśmy na prostotę użycia, bo na tym zwykle polegają genialne pomysły :)

## 🧠 Zadanie 1.1.1: Jaki to poziom autonomii AI?

Pamiętaj, **minimalny** jest tu słowem-kluczem. Oczywiście, problem z poziomu 2 można rozwiązać narzędziami z poziomu 4, ale nigdy odwrotnie – problemu z poziomu 4 nie da się poprawnie rozwiązać, mając tylko kompetencje z poziomu 2.

Wynik dla każdej opinii musi być zwrócony w ściśle określonym formacie JSON, gotowym do dalszej analizy.

Cała jego wiedza pochodzi **wyłącznie** z firmowej bazy dokumentów w SharePoint, która jest regularnie aktualizowana przez dział HR. System nie wykonuje żadnych innych akcji.

**Jaki jest minimalny wymagany poziom?**

**Przypadek #3: Asystent rezerwacji podróży**

Agent AI prowadzi konwersację z pracownikiem na temat planowanej podróży służbowej. Po zebraniu kluczowych informacji (cel, daty, preferencje), agent musi **wywołać zewnętrzne API** linii lotniczych, aby sprawdzić dostępne loty. Dodatkowo, system **pamięta preferencje użytkownika** (np. miejsce przy oknie, preferowana linia lotnicza) z jego poprzednich rezerwacji i uwzględnia je w propozycjach.

**Jaki jest minimalny wymagany poziom?**
<br>

**Przypadek #4: Analityk danych SQL**

System, który pozwala analitykom biznesowym zadawać pytania w języku naturalnym (np. "pokaż mi sumę sprzedaży dla produktu X w ostatnim kwartale"). Agent ma za zadanie **przetłumaczyć to pytanie na poprawne zapytanie SQL**, wykonać je na firmowej bazie danych i zwrócić użytkownikowi odpowiedź w formie tabeli lub podsumowania.

**Jaki jest minimalny wymagany poziom?**
<br>

**Przypadek #5: Diagnosta systemów IT**

Agent AI, którego zadaniem jest zdiagnozowanie problemu technicznego w złożonym systemie. Otrzymuje od inżyniera opis błędu. Musi **samodzielnie zdecydować**, jakie logi systemowe sprawdzić, jakie testy diagnostyczne uruchomić (za pomocą dostępnych narzędzi/skryptów), a na podstawie zebranych wyników **postawić hipotezę** co do przyczyny problemu i **zaproponować wieloetapowy plan naprawczy**.

**Jaki jest minimalny wymagany poziom?**

**Przypadek #1: Analizator opinii → poziom 2**

**Uzasadnienie:** Kluczowym i jedynym wymaganiem jest **Structured Output** (wyjście w formacie JSON). System nie potrzebuje zewnętrznej wiedzy (RAG), pamięci ani narzędzi do działania w świecie. To klasyczny przykład inteligentnej automatyzacji, czyli esencja Poziomu 2.

---
**Przypadek #2: Firmowy ekspert HR → poziom 3**

*   **Uzasadnienie:** Sygnałem jest potrzeba sięgnięcia do **zewnętrznej bazy wiedzy** (dokumenty w SharePoint), aby odpowiedzieć na pytanie. To definicja wzorca **RAG (Retrieval-Augmented Generation)**, który jest cechą charakterystyczną Poziomu 3. System nie działa na zewnątrz ani nie potrzebuje pamięci między sesjami.

---
**Przypadek #3: Asystent rezerwacji podróży → poziom 4**

**Uzasadnienie:** Mamy tu dwa jednoznaczne sygnały Poziomu 4:
1.  **Narzędzia (Tools/Actions):** System musi wykonać akcję w świecie zewnętrznym – wywołać API linii lotniczych.
2.  **Pamięć (Memory):** System musi pamiętać preferencje użytkownika z poprzednich interakcji.

Obecność choćby jednej z tych cech kwalifikuje problem na poziom 4.

---

**Przypadek #4: Analityk danych SQL → Poziom 3**

**Uzasadnienie:** To absolutnie kluczowy i celowo podchwytliwy przykład, który testuje głębokie zrozumienie różnicy między Poziomem 3 a 4. Na pierwszy rzut oka, interakcja z bazą danych wygląda jak użycie zewnętrznego narzędzia, co sugerowałoby Poziom 4. Jednak kluczem jest zrozumienie **celu (intencji)** tej interakcji.

Wzorzec RAG (Retrieval-Augmented Generation), będący sercem Poziomu 3, składa się z dwóch kroków:
1.  **Retrieval (Pobranie):** System sięga do zewnętrznego źródła, aby pobrać informacje potrzebne do odpowiedzi.
2.  **Generation (Generowanie):** System wykorzystuje te pobrane informacje (kontekst) do sformułowania finalnej odpowiedzi.

W tym przypadku **bazą wiedzy jest po prostu baza SQL**, a mechanizmem pobierania – dynamicznie wygenerowane zapytanie `SELECT`. Agent nie zmienia stanu świata zewnętrznego (nie modyfikuje danych, nie tworzy nowych rezerwacji). On jedynie **wzbogaca swój kontekst o wiedzę** zamkniętą w bazie danych, aby móc odpowiedzieć na pytanie.

Spójrzmy na bezpośrednie porównanie, które rozwieje wszelkie wątpliwości:

| | **Poziom 3 (RAG z użyciem SQL)** | **Poziom 4 (Akcja z użyciem SQL)** |
| :--- | :--- | :--- |
| **Intencja** | "Dowiedz się czegoś" | "Zmień coś" / "Wykonaj coś" |
| **Przykład** | Agent tłumaczy "pokaż sprzedaż" na `SELECT SUM(sales) FROM...` | Agent tłumaczy "anuluj rezerwację X" na `UPDATE bookings SET status='cancelled'...` |
| **Rola bazy danych** | Źródło wiedzy (pasywne) | System do interakcji (aktywny) |

Dlatego właśnie jest to zaawansowana, ale wciąż implementacja **Poziomu 3**. Pokazuje ona, że "baza wiedzy" w systemach RAG nie musi być zbiorem plików tekstowych czy stron w Confluence. Może to być dowolne, ustrukturyzowane źródło danych, z którego agent potrafi czerpać informacje, aby mądrzej odpowiadać. Dopiero gdyby agent zaczął używać zapytań `INSERT`, `UPDATE` lub `DELETE` do modyfikacji danych, przekroczyłby czerwoną linię i wszedł na Poziom 4.

---
**Przypadek #5: Diagnosta systemów IT → poziom 5**

**Uzasadnienie:** Tutaj nie wystarczy już samo użycie narzędzi (Poziom 4). Kluczowe słowa to "samodzielnie zdecydować", "postawić hipotezę" i "zaproponować plan". Wskazują one na potrzebę **adaptacyjnego planowania i ustrukturyzowanego rozumowania (reasoning)**. Agent musi dynamicznie tworzyć strategię działania w oparciu o napływające dane, a nie tylko wykonywać pojedyncze, z góry określone akcje. To jest esencja Poziomu 5.

## ☝️Ważna uwaga, a co dokładnie oznacza, że agent AI "myśli"?

Spójrzmy na poziom 5: `Strategiczne AI`. Definiujemy go jako system, który potrafi nie tylko używać narzędzi, ale także adaptacyjnie planować ich użycie. Zaczyna "myśleć" w ustrukturyzowany sposób, tworzyć strategie i modyfikować je w trakcie działania.

Zatrzymajmy się na chwilę przy słowie "myśleć", które celowo umieściłem w cudzysłowie. To kluczowe, abyśmy od razu wyjaśnili: **nie chodzi tu o myślenie w ludzkim, świadomym sensie.**

Szczerze mówiąc, precyzyjne zdefiniowanie "ludzkiego myślenia strategicznego" samo w sobie jest niezwykle trudne. A obserwując rzeczywistość, można zaryzykować stwierdzenie, że wielu ludzi na co dzień w ogóle nie myśli. Niestety. Proces ludzkiego myślenia jest skomplikowany, często chaotyczny i nie do końca poznany.

W kontekście AI na poziomie 5, interesuje nas nie tyle naśladowanie ludzkiego *procesu* myślowego (który jest fundamentalnie inny od naszego), co osiągnięcie **mierzalnego wyniku**.

Co więc wkładamy w sens słowa "myśleć" w świecie agentów?
> **"Myślenie" agenta AI to ustrukturyzowany, powtarzalny i weryfikowalny proces podejmowania decyzji w celu osiągnięcia zadanego celu.**

Rozłóżmy to na czynniki pierwsze:
1. **Ustrukturyzowany:** bo opiera się na konkretnych, inżynierskich schematach i architekturach (których będziemy się uczyć), a nie na ulotnej intuicji
2. **Powtarzalny:** dlatego że w tych samych warunkach agent powinien dojść do podobnych, logicznych wniosków.
3. **Weryfikowalny:** w przeciwieństwie do ludzkiego umysłu – możemy prześledzić "tok rozumowania" agenta krok po kroku i zrozumieć, dlaczego podjął taką, a nie inną decyzję.

Im wyższy poziom w naszym spektrum autonomii, tym więcej decyzji system jest w stanie podejmować **samodzielnie** (autonomicznie). Aby to robić skutecznie, potrzebuje właśnie takiego mechanizmu.

Innymi słowy, z inżynierskiego punktu widzenia jest nam wszystko jedno, *jak* Agent AI myśli. Ważne jest tylko to, że **podejmuje słuszne decyzje** i mamy nad tym **kontrolę** (rozumiemy, jak to działa). I tego właśnie będziemy uczyć nasze systemy.

## Struktura kursu i dwa wymiary progresji

Struktura tego programu została zaprojektowana tak, aby zapewnić Ci systematyczny i głęboki rozwój kompetencji. Mamy tu do czynienia z dwoma wymiarami progresji: **pionowym** (per moduł) i **poziomym** (per use case).

### Wymiar Pionowy - budowanie kompetencji krok po kroku

Pionowy podział to klasyczna struktura modułowa, gdzie każdy kolejny moduł wprowadza nową, kluczową zdolność inżynierską:

- **Moduł 1: Domain-Driven Design** – Zaczynamy od fundamentów. Nauczysz się, jak mapować problem biznesowy na architekturę systemu AI i zbudujesz szkielet pierwszego dialogu.
- **Moduł 2: Memory Engineering** – Twój agent AI zyskuje supermoc: pamięć. Zanurzymy się w techniki budowania systemów, które pamiętają kontekst nie tylko w trakcie jednej, ale wielu rozmów.
- **Moduł 3: Actions/Tools Engineering** – Czas wyjść w świat. Wyposażymy agenta w narzędzia, które pozwolą mu wchodzić w interakcje z zewnętrznymi systemami, takimi jak API czy strony internetowe.
- **Moduł 4: Decision Engineering** – To tutaj agent AI zaczyna "myśleć". Wprowadzimy techniki adaptacyjnego planowania (czy też *custom reasoning*), dzięki którym system będzie potrafił samodzielnie tworzyć strategie działania, gdy plan A zawiedzie.
- **Moduł 5: Evaluation Engineering** – Zanim system trafi na produkcję, musisz mieć pewność, że działa. Nauczysz się, jak projektować i implementować zautomatyzowane testy jakości dla złożonych systemów agentowych.



### ☝️Ciekawostka - moduł 5 i dlaczego jakość to fundament, a nie dodatek

Zwróć uwagę, że **moduł 5** nie jest *o* konkretnym poziomie w spektrum AI. Jest on dyscypliną inżynierską, która staje się absolutnie krytyczna, gdy tylko przekroczysz próg poziomu 2-3.

Pomyśl o tym w ten sposób: kontrola jakości istnieje na każdym poziomie, ale jej forma i znaczenie dramatycznie ewoluują. Zobacz, na **poziomie 1**, gdy zadajesz pytanie w ChatGPT, jakość oceniasz `ad-hoc`, własnymi oczami. Czytasz odpowiedź i decydujesz, czy jest dobra. To prosta, manualna weryfikacja.

Jednak im wyżej w spektrum autonomii, tym bardziej ta zasada się zmienia. Na **poziomach 4 i 5** zautomatyzowana kontrola jakości przestaje być `dodatkiem`. Staje się absolutnym, niepodważalnym **obowiązkiem** i punktem wyjścia do projektowania.

Dlaczego? Wyobraź sobie złożony system, w którym zmiana jednego małego komponentu (A - np. modyfikujesz strukturę w pamięci) może nieoczekiwanie zepsuć coś w zupełnie innym miejscu (Z - cały obszar przestał działać). Bez zautomatyzowanej siatki bezpieczeństwa, która sprawdza cały system po każdej zmianie, dalszy rozwój staje się niemożliwy. Utkniesz. Nie będziesz w stanie poruszać się do przodu z pewnością, że nie psujesz czegoś, co już działało.

I tu dochodzimy do kluczowego paradoksu: **im więcej autonomii oddajesz agentowi (i mniej jest `Human-in-the-Loop`), tym więcej zautomatyzowanej kontroli musisz wdrożyć.** To, że człowiek nie weryfikuje każdego kroku, nie oznacza, że chcesz stracić kontrolę. Wręcz przeciwnie – musisz zbudować system, który kontroluje sam siebie.

Potrzebujesz mechanizmów, które błyskawicznie wychwytują anomalie. Właśnie dlatego tworzenie własnych **leaderboardów** do śledzenia wydajności czy projektowanie precyzyjnych, **binarnych testów** sprawdzających kluczowe zachowania jest tak krytycznie ważne.

Dlatego moduł 5 nie jest `o` poziomie 5. Jest on **niezbędnym zestawem narzędzi i myślenia**, abyś mógł w ogóle bezpiecznie i skutecznie budować, rozwijać i utrzymywać systemy na Poziomach 4 i 5.

### Wymiar poziomy

A teraz inna ciekawa część: podział poziomy. Zamiast budować w każdym module małe, niezależne zabawki, będziemy pracować nad tymi samymi, kluczowymi *use case'ami* przez cały kurs.

Oznacza to, że system, który zamodelujesz w module 1, w module 2 zyska pamięć, w module 3 — narzędzia, i tak dalej. Dzięki temu na koniec programu będziesz mieć nie tylko ciekawe portfolio, lecz także wielowarstwowe systemy, a co ważne — zrozumiesz ich ewolucję od deski do deski. Mniej magii, więcej rozumienia. To jest prawdziwa inżynieria, prawda?

### Jakość i bezpieczeństwo

Pracując z LLM-ami, nie możemy też ignorować tematu **bezpieczeństwa**. I tu ważna uwaga: **ten kurs nie jest dedykowanym programem o cyberbezpieczeństwie AI**. To jest temat na osobny, bardzo głęboki kurs. Moim celem nie jest nauczyć Cię wszystkich możliwych technik obrony przed *prompt injection*. Zamiast tego chcę dać Ci coś znacznie cenniejszego: **inżynierski zdrowy rozsądek i umiejętność identyfikowania słabych punktów.** Pokażę Ci, gdzie w architekturze agenta znajdują się jego "pięty achillesowe". Gdy zrozumiesz te fundamentalne słabości, zaczniesz myśleć krytycznie o zarządzaniu ryzykiem. To jest o wiele potężniejsze niż ślepe kopiowanie gotowych szablonów obronnych.

### Narzędzia, Akcje i MCP - patrzymy na Wizję, nie na Hype

W Module 3 będziemy mówić o narzędziach i akcjach. Na pewno usłyszysz o takich terminach jak **MCP ([Model Context Protocol](https://modelcontextprotocol.io/))** od Anthropic. Tak, pojawi się ono u nas. Kluczowe jest zrozumienie naszej filozofii: MCP to sposób realizacji pewnej *wizji*, a nie cel sam w sobie.

Wizja jest prosta: agent musi mieć bezpieczny i ustrukturyzowany sposób na wykonanie czynności w świecie zewnętrznym (np. wyszukanie czegoś w internecie, wysłanie maila, zarezerwowanie terminu). Naszym celem nie jest "nauczyć się MCP", ale **zrozumieć zasady, które pozwalają bezpiecznie zarządzać takimi akcjami**, niezależnie od konkretnej implementacji. W świecie, gdzie co chwilę pojawiają się nowe frameworki, to właśnie zrozumienie fundamentalnych zasad daje Ci przewagę.

Na koniec praktyczna porada. Rekomenduję zaplanować **co najmniej godzinę** na każdą lekcję. Tyle powinno wystarczyć na przerobienie materiału i wykonanie podstawowych zadań.

Jednak w kursie znajdziesz też zadania oznaczone gwiazdką (`*` lub `**`). Są one opcjonalne, ale gorąco zachęcam do ich wykonania. Jeśli chcesz naprawdę głęboko zanurzyć się w temat i zmierzyć z bardziej złożonymi problemami, zaplanuj sobie na takie lekcje raczej **dwie godziny**. Oczywiście, wszystko zależy od Twojego dotychczasowego doświadczenia – jeśli dopiero zaczynasz, niektóre tematy mogą naturalnie zająć trochę więcej czasu.

### Nasza filozofia działania (w tym też przekazanie wiedzy) - inżynieria ponad iluzją

Zanim przejdziemy do konkretów, musimy postawić sprawę jasno. W DataWorkshop skupiamy się na fundamentalnej wiedzy i inżynierskiej praktyce. Uczymy, jak budować systemy, które można wdrożyć na produkcję i za które można wziąć pełną odpowiedzialność. Systemy, które rozumiemy, kontrolujemy i których jakość potrafimy mierzyć.

To jest trudniejsza, ale ostatecznie jedyna wartościowa droga. Jest ona przeciwieństwem tworzenia szybkich, modnych demonstracji, które robią wrażenie przez jeden dzień, a następnego znikają jak motyle.

A teraz najważniejsze: **prawdziwej kompetencji nie da się zdobyć, jedynie oglądając filmy czy przeklikując tutoriale.** Trzeba się zanurzyć w realne problemy, zmierzyć z niejednoznacznością i napisać kod, który działa nie tylko w idealnym scenariuszu. 

Swoją drogą, to jest trudne. A przygotowanie takiego doświadczenia w ramach pięciotygodniowego kursu jest jeszcze trudniejsze.

Jeden z *use case'ów*, które tu zbudujemy, w świecie komercyjnym mógłby być projektem co najmniej na pół roku, a nawet dłużej. Wiem to, bo realizowaliśmy podobne systemy, a jeden z nich zajął naszemu zespołowi prawie rok.

Dlatego to, co tu zrobimy, to mistrzowska synteza. Wziąłem esencję z tych wielomiesięcznych projektów – kluczowe wyzwania architektoniczne, najczęstsze pułapki i najważniejsze decyzje inżynierskie – i skondensowałem je w formie, którą jesteśmy w stanie przerobić.

Celem nie jest wmówić Ci, że w pięć tygodni zbudujesz produkt warty miliony. Celem jest dać Ci **realne zderzenie z rzeczywistością** i pozwolić Ci zdobyć kompetencje, które faktycznie umożliwią Ci budowanie takich produktów w przyszłości.

Nie dam Ci gotowej ryby. Dam Ci wędkę, nauczyć Cię łowić i pokazać najlepsze łowiska. Chcę, aby po tym programie w Twojej głowie kłębiły się pomysły i sprawić, byś na końcu z pełnym przekonaniem powiedział(a): "Tak, wiem, co z tym zrobić dalej!".

A teraz zobacz, co dokładnie złowimy 🐟.

3.  **Normalizacja umiejętności:** Ustalenie, czy wpisany w wymaganiach "Jenkins" oznacza znajomość konkretnego narzędzia, czy ogólną wiedzę o *continuous integration*.

Skupimy się na tym precyzyjnie zdefiniowanym, ale niezwykle wartościowym kroku w procesie HR.
  

### ⚖️ **AI Quality Manager**
Nasz meta-system. Zbudujemy framework do **automatycznej ewaluacji** wszystkich pozostałych agentów. Bo jak inaczej sprawdzisz z całą pewnością, czy Twój agent jest gotowy na produkcję?

### Metoda nauczania - system bliźniaczych par

Zauważasz ten wzorzec: każdy główny *use case* ma swojego "bliźniaka"? To nie jest przypadek. To sedno naszej metodyki nauczania, którą nazywamy **Systemem bliźniaczych par**. Wypracowaliśmy ją w DataWorkshop, aby rozwiązać fundamentalny problem: jak przekazać prawdziwą, praktyczną kompetencję, a nie tylko teoretyczną wiedzę.

Co to jest? Już tłumaczę:

1.  **Ja prowadzę, Ty podążasz:** Najpierw wspólnie pracujemy nad głównym *use case'em* (np. `AI Restaurant Concierge`). Krok po kroku pokazuję Ci, jak analizować problem, projektować architekturę i pisać kod. Tłumaczę nie tylko "co" i "jak" robimy, ale przede wszystkim "**dlaczego**" podejmujemy takie, a nie inne decyzje inżynierskie. To jest etap budowania fundamentu.

2.  **Twoja kolej na działanie:** jak już opanujemy konkretny koncept (będzie to podzielone na mniejsze kroki), otrzymujesz swoje wyzwanie: **bliźniaczy *use case*** (np. `AI Taxi Dispatcher`). I tu zaczyna się "kontrolowalna magia". Ten przypadek może pochodzić z zupełnie innej branży, ale został celowo zaprojektowany tak, aby mieć **ten sam poziom złożoności technicznej.**

Dzięki temu nie możesz po prostu skopiować mojego kodu. Musisz **samodzielnie zaadaptować poznane zasady** do nowego kontekstu. Musisz włączyć swoje myślenie, rozważyć nowe przypadki brzegowe i przełożyć logikę z jednego problemu na drugi. To jest moment, w którym teoria zamienia się w umiejętność.

To właśnie ta metoda jest naszym unikalnym know-how. Jest ona odzwierciedleniem tego, jak my sami zdobywaliśmy kompetencje – nie przez pasywne oglądanie, ale przez aktywne działanie i zderzanie się z realnymi problemami. Różnica między "obejrzałem, jak ktoś to robi" a "sam potrafię to zbudować" jest jak między teorią jazdy a prowadzeniem samochodu w godzinach szczytu. Ten program jest Twoją szkołą jazdy.

## Czym jest Software 3.0? Nasz inżynierski plan

Zanim zakończymy tę lekcję wprowadzającą, musimy wyjaśnić ostatni, kluczowy wątek. Mamy już naszą mapę – 7 poziomów autonomii AI. Teraz czas zdefiniować "pojazd", którym będziemy się po tej mapie poruszać. Czas wyjaśnić, co dokładnie oznaczają terminy z nazwy kursu: **Software 3.0** i **Agentic AI Systems**.

To nie są przypadkowe hasła. To precyzyjne określenia opisujące fundamentalną zmianę paradygmatu w tworzeniu oprogramowania – zmianę, która dzieje się na naszych oczach i w której LLM-y odgrywają główną rolę.

### Ewolucja Paradygmatów: Od 1.0 do 3.0

Świat IT nieustannie ewoluuje. Aby zrozumieć, czym jest Software 3.0, musimy zobaczyć, skąd przyszliśmy.

#### Software 1.0: świat jawnych reguł

To jest klasyczne IT, świat, w którym programiści jawnie piszą kod. Proces jest dobrze znany: analityk lub Product Owner zbiera wymagania biznesowe i tworzy specyfikację. Programista, niczym tłumacz, przekłada język ludzkich potrzeb na precyzyjny język maszynowy – kod. Czasem żartobliwie mówi się, że na wejściu jest kawa, a na wyjściu kod.

Kluczowe w tym paradygmacie jest to, że cała logika systemu jest *jawna i z góry zdefiniowana*. Każda reguła, każdy warunek `if-else`, każda pętla jest wynikiem świadomej **decyzji człowieka**.

#### Software 2.0: Kod "pisany" przez dane

To jest świat klasycznego uczenia maszynowego (Machine Learning). Tutaj następuje zmiana: zamiast pisać skomplikowane algorytmy, dostarczamy dane i pozwalamy, aby algorytm **sam znalazł w nich wzorce**.

Rola człowieka (tę rolę zwykle nazywa się *Data Scientist* lub *Machine Learning Engineer*) zmienia się. Zamiast pisać reguły, przygotowuje dane, definiuje cel (np. prognoza odejścia klienta) i dobiera odpowiednie metryki oraz modele. Wytrenowany model staje się sercem logiki. Oczywiście, cała otoczka dookoła niego – API, bazy danych, interfejsy – to nadal klasyczne Software 1.0. Dlatego te dwa światy od dawna ze sobą współistnieją.

Co ważne, każdy nowy paradygmat umożliwia rozwiązywanie problemów, które wcześniej były niemożliwe. Software 2.0 pozwoliło nam prognozować, rozpoznawać obrazy czy analizować sentyment na skalę niedostępną dla jawnie pisanych reguł.

#### Software 3.0: Orkiestracja inteligencji

I tak dochodzimy do miejsca, w którym jesteśmy dzisiaj. Pojawił się nowy, potężny budulec: **wielkie modele językowe (LLM)**. Software 3.0 to nowy paradygmat inżynierii oprogramowania, w którym **LLM jest centralnym komponentem systemu**.

Jednak Software 3.0 to znacznie więcej niż tylko wywoływanie API modelu. To cała **architektura i zestaw praktyk inżynierskich** dookoła niego. To sztuka stabilizowania i kontrolowania potężnego, ale z natury niepewnego komponentu. Czerpiemy tu z obu poprzednich światów:
*   Z **Software 1.0** bierzemy rygor inżynierski, dbałość o architekturę i testowalność
*   Z **Software 2.0** bierzemy doświadczenie w pracy z niepewnością i podejściem opartym na danych

W skrócie: w Software 3.0 nie programujemy już każdej reguły, ale **projektujemy system, który inteligentnie orkiestruje** działaniem potężnego, pre-trenowanego modelu.

### Jak to wszystko łączy się w całość?

Teraz połączmy kropki z naszym Spektrum Autonomii:
- **Software 3.0** to szeroki paradygmat, który zaczyna się już na **Poziomie 2** (np. użycie LLM do generowania ustrukturyzowanych danych).
- **Agentic AI Systems** to zaawansowane implementacje tego paradygmatu, które odpowiadają **Poziomom 4 i 5**. To systemy, które nie tylko korzystają z LLM, ale posiadają pamięć, narzędzia i zdolność do planowania.

Dlatego ten kurs nazywa się `Software 3.0 – Agentic AI Systems`: uczymy się budować **zaawansowane systemy (Poziomy 4 i 5)** w ramach **nowego paradygmatu (Software 3.0)**.

### Software 3.0 z lotu ptaka 

Przygotowałem dla Ciebie schemat, który będzie naszym planem i mapą drogową przez cały kurs. Tak z lotu ptaka wygląda architektura typowego systemu Software 3.0 (poziom 4-5), który będziemy budować.

![](../images/helicopter_view.jpg)

1.  **Input** dociera do systemu (tekst, głos, obraz)
2.  **Perception System** przetwarza go na zrozumiałą dla maszyny formę (nasze "oczy i uszy")
3.  **Decision Making System** to serce operacji. To tutaj odbywa się całe planowanie i rozumowanie
4.  Ten system ma dostęp do **pamięci (Memory)**...
5.  ...oraz do **bazy wiedzy (Knowledge Base)**...
6.  ...a także do **narzędzi (Tools/Actions)**, by działać w świecie
7.  Na podstawie tych wszystkich elementów generowany jest **Output**

### ☝️ Zwróć uwagę, zdolności a autonomia: gdzie w architekturze ukryty jest poziom?

Ten program skupia się na budowaniu systemów na **poziomach 4 i 5**. Jednak patrząc na nasz schemat architektury, musisz zrozumieć kluczową rzecz: **nie wszystkie jego elementy definiują poziom autonomii**. Niektóre z nich to potężne, ale uniwersalne zdolności, dostępne na każdym poziomie.

Najlepszym przykładem jest **System Percepcyjny** (bloczek #2 na naszym schemacie).

Jeszcze kilka lat temu, zdolność systemu do płynnego rozumienia mowy czy analizy obrazu należała do sfery *science fiction*. Dziś te technologie stały się powszechnie dostępne i wysoce niezawodne. Oznacza to, że system percepcyjny nie jest cechą definiującą poziom zaawansowania. Możemy mieć system na poziomie 1, który przyjmuje polecenia głosowe, i nie czyni go to bardziej autonomicznym. Jest to po prostu inny, bogatszy sposób interakcji. To samo dotyczy wyjścia systemu (Output) – agent na każdym poziomie może odpowiadać tekstem, głosem czy generować obraz.

**Jednak** gdy wchodzimy na poziomy 4 i 5, gdzie nasi agenci mają wchodzić w interakcje z realnym światem (np. dzwonić do restauracji i rozumieć, co mówi się do słuchawki po drugiej stronie), te zdolności percepcyjne stają się **praktyczną koniecznością**.

Dlatego właśnie pracą z głosem zajmiemy się już w **module 1**. Traktujemy to jako fundamentalną, ale pomocniczą zdolność – coś, co musimy mieć, aby móc zbudować naprawdę zaawansowane rozwiązania, ale co samo w sobie nie stanowi o ich inteligencji. Opanujemy ten element, aby później móc w pełni skupić się na prawdziwym rdzeniu tego kursu: **pamięci, narzędziach i adaptacyjnym podejmowaniu decyzji**. To właśnie w tych elementach kryje się prawdziwa autonomia agenta AI.

## Podsumowanie

Kończymy na tym tę pierwszą lekcję, która ma być kompasem dla Ciebie w naszej podróży. Zamiast zwykłej listy punktów, podsumujmy ją inaczej. Jeśli cokolwiek masz zapamiętać z niej, niech to będzie tych siedem potężnych idei, które od teraz stają się Twoim manifestem:

1.  **Inżynieria ponad iluzją.** Odrzucasz pogoń za chwilowymi trendami. Wybierasz trudniejszą, ale jedyną wartościową drogę: budowanie solidnych, mierzalnych systemów, za które bierzesz pełną odpowiedzialność. Jesteś inżynierem, nie iluzjonistą.

2.  **Wiedza z pola bitwy.** Uczysz się nie z teorii, ale z praktyki. Wiedza, którą zdobywasz, to esencja tysięcy godzin testów, porażek i sukcesów z realnych wdrożeń produkcyjnych. To nie jest mapa z książki – to mapa narysowana w ogniu walki.

3.  **Mapa nowego świata.** Spektrum Autonomii AI to nie jest kolejna klasyfikacja. To Twoja osobista mapa do nawigacji po chaotycznym świecie AI. Pozwala Ci precyzyjnie ocenić problem, dobrać narzędzia i, co najważniejsze, pokazać, gdzie Twoje ludzkie kompetencje są niezastąpione.

4.  **Twój cel: poziomy 4 i 5 – architektura myślenia.** Nie uczysz się tylko dodawać funkcje. Uczysz się projektować architekturę "myślenia" dla systemów AI. Twoim celem jest opanowanie budowy agentów AI, którzy posiadają **pamięć, narzędzia i zdolność do adaptacyjnego planowania**.

5.  **Nowy paradygmat: Software 3.0.** Już wiesz, że jesteś świadkiem fundamentalnej zmiany. Nie uczysz się nowego języka programowania – uczysz się zupełnie nowego sposobu myślenia o tworzeniu oprogramowania, w którym Twoim zadaniem jest orkiestracja potężnej, zewnętrznej inteligencji (zwróć uwagę, że tutaj nie chodzi o to, aby pisać kod 10 razy szybciej z Software 1.0).

6.  **Kompetencje przez działanie.** Wiesz już, że prawdziwej umiejętności nie da się zdobyć przez oglądanie. Metoda bliźniaczych par zmusi Cię do działania, do adaptacji, do samodzielnego rozwiązywania problemów. Bo tylko tak wiedza zamienia się w realną kompetencję.

7.  **Twoja rola w przyszłości.** To jest najważniejsza idea. Ten kurs nie uczy Cię, jak przetrwać w nowej erze AI. On uczy Cię, **jak w niej przewodzić**. Dając Ci mapę i narzędzia do budowy zaawansowanych systemów, pokazujemy Ci, gdzie Twoja ludzka kreatywność, strategia i mądrość będą miały największe znaczenie.

## Co dalej?

Zanurzamy się w inżynierię kontekstu (Context Engineering).

W kolejnej lekcji przejdziemy od "dlaczego" do pierwszego, kluczowego "jak". Zanurzymy się w temat, który jest absolutnym fundamentem dla systemów z poziomów 4 i 5: **inżynierię kontekstu**, którą w DataWorkshop nazywamy **structured dynamic prompting**.

Zrozumiesz, dlaczego statyczne, szablonowe prompty to za mało, gdy budujemy inteligentne systemy. Pokażę Ci, jak dynamicznie konstruować kontekst dla modelu, aby był on świadomy historii rozmowy, dostępnych narzędzi i swojej aktualnej misji.

Prawdziwe mistrzostwo osiąga się na przecięciu tych dwóch światów: musisz rozumieć, **dlaczego** inżynieria kontekstu jest sercem systemów poziomu 4, a jednocześnie potrafić ją **sprawnie zaimplementować** na poziomie kodu.

Przygotuj się na pierwszy, kluczowy krok w budowaniu agentów AI (z poziomu 4 i wyżej). Więcej poznasz w kolejnej lekcji!
