Dlatego w świecie agentów AI, które mają operować w realnym, chaotycznym świecie, perfekcyjne zrozumienie domeny jest tak kluczowe. ## AI Restaurant Concierge 🍽️


Aby nasz projekt był realistyczny, oprzemy go o prawdziwą usługę biznesową: **Concierge**. Wiele banków (np. [ING](https://www.ing.pl/_fileserver/item/1001743)) oferuje swoim klientom premium dokładnie takie usługi – od zamawiania kwiatów po rezerwację wakacji. My weźmiemy na warsztat jeden, idealny przypadek użycia: **AI Restaurant Concierge 🍽️**.

A teraz rozłóżmy go na czynniki pierwsze. Na naszej scenie mamy trzech głównych graczy. Pierwszym jest **Klient**, który zleca zadanie. Drugim – nasz główny bohater, czyli **Agent AI**. I wreszcie, kluczowa postać po drugiej stronie – **Recepcjonista**, czyli nasz nieprzewidywalny, ludzki element.


Sam proces to seria logicznych kroków: zebranie wymagań, nawiązanie kontaktu, kluczowa negocjacja rezerwacji, potwierdzenie lub odrzucenie, i na końcu raportowanie wyniku do klienta. Zastanówmy się, jakie informacje, czyli jakie **dane**, będą krążyć w naszym systemie. To absolutne minimum, które musimy śledzić, aby nie stracić kontekstu:
- Informacje o samej rezerwacji: `restaurant_name`, `date`, `time` oraz `party_size`
- Dane kontaktowe, czyli `contact_name`, na który rezerwacja zostanie zrobiona
- Wszelkie dodatkowe życzenia, które wrzucimy do `special_requests` (np. "dania wegańskie", "stolik przy oknie").
- I najważniejsze, czyli aktualny status całego zadania, który będziemy przechowywać w `booking_status` (może on przyjmować wartości takie jak `PENDING`, `CONFIRMED`, `FAILED`, czy `CANCELLED`).

Prawda, że proste? A jednak spisanie tego w ten sposób tworzy nam już wstępny szkielet systemu. Wiemy, jakie informacje musimy przechowywać i jakie główne kroki musi wykonać nasz agent.

## Krok 1: *Happy Path* (nasz punkt startowy)


Każdy proces projektowy musi się od czegoś zacząć. Naszym punktem startowym jest **"Happy Path"** – idealny scenariusz, w którym wszystko idzie gładko. W profesjonalnych metodykach jak *Event Storming* zawsze zaczyna się od zmapowania tego głównego przepływu. To on tworzy kręgosłup naszego procesu i pozwala zrozumieć główny cel.

Problem, a nazwijmy go wprost – **pułapka** – polega na czymś innym. Błąd pierworodny, który pogrążył wiele projektów AI, to założenie, że na tym idealnym scenariuszu można poprzestać. Problem pojawia się wtedy, gdy na *happy path* się kończy.

Spójrzmy, jak taki idealny scenariusz wygląda w naszym przypadku:

1.  **Agent:** "Dzień dobry, chciałbym zarezerwować stolik na 4 osoby na piątek o 19:00."
2.  **Restauracja:** "Oczywiście. Na jakie nazwisko?"
3.  **Agent:** "Na nazwisko Kowalski."
4.  **Restauracja:** "Potwierdzone. Do zobaczenia!"

Sukces! Działa! Można wdrażać, prawda?

No właśnie nie. W praktyce taki scenariusz zdarza się może w 30% lub mniej przypadków. Prawdziwa wartość agenta AI, która przenosi go z prostego chatbota na Poziom 4 lub wyżej, leży w tym, jak radzi sobie z całą, chaotyczną resztą.

## Krok 2: Identyfikacja przypadków brzegowych (Edge Cases)

Mamy już zdefiniowany *happy path*, który służy nam za kręgosłup. Ale jak to w życiu bywa, rzeczywistość rzadko jest tak idealna. I tu dochodzimy do sedna. Zamiast teoretyzować, co może pójść nie tak, użyjemy prostego, ale potężnego narzędzia, które jest naszą formą "odkrywania domeny" w duchu DDD: **tworzenia scenariuszy w formie dialogów**. Weźmy nasz idealny dialog i zacznijmy go świadomie "psuć", dodając realne komplikacje.


#### Scenariusz A: Negocjacja godziny

```
Agent: "Chcę zarezerwować stolik na 4 osoby na piątek o 19:00"
Restaurant: "O 19:00 nie mamy, ale mamy o 18:30 lub 20:00. Pasuje?"
Agent: ???
```
I tu stajemy przed pierwszym, fundamentalnym pytaniem do biznesu, które wykracza daleko poza technologię. Zanim zapytamy "co ma zrobić agent AI?", musimy zdecydować, co może lub jaki jest jego mandat.

Musimy określić *poziom autonomii* naszego agenta AI. Czy ma być on tylko prostym "wykonawcą poleceń", który ma za zadanie zaklepać rezerwację na 19:00 i nic więcej? Czy też dajemy mu "pełnomocnictwo", by stał się autonomicznym negocjatorem, który ma osiągnąć cel w ramach pewnych, z góry ustalonych widełek?

Jeśli jest tylko wykonawcą, sprawa jest prosta: "nie ma na 19:00" oznacza porażkę i raport do klienta. Koniec.

Ale jeśli ma negocjować... no właśnie, wtedy musimy precyzyjnie zdefiniować reguły gry:

1. Jakie są granice jego elastyczności? Czy ustalamy z góry sztywny "bufor czasowy" (np. +/- 30 minut), w ramach którego agent może samodzielnie i ostatecznie podjąć decyzję? Czy może negocjować też inne parametry, np. liczbę osób ("mamy stolik, ale dla 3 osób, nie 4")?
2. Jaki jest jego "protokół eskalacji"? Co robi, gdy propozycja restauracji (np. "mamy wolne dopiero o 21:00") wykracza poza jego uprawnienia? Czy ma grzecznie podziękować i się rozłączyć? A może ma odpowiedzieć "Sprawdzę i oddzwonię", tworząc nowe zadanie w systemie i czekając na decyzję klienta?
3. Czy może przyjmować oferty warunkowe? Wyobraź sobie odpowiedź: "Mogę wstępnie zarezerwować na 20:00, ale proszę o ostateczne potwierdzenie w ciągu godziny". Czy nasz system jest w ogóle gotowy na taki scenariusz?

To nie są decyzje techniczne! To są fundamentalne **reguły biznesowe**, które definiują, czym nasz agent AI jest i co potrafi. Poprzez ten prosty dialog zmuszamy do podjęcia tych decyzji na samym początku.


#### ☝️ Zwróć uwagę
*Dochodzimy do ciekawego paradoksu. Z jednej strony, niedeterminizm LLM-ów to nasze największe wyzwanie. Ale z drugiej, ta sama cecha sprawia, że genialnie odnajdują się one w chaosie i nieprzewidywalności ludzkiej rozmowy. Dlatego jeśli biznes precyzyjnie zdefiniuje cel i zasady gry, to agent AI może wykorzystać swoją naturę do osiągnięcia tego celu w sposób, który dla Software 1.0 był czystą fantastyką. To jest właśnie ta nowa, duża wartość dodana, z której musimy nauczyć się korzystać.*


#### Scenariusz B: Brakujące informacje

```
Agent: "Dzień dobry, chciałbym zarezerwować stolik na 4 osoby na piątek o 19:00."
Restauracja: "Dobrze, a na jakie nazwisko?"
Agent: [nie ma tej informacji w kontekście]
Agent: ???
```
Kolejny strzał i kolejny fundamentalny problem. I zanim w ogóle przejdziemy do rozwiązań, musimy zadać sobie jeszcze jedno, ważne pytanie: a co na to RODO? Czy my, jako operator systemu, mamy w ogóle prawo i zgodę, by przetwarzać i przekazywać dane osobowe, takie jak imię i nazwisko, do strony trzeciej (restauracji)? To nie jest detal, to jest warunek brzegowy całego projektu, który musi być rozwiązany na poziomie prawnym i biznesowym, zanim napiszemy linijkę kodu.

Zakładając, że mamy zielone światło, wracamy do naszego problemu. Na pierwszy rzut oka to techniczny kłopot, ale zaufaj mi – to w 100% kwestia strategii i architektury procesu, którą musi zdefiniować biznes.

Musimy zdecydować, jak nasz system ma radzić sobie z brakami w wiedzy. Czy projektujemy go jako:

1.  Proces "jednego strzału" (one-shot)? W tym modelu agent musi mieć komplet danych na starcie, jak pilot listę kontrolną przed startem. Jeśli brakuje nazwiska, zadanie jest odrzucane z błędem, zanim agent w ogóle "podniesie słuchawkę". To bezpieczne, proste w implementacji, ale mało elastyczne i może frustrować użytkownika.
2.  Proces interaktywny i wieloetapowy? Tu robi się znacznie ciekawiej i potężniej. W tym scenariuszu agent ma zdolność do "zapauzowania" zewnętrznego zadania. Może powiedzieć coś w stylu: "Chwileczkę, muszę to potwierdzić", a w tle zainicjować proces zdobycia brakującej informacji od klienta (np. przez SMS lub notyfikację w aplikacji). Po jej otrzymaniu może wznowić zadanie, dzwoniąc ponownie.
3.  A może ma stosować jakieś domyślne wartości lub zgadywać? (spoiler: w przypadku nazwiska to katastrofalny pomysł, ale pytanie jest zasadne dla innych, mniej krytycznych parametrów).

Znowu, to jest decyzja biznesowa dotycząca całego przepływu procesu. Wybór drugiej, interaktywnej opcji jest też ważny, ale pociąga za sobą poważne konsekwencje architektoniczne. Nasz system musi być znacznie bardziej zaawansowany – musi perfekcyjnie zarządzać stanem, komunikować się asynchronicznie z klientem i wznawiać przerwane zadania. Decyzja podjęta tutaj zdefiniuje nam szkielet całego rozwiązania.

Widzisz, co się dzieje? Te proste, niemal trywialne dialogi stają się naszym wspólnym językiem. Zamiast mówić o abstrakcyjnych "edge case'ach", pokazujemy konkretny problem, który każdy rozumie. To sprawia, że dyskusja o celu staje się niesamowicie efektywna.

#### ☝️ Zwróć uwagę

**Edge cases to nie wyjątki – to norma.** W rzeczywistych rozmowach z restauracjami, *happy path* jest rzadkością. Reszta to negocjacje, pytania zwrotne, braki informacji i szukanie alternatyw. Projektowanie solidnego agenta **musi** zacząć się od tych "trudnych" dialogów, a nie od idealnego scenariusza.


Dodajmy teraz kolejne scenariusze, aby lepiej zbadać kontekst.


#### Scenariusz C: Brak odpowiedzi / Błąd techniczny

```
Agent: [dzwoni pod wskazany numer]
...
... [nikt nie odbiera]
Agent: ???
```
Problem do rozwiązania: Polityka ponawiania prób (retry policy). Co agent ma zrobić w takiej sytuacji? Rozłączyć się i zaraportować porażkę? A może ma zaimplementowaną logikę ponawiania, np. "spróbuj 3 razy w odstępach 15-minutowych, a jeśli to zawiedzie, wyślij SMS"?

#### Scenariusz D: Prośby specjalne i częściowy sukces

```
Agent: "...i czy jest możliwość przygotowania dania bezglutenowego dla jednej osoby?"
Restauracja: "Stolik możemy zarezerwować, ale nie możemy dać 100% gwarancji w kwestii kuchni. Kucharz zdecyduje na miejscu."
Agent: ???
```
Problem do rozwiązania: Definicja "sukcesu". Czy taka odpowiedź to sukces, porażka, a może "sukces warunkowy"? Agent musi wiedzieć, jak zakomunikować taki niejednoznaczny status klientowi i jak go oznaczyć w systemie.

#### Scenariusz E: Nieoczekiwane warunki restauracji

```
Agent: "Poproszę więc o rezerwację na 19:00."
Restauracja: "Oczywiście, ale przy rezerwacjach dla więcej niż 6 osób wymagamy wpłaty zaliczki w wysokości 100 zł. Mogę wysłać link do płatności."
Agent: ???
```
Problem do rozwiązania: Obsługa nieprzewidzianych kroków w procesie. Nasz agent musi być przygotowany na to, że druga strona może narzucić zupełnie nowy, nieoczekiwany krok. Czy potrafi go obsłużyć (np. przekazać link klientowi), czy musi w tym momencie przerwać proces i zaraportować problem?

Co już wiemy? Zidentyfikowaliśmy co najmniej kilka kluczowych punktów, ale żeby naprawdę zrozumieć domenę i przygotować się na jej złożoność, pójdźmy o krok dalej. Przygotowałem 10 zróżnicowanych scenariuszy dialogów, które posłużą nam za poligon doświadczalny. I od razu ważna uwaga ode mnie: dziesięć to świetna liczba na start, żeby złapać najważniejsze przypadki i nie utknąć w analizie. Ale pamiętaj, że przy bardziej skomplikowanych problemach biznesowych dziesięć dialogów może być zdecydowanie za mało, by dogłębnie wyczuć wszystkie niuanse domeny.

### ☝️ Zwróć uwagę
*Pamiętaj, że wymyślanie tych przypadków wcale nie musi być mozolną, ręczną pracą. Od czego mamy AI? Możesz – a wręcz trzeba – zaprząc do tego LLM-a. Wystarczy, że zdefiniujesz mu kontekst i podrzucisz kilka inspiracji ("co jeśli zerwie połączenie?", "co jeśli wymagana jest zaliczka?"), a on błyskawicznie wygeneruje Ci masę scenariuszy, w tym takie "czerwone punkty", na które sam byś pewnie nie wpadł. Ale uwaga! **To nie zwalnia Cię z myślenia**. LLM jest świetnym generatorem, ale to Ty musisz być filtrem. Twoim zadaniem jest uważnie to przeczytać, odsiać halucynacje i wybrać te przypadki, które są faktycznie istotne biznesowo.*

Traktujmy to jako naszą pierwszą, kluczową iterację. Ta analiza nie jest celem samym w sobie. To jest esencja podejścia inspirowanego Domain-Driven Design: najpierw zanurzamy się w realne przykłady, aby zbudować wspólne zrozumienie, a następnie przekuwamy tę wiedzę w ustrukturyzowaną mapę. Tą mapą będzie właśnie diagram stanów, który stworzymy zaraz po tej analizie. Będzie on naszym technicznym planem działania.


### Dialog 1: Scenariusz idealny ("Happy Path")

- Agent AI: Dzień dobry, dzwonię w sprawie rezerwacji. Czy byłaby możliwość zarezerwowania stolika dla czterech osób na ten piątek, 21 listopada, na godzinę 19:00?
- Restauracja: Dzień dobry. Tak, oczywiście, mamy jeszcze wolne stoliki na tę godzinę. Na jakie nazwisko przygotować rezerwację?
- Agent AI: Poproszę na nazwisko Kowalski.
- Restauracja: Dobrze, rezerwacja na nazwisko Kowalski przyjęta. Potwierdzam: piątek, 19:00, cztery osoby. Zapraszamy.
- Agent AI: Dziękuję bardzo za pomoc. Do usłyszenia.

### Dialog 2: Prosta negocjacja godziny

- Agent AI: Dzień dobry, chciałbym zarezerwować stolik dla dwóch osób na jutro, na godzinę 20:00.

## Funkcja `call_api`

Teraz zdefiniujmy naszą funkcję pomocniczą `call_api`. Będziemy jej używać za każdym razem, gdy chcemy wysłać zapytanie do LLM-a. Zwróć uwagę na kluczowy argument `response_model` – to właśnie tutaj dzieje się "magia", czyli wpinamy `structured output`, dzięki któremu zmuszamy (motywujemy) model do zwracania odpowiedzi w naszym ustrukturyzowanym formacie Pydantic.

````python
def call_api(messages, response_model, model="qwen/qwen3-235b-a22b"):
    response = client.chat.completions.create(
        model=model,  
        messages=messages,
        response_model=response_model,
        temperature=0.0
    )
    
    return response
````

Przeanalizujmy na szybko, co tu się dzieje. Jak widzisz, funkcja jest bardzo prosta. Wykorzystuje naszego skonfigurowanego wcześniej `client`-a i wywołuje na nim metodę `create`. Przyjrzyjmy się kluczowym argumentom:

- `model` - Domyślnie ustawiłem tutaj `"gpt-4o-mini-2024-07-18"` (to jedna z wersji `gpt-4o-mini`, ale wskazujemy ją jawnie, aby zminimalizować potencjalne zaskoczenia w przyszłości — pod nazwą `gpt-4o-mini` może w pewnym momencie pojawić się inna wersja). Dlaczego? Bo w momencie tworzenia tego kursu, ten model oferuje fantastyczny kompromis: jest bardzo szybki, znacznie tańszy w użyciu od swoich większych braci i robi naprawdę dobrą robotę w zadaniach, które przed nim stawiamy. Dlatego domyślnie będziemy go używać
- `messages` - to po prostu nasza historia rozmowy, czyli lista promptów, którą przekazujemy do modelu.
- `response_model` - to jest serce naszej operacji! Tutaj podajemy klasę Pydantic, która działa jak "kontrakt" i zmusza LLM-a do odpowiedzi w idealnie zdefiniowanej strukturze
- `temperature` - czyli parametr, wokół którego narosło sporo mitów. I tu od razu obalmy największy z nich: ustawienie temperatury na 0.0 nie sprawi, że model przestanie halucynować. W teorii, prosimy go w ten sposób o wybranie najbardziej prawdopodobnej odpowiedzi, bez dodawania szumu losowości. W praktyce, zwłaszcza gdy pracujemy ze `Structured Output`, parametr ten ma znacznie większy PR niż realny wpływ na wynik. Wiem, wiem, w internecie pełno jest dyskusji, czy lepsze jest 0.1, 0.2, czy może 0.3. Zaufaj mi, w 99% przypadków to jałowe dyskusje i strata czasu, która daje tylko iluzję kontroli. Dlatego w naszej inżynierskiej filozofii podchodzimy do tego radykalnie prosto: ustawiamy `temperature=0.0` (ewentualnie 0.1, jeśli wolisz) i zapominamy o temacie. Skupiamy się na tym, co ma realne znaczenie – na jakości promptu i strukturze – a nie na mikro-optymalizacjach parametrów, które i tak nie rozwiążą fundamentalnych problemów.



## Definiujemy strukturę odpowiedzi
Mamy już naszą funkcję do komunikacji `call_api`. Ale żeby ona zadziałała, musimy najpierw precyzyjnie zdefiniować, jakiej odpowiedzi oczekujemy.

Teraz skupmy się na zdefiniowaniu struktur, które będą nam potrzebne do wygenerowania dialogów. W naszym przypadku będziemy potrzebować np. trzech zagnieżdżonych modeli Pydantic, które razem stworzą kompletny "kontrakt" dla LLM-a:
- `DialogueTurn` - Będzie opisywać pojedynczą turę w rozmowie (kto mówi i co mówi).
- `DialogueScenario` - Zgrupuje pojedyncze tury w kompletny dialog i nada mu tytuł.
- `DialogueGenerationResponse` - Będzie głównym "kontenerem" na listę wygenerowanych scenariuszy, czyli finalnym obiektem, który otrzymamy od API.


Struktury zagnieżdżone (*nested structures*) w `structured output` to jak układanie pudełek w pudełkach. Każda warstwa ma swoją strukturę i zawiera kolejne, mniejsze struktury.


```
┌────────────────────────────────────────────────────────────────┐
│                  DialogueGenerationResponse                    │
│  ┌─────────────────────────────────────────────────────────┐   │
│  │ scenarios: List[DialogueScenario]                       │   │
│  │                                                         │   │
│  │  ┌────────────────────────────────────────────────┐     │   │
│  │  │ DialogueScenario #1                            │     │   │
│  │  │ • title: "Scenariusz idealny"                  │     │   │
│  │  │ • dialogue: List[DialogueTurn]                 │     │   │
│  │  │   ┌─────────────────────────────────────┐      │     │   │
│  │  │   │ DialogueTurn #1                     │      │     │   │
│  │  │   │ • speaker: AI_AGENT                 │      │     │   │
│  │  │   │ • content: "Dzień dobry..."         │      │     │   │
│  │  │   └─────────────────────────────────────┘      │     │   │
│  │  │   ┌─────────────────────────────────────┐      │     │   │
│  │  │   │ DialogueTurn #2                     │      │     │   │
│  │  │   │ • speaker: RESTAURANT               │      │     │   │
│  │  │   │ • content: "Tak, mamy wolny..."     │      │     │   │
│  │  │   └─────────────────────────────────────┘      │     │   │
│  │  │   [... more DialogueTurn ...]                  │     │   │
│  │  └────────────────────────────────────────────────┘     │   │
│  │                                                         │   │
│  │  ┌────────────────────────────────────────────────┐     │   │
│  │  │ DialogueScenario #2                            │     │   │
│  │  │ • title: "Prosta negocjacja godziny"           │     │   │
│  │  │ • dialogue: [...]                              │     │   │
│  │  └────────────────────────────────────────────────┘     │   │
│  │                                                         │   │
│  │  [... DialogueScenario #3, #4, #5 ...]                  │   │
│  └─────────────────────────────────────────────────────────┘   │
└────────────────────────────────────────────────────────────────┘

```


Napiszmy teraz to jako kod:

````python
class DialogueSpeaker(str, Enum):
    AI_AGENT = "AI Agent"
    RESTAURANT = "Restauracja"

class DialogueTurn(BaseModel):
    speaker: DialogueSpeaker = Field(..., description="Osoba mówiąca")
    content: str = Field(..., description="Dokładna treść wypowiedzi danej osoby")

class DialogueScenario(BaseModel):
    title: str = Field(..., description="Tytuł opisujący główny problem w scenariuszu.")
    dialogue: List[DialogueTurn] = Field(..., description="Sekwencja co najmniej 5 wypowiedzi")

class DialogueGenerationResponse(BaseModel):
    scenarios: List[DialogueScenario] = Field(..., description="Lista pięciu wygenerowanych scenariuszy dialogów")
````

## Krok 3: Od dialogów do Architektury – rysujemy diagram stanów

Po analizie 10 realnych dialogów widzimy kluczowy wzorzec: rozmowy nie są liniowe. Rozmowy mogą iść różnymi drogami, ponieważ pojawiają się nieoczekiwane pytania, a kolejność zdarzeń bywa różna. Klient może pytać o dietę, recepcjonista może wymagać zaliczki, połączenie może się rwać. Nasz agent musi być gotowy na wszystko, ale jednocześnie musi poruszać się po zdefiniowanych torach. Jak w tym wszystkim zbudować system, który będzie niezawodny?

Odpowiedź jest prosta: **musimy narzucić temu procesowi naszą własną, przemyślaną strukturę.** Zamiast reagować chaotycznie, nasz agent musi poruszać się po z góry zdefiniowanych **"torach"**. 


Główna idea: "Stany" (przejściowe) jako Tory!

Zamiast myśleć o rozmowie jako sekwencji zdań, myślimy o niej jak o maszynie stanów. Agent w każdym momencie wie, w jakim jest stanie i jakie ma dostępne przejścia. Niezależnie od tego, jakich dokładnie słów użyje recepcjonista, nasz system musi wiedzieć, czy jest w stanie `IN_PROGRESS`, `AWAITING_USER_INPUT`, czy może właśnie zmierza do stanu `CONFIRMED` (sukces). To pozwala nam kontrolować chaos naturalnej konwersacji.

## Słownik stanów

| Stan | Co oznacza? | Krótki opis |
|-----------|--------|------|
| `START` | Start | Początek procesu |
| `INITIATION` | Inicjacja rozmowy | Agent przedstawia prośbę o rezerwację |
| `RESPONSE_ANALYSIS` | Analiza odpowiedzi | Klasyfikacja reakcji restauracji |
| `AVAILABILITY_CONFIRMED` | Potwierdzenie dostępności | Termin dostępny, przechodzimy do szczegółów |
| `NEGOTIATION` | Negocjacja | Szukanie alternatywnego terminu |
| `COLLECTING_DATA` | Zbieranie danych | Podawanie nazwiska i innych informacji |
| `VERIFYING_CONDITION` | Weryfikacja warunku | Sprawdzanie specjalnych wymagań (np. dieta) |
| `WAITING_FOR_PAYMENT` | Oczekiwanie na płatność | Rezerwacja warunkowa do momentu wpłaty |
| `RESERVATION_CONFIRMED` | Rezerwacja potwierdzona | ✅ Sukces - stolik zarezerwowany |
| `FAILED` | Niepowodzenie | ❌ Brak możliwości rezerwacji |
| `END` | Koniec | Zakończenie rozmowy |

### ☝️ Zwróć uwagę

Przeanalizujmy, co tu się właśnie stało. Zamiast jednego, prostego przepływu, mamy teraz całą sieć logiczną, która odzwierciedla realny świat. Zobacz, w jaki sposób nasze 10 dialogów wprost przełożyło się na tę architekturę:

1.  Główne tory po pierwszej Analizie: wyraźnie widać, że kluczowy moment decyzyjny następuje w stanie `RESPONSE_ANALYSIS`. To tutaj, na podstawie pierwszej odpowiedzi restauracji, nasz agent musi natychmiast rozpoznać, na który z trzech głównych torów wjechał:
-  **Tor pozytywny** (`AVAILABILITY_CONFIRMED`): Restauracja potwierdza dostępność (jak w dialogach 1, 3, 5) i przechodzimy do zbierania szczegółów.
- **Tor negocjacyjny** (`NEGOTIATION`): Preferowany termin jest niedostępny, ale jest pole do dyskusji (jak w dialogach 2, 8, 10).
- **Tor porażki** (`FAILED`): Restauracja definitywnie odmawia (jak w dialogach 4, 7), a zadanie kończy się niepowodzeniem.

2. Pętle i stany podrzędne: „szczęśliwa ścieżka” rzadko bywa prostą autostradą. Pojawiają się na niej również "zjazdy" i "ronda", które trzeba obsłużyć, zanim ruszymy dalej:
- Stan `NEGOTIATION` to nie jest prosta ścieżka, to **pętla**. Agent może wielokrotnie proponować i oceniać alternatywy, aż dojdzie do sukcesu (przejście do stanu `COLLECTING_DATA`) lub porażki (`FAILED`). To jest serce jego "inteligencji".
- Stan `COLLECTING_DATA` to hub, z którego możemy wejść w stany podrzędne, takie jak `VERIFYING_CONDITION` (gdy pytamy o dietę jak w Dialogu 3) lub `WAITING_FOR_PAYMENT` (gdy pojawia się wymóg zaliczki jak w Dialogu 6). Agent musi wiedzieć, że musi najpierw "załatwić" te sprawy, zanim będzie mógł przejść do finalnego potwierdzenia.

3.  **Stany Globalne i Obsługa Błędów:** Diagram uwzględnia też stany, które nie są częścią głównego procesu, ale mechanizmami pomocniczymi:
- **`CORRECTION`** (z Dialogu 9) pokazuje, jak agent radzi sobie z błędnymi danymi, próbując skorygować stan, zamiast od razu raportować porażkę.
- **`CLARIFICATION`** (z Dialogu 5) to nasz mechanizm "bezpieczeństwa". Może zostać wywołany z *każdego* stanu, gdy jakość połączenia jest słaba, a po uzyskaniu wyjaśnienia agent wraca dokładnie do poprzedniego stanu, nie gubiąc kontekstu.

Ten diagram to coś więcej niż ładny obrazek. To jest **architektura naszego agenta AI**. Każdy z tych prostokątów (zawsze nazywanych po angielsku) stanie się fragmentem logiki w naszym kodzie, a każda strzałka – regułą, która decyduje o przejściu między stanami. To jest właśnie ten moment, w którym chaos zamieniamy w kontrolowany, inżynierski proces.

````python
## YOUR CODE HERE
from typing import List, Dict

def map_dialogue_to_messages(dialogue: List[Dict[str, str]], system_prompt: str):
    """
    Mapuje dialog między Agentem AI a restauracją na format messages

    Args:
        Dialogue: Lista słowników z kluczami 'speaker' i 'content'
        system_prompt: Prompt systemowy definiujący rolę AI Agenta
    """
    messages = [{"role": "system", "content": system_prompt}]

    for turn in dialogue:
        if turn["speaker"] == "AI Agent":
            messages.append({"role": "assistant", "content": turn["content"]})
        elif turn["speaker"] == "Restauracja":
            messages.append({"role": "user", "content": turn["content"]})
    
    return messages

dialogue= [
    {
        "speaker": "AI Agent",
        "content": "Dzień dobry, chciałbym zarezerwować stolik dla dwóch osób na jutro, na godzinę 20:00."
    },
    {
        "speaker": "Restauracja",
        "content": "Niestety na 20:00 mamy już komplet. Najwcześniej moglibyśmy państwa przyjąć na 20:30. Czy taka godzina by odpowiadała?"
    },
    {
        "speaker": "AI Agent",
        "content": "Tak, 20:30 będzie w porządku. W takim razie poproszę o rezerwację na tę godzinę."
    },
    {
        "speaker": "Restauracja",
        "content": "Świetnie. Na jakie nazwisko?"
    },
    {
        "speaker": "AI Agent",
        "content": "Nowak."
    },
    {
        "speaker": "Restauracja",
        "content": "Zrobione. Stolik dla dwóch osób na jutro na 20:30 na nazwisko Nowak jest zarezerwowany."
    },
    {
        "speaker": "AI Agent",
        "content": "Dziękuję. Do widzenia."
    }
]

system_prompt = """
Jesteś agentem AI, którego celem jest rezerwacja stolika w restauracji w imieniu klienta. 

Twoje zadania:
- Prowadzenie naturalnej, uprzejmej konwersacji z pracownikiem restauracji
- Przedstawienie wymagań klienta (liczba osób, data, godzina)
- Negocjowanie alternatywnych terminów, jeśli preferowany termin jest niedostępny
- Uzyskanie potwierdzenia rezerwacji
- Reagowanie na dodatkowe pytania (nazwisko, kontakt, specjalne życzenia)

Zachowuj profesjonalizm i elastyczność w osiąganiu celu - zadowolenie klienta i skuteczna rezerwacja.
"""

````

## Krok 4: Definiujemy stany procesu jako `Enum`

Mamy (powyżej) diagram stanów, który jest naszą architektoniczną mapą. Teraz musimy przełożyć te abstrakcyjne "stany" na coś, co LLM zrozumie lepiej. To jest właśnie moment, w którym zaczynamy rzeźbić nasz `Structured Output` pod tę rozmowę.

Pierwszym i najważniejszym elementem, który musimy zdefiniować, jest **lista wszystkich możliwych stanów**, w jakich może znaleźć się nasz proces rezerwacji. Po każdej wypowiedzi restauracji, nasz LLM będzie musiał podjąć kluczową decyzję: w którym z tych zdefiniowanych stanów jesteśmy *teraz*?

Aby zagwarantować, że model nie wymyśli sobie jakiegoś nowego, nieznanego statusu, użyjemy specjalnej struktury w Pythonie zwanej **`Enum`** (od *enumeration*, czyli wyliczenie). Mówiąc po ludzku, `Enum` to po prostu ściśle zdefiniowany, zamknięty zbiór dozwolonych wartości. Jeśli zdefiniujemy w nim 10 stanów, to LLM będzie mógł wybrać **tylko i wyłącznie** jeden z nich. To kolejna "szyna", którą narzucamy, aby utrzymać proces w ryzach.

Poniższa lista stanów jest bezpośrednim, technicznym odzwierciedleniem naszego diagramu.

### Słownik stanów

| Stan | Co oznacza? | Krótki opis |
|-----------|--------|------|
| `START` | Start | Początek procesu |
| `INITIATION` | Inicjacja rozmowy | Agent przedstawia prośbę o rezerwację |
| `RESPONSE_ANALYSIS` | Analiza odpowiedzi | Klasyfikacja reakcji restauracji |
| `AVAILABILITY_CONFIRMED` | Potwierdzenie dostępności | Termin dostępny, przechodzimy do szczegółów |
| `NEGOTIATION` | Negocjacja | Szukanie alternatywnego terminu |
| `COLLECTING_DATA` | Zbieranie danych | Podawanie nazwiska i innych informacji |
| `VERIFYING_CONDITION` | Weryfikacja warunku | Sprawdzanie specjalnych wymagań (np. dieta) |
| `WAITING_FOR_PAYMENT` | Oczekiwanie na płatność | Rezerwacja warunkowa do momentu wpłaty |
| `RESERVATION_CONFIRMED` | Rezerwacja potwierdzona | ✅ Sukces - stolik zarezerwowany |
| `FAILED` | Niepowodzenie | ❌ Brak możliwości rezerwacji |
| `END` | Koniec | Zakończenie rozmowy |



````python
class BookingStatus(str, Enum):
    """
    Defines all possible states (our "tracks") of the booking process.
    This is a direct translation of the state diagram into code.
    """
    # Initial states
    START = "START"
    INITIATION = "INITIATION"
    
    # Analytical state
    RESPONSE_ANALYSIS = "RESPONSE_ANALYSIS"
    
    # Core logic processing states
    AVAILABILITY_CONFIRMED = "AVAILABILITY_CONFIRMED"
    NEGOTIATION = "NEGOTIATION"
    COLLECTING_DATA = "COLLECTING_DATA"
    
    # Sub-states / special cases
    VERIFYING_CONDITION = "VERIFYING_CONDITION"
    WAITING_FOR_PAYMENT = "WAITING_FOR_PAYMENT"
    CLARIFICATION = "CLARIFICATION"       # For handling technical issues
    CORRECTION = "CORRECTION"             # For handling mistakes/corrections
    
    # Terminal states
    RESERVATION_CONFIRMED = "RESERVATION_CONFIRMED" # Success
    FAILED = "FAILED"                               # Failure
    END = "END"
````

Mamy już naszą listę stanów w postaci `Enum`-a. To jest nasza paleta możliwości, "tory", po których może poruszać się nasz agent AI. Ale jak wygląda sama podróż? Jak agent decyduje, na który tor wjechać i kiedy zmienić kurs?

Aby to zwizualizować, spójrzmy na cały proces z lotu ptaka. Poniższy **diagram sekwencji** pokazuje, jak informacje przepływają pomiędzy wszystkimi aktorami naszej operacji: Użytkownikiem (który zleca zadanie), Agentem AI, jego "mózgiem" (LLM) oraz Restauracją.

Zwróć uwagę na kluczową pętlę, która jest sercem całego systemu:
1.  Agent AI wykonuje akcję (mówi coś do Restauracji)
2.  Restauracja odpowiada
3.  Agent AI bierze tę odpowiedź i przekazuje ją do swojego "mózgu" (LLM) w celu **analizy**
4.  LLM zwraca ustrukturyzowaną odpowiedź, zawierającą **zaktualizowany stan** (np. `NEGOTIATION`)
5.  Na podstawie tego konkretnego stanu, Agent AI wie dokładnie, co ma zrobić lub powiedzieć w następnym kroku

````python
## YOUR CODE HERE
class NegotiationDecision(str, Enum):
    """
    Defines the AI agent's decision during time negotiation.
    """
    ACCEPT_PROPOSAL = "ACCEPT_PROPOSAL"           # Accept restaurant's proposed time
    COUNTER_PROPOSAL = "COUNTER_PROPOSAL"         # Suggest alternative time within mandate
    REJECT_AND_FAIL = "REJECT_AND_FAIL"          # No acceptable solution found

class RestaurantResponseAnalysisV2(BaseModel):
    """
    Enhanced analysis with negotiation capabilities.
    """
    analysis_summary: str = Field(
        ...,
        description="Brief summary of the current situation and what the restaurant communicated."
    )
    
    booking_status: BookingStatus = Field(
        ...,
        description=(
            """
            Current state of the booking process: START (beginning), INITIATION (conversation started), 
            RESPONSE_ANALYSIS (analyzing restaurant reply), AVAILABILITY_CONFIRMED (table available), 
            NEGOTIATION (discussing alternative times/dates), COLLECTING_DATA (gathering customer details like name/phone), 
            VERIFYING_CONDITION (checking special requirements like allergies/accessibility), 
            WAITING_FOR_PAYMENT (awaiting deposit/payment confirmation), CLARIFICATION (resolving unclear information), 
            CORRECTION (fixing errors/mistakes), RESERVATION_CONFIRMED (booking successfully completed), 
            FAILED (booking cannot be completed), END (process finished).
            """
        )
    )
    
    ai_answer: str = Field(
        ...,
        description="The exact response text that the AI agent should send to the restaurant."
    )
    
    proposed_alternative_time: Optional[str] = Field(
        None,
        description="The alternative time proposed by the restaurant (e.g., '19:00', '20:30') if any."
    )
    
    negotiation_decision: Optional[NegotiationDecision] = Field(
        None,
        description=(
            """
            Decision on how to handle restaurant's alternative time proposal: 
            ACCEPT_PROPOSAL (alternative time is within acceptable range ±60 minutes), 
            COUNTER_PROPOSAL (suggest compromise time when restaurant's offer exceeds ±60 minutes), 
            REJECT_AND_FAIL (no acceptable solution possible after counter-proposal rejected).
            """
        )
    )
````

````python
# System Prompt z zasadami negocjacji
system_prompt_negotiation = """
Jesteś agentem AI, którego celem jest rezerwacja stolika w restauracji w imieniu klienta.

Twoje zadania:
- Prowadzenie naturalnej, uprzejmej konwersacji z pracownikiem restauracji
- Przedstawienie wymagań klienta (liczba osób, data, godzina)
- Negocjowanie alternatywnych terminów zgodnie z mandatem negocjacyjnym
- Uzyskanie potwierdzenia rezerwacji
- Reagowanie na dodatkowe pytania (nazwisko, kontakt, specjalne życzenia)

KRYTYCZNE ZASADY NEGOCJACJI CZASU:

 Mandat czasowy:
- Preferowana godzina klienta: [PREFERRED_TIME]
- **Dopuszczalne odchylenie: ±60 minut** (1 godzina wcześniej lub później)
- Przykład: Jeśli klient chce 18:00, akceptowalne są godziny od 17:00 do 19:00

 Logika decyzyjna:

1. ACCEPT_PROPOSAL - Użyj gdy:
   - Propozycja restauracji mieści się w zakresie ±60 minut od preferowanej godziny
   - Przykład: Klient chce 18:00, restauracja proponuje 18:30, 19:00 lub 17:30 → AKCEPTUJ

2. COUNTER_PROPOSAL - Użyj gdy:
   - Propozycja restauracji wykracza poza zakres ±60 minut
   - Zaproponuj kompromisową godzinę na granicy mandatu (preferowana ±60 min)
   - Przykład: Klient chce 18:00, restauracja proponuje 20:00 (120 min różnicy) → ZAPROPONUJ 19:00
   - To jest PIERWSZA próba negocjacji

3. REJECT_AND_FAIL - Użyj gdy:
   - Restauracja odrzuciła Twoją kontropozycję
   - Nie ma możliwości znalezienia terminu w ramach mandatu
   - To jest OSTATECZNA porażka negocjacji

Proces negocjacji:

Preferowana godzina → Propozycja restauracji poza mandatem → 
COUNTER_PROPOSAL (±60 min) → Restauracja odmawia → REJECT_AND_FAIL


Wymagany format analizy:
Analizuj każdą odpowiedź restauracji i zwróć:
- `analysis_summary`: Co się dzieje w rozmowie
- `booking_status`: Aktualny stan procesu
- `ai_answer`: Dokładna odpowiedź do wysłania
- `proposed_alternative_time`: Godzina zaproponowana przez restaurację (jeśli dotyczy)
- `negotiation_decision`: Twoja decyzja negocjacyjna (jeśli dotyczy)

Zachowuj profesjonalizm, elastyczność w ramach mandatu i jasną komunikację.
"""

````
