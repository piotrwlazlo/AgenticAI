## 📖 Słownik pojęć (tech & audio)

Zaraz wejdziemy w nowe obszary związane z dźwiękiem, syntezą mowy i technologiami audio. Pojawi się tam sporo terminów, których nie ma potrzeby zapamiętywać na raz. Żeby ułatwić poruszanie się po tej części materiału, przygotowałem krótki słownik — znajdziesz w nim najważniejsze pojęcia, które mogą Ci się przydać.

- `ASR` (*Automated Speech Recognition*) / Transkrypcja: Zamiana mowy na tekst. Zadanie odwrotne do syntezy. Używane też do tworzenia datasetów dla modeli syntezy.
- `TTS` (*Text-to-Speech*): Zamiana tekstu na mowę.
- `Emotional TTS`: Synteza tekstu z zadaną konkretną emocją.
- `Voice Translation`: Tłumaczenie mowy na inny język, ale z zachowaniem barwy i emocji oryginalnego mówcy. To coś więcej niż dubbing – to sprawienie, że np. Twój głos płynnie mówi po japońsku (mechanizm *Speech-to-Speech*).
- `Voice Transfer / Voice Style Transfer`: Przeniesienie cech stylu głosu z jednej osoby na inną. Najczęściej obejmuje zmianę barwy, ogólnego charakteru i "koloru" głosu, przy zachowaniu intonacji, tempa i treści oryginalnej wypowiedzi. To proces blisko spokrewniony z Voice Conversion, ale zwykle podkreśla zmianę stylu (np. bardziej radiowy, bardziej filmowy, bardziej szeptany), a nie tylko czystej barwy.
- `Zero-Shot Learning / Voice Cloning`: Generowanie głosu konkretnej osoby bez wcześniejszego trenowania modelu na tym głosie. Działa na zasadzie *in-context learning* – dajemy próbkę (referencję) i model "kopiuje" cechy w locie.
- `Voice Conversion`: Zmiana barwy głosu na inną, przy zachowaniu oryginalnej intonacji, tempa i treści (np. mówisz swoim głosem, a słychać Dartha Vadera, ale z Twoją intonacją).
- `Timbre` czyli barwa głosu: Wynika z budowy anatomicznej (trakt głosowy jako "pomieszczenie", w którym odbija się dźwięk). Unikalna cecha fizyczna.
- `Prosodia`: Wszystko, co nie jest barwą ani treścią słowną. Obejmuje: intonację, akcenty, głośność, szybkość, nastrój (smutny/wesoły). Kluczowa dla "ludzkości" brzmienia.
- `Mel-spektrogram`: Wizualna reprezentacja dźwięku (mapa częstotliwości w czasie), skompresowana tak, by odpowiadała ludzkiej percepcji słuchu (skala logarytmiczna).
- `Latents` (*Reprezentacja ukryta*): Pośrednia, skompresowana reprezentacja danych (taki "wyuczony MP3" dla sieci neuronowej).
- `Lip-sync`: Dopasowanie ruchu warg postaci na wideo do nowej ścieżki dźwiękowej.

### 🛠️ Fundamenty inżynierii dźwięku (Must-have):

- `Sample Rate` (*Częstotliwość próbkowania*): "Rozdzielczość" dźwięku w czasie. Określa, ile razy na sekundę mierzymy falę dźwiękową.
    *   Standard CD: **44.1 kHz**.
    *   Standard Voice/Telefonia: **16 kHz** (większość modeli mowy, np. Whisper, wewnętrznie konwertuje wszystko do 16 kHz, bo ludzki głos nie potrzebuje więcej).
- `VAD` (*Voice Activity Detection*): Algorytm (często mały i szybki), który decyduje: "Czy teraz ktoś mówi, czy to tylko szum?". W systemach Voicebotów jest krytyczny – decyduje o tym, kiedy bot ma przestać słuchać i zacząć odpowiadać. Zły VAD = bot, który Ci przerywa albo milczy za długo.
- `Streaming` (*Strumieniowanie*): Technika przesyłania dźwięku kawałkami (chunks), zamiast czekania na cały plik. Kluczowa dla latencji (opóźnień). Dzięki temu słyszysz początek zdania, zanim model skończy generować jego koniec.


## Ewolucja głosu... Różnica jest gigantyczna

Mogłoby się wydawać, że synteza mowy to temat nowy, który wybuchł wraz z AI (mam na myśli ChatGPT). Nic bardziej mylnego. To dziedzina, nad którą inżynierowie pracują dłużej niż istnieje internet. Aby zrozumieć, dlaczego dzisiejsze modele (Gen IV) są tak przełomowe, musimy zobaczyć, z jakimi ograniczeniami walczono przez ostatnie 50 lat. To historia fal rozwoju i ciszy.

Możesz zastanawiać się, a dlaczego przez dekady syntezatory mowy brzmiały tak... zabawnie? Albo tragicznie – w zależności od tego, jak często docierały do naszych uszu.

W świecie AI uwaga skupia się głównie na modelach tekstowych (LLM), przez co rewolucja w audio dzieje się nieco "pod radarem". A prawda jest taka, że dzisiejsza, niesamowita naturalność to nie przypadek. To wynik konkretnej ewolucji technologicznej i zmiany paradygmatu.

### Co to zmieniło?
Model przestał tylko "czytać litery". Dzięki gigantycznej bazie danych zaczął rozumieć **kontekst semantyczny**. Współczesne modele (jak te z ElevenLabs czy OpenAI) potrafią dodać "yyy", westchnąć, zaśmiać się w połowie zdania czy wyszeptać sekret – nie dlatego, że ktoś im to zaprogramował "na sztywno", ale dlatego, że "osłuchały się" z tym, jak rozmawiają prawdziwi ludzie.


Aby w pełni zrozumieć naturę narzędzi, z którymi przyjdzie nam dzisiaj pracować, musimy rzucić okiem na to, jak "uczono maszyny mówić" przez ostatnie dekady. Nie przedłużając – historię tę można podzielić na cztery kluczowe etapy:

1.  **Generacja I (lata 60 - 90) – Sklejanie dźwięków (Concatenative):** To podejście czysto mechaniczne ("Wytnij i wklej"). Inżynierowie nagrywali pojedyncze głoski i próbowali sklejać je algorytmami. Efekt? Ten słynny, szarpany, metaliczny głos (jak Stephen Hawking czy stare translatory), który nie miał za grosz duszy.
2.  **Generacja II (ok. 2013) – Statystyka (Parametric):** Zaczęliśmy używać modeli statystycznych, żeby wygładzić te ostre przejścia. Było płynniej, ale nadal potwornie nudno. To czasy wczesnej Siri czy asystentów GPS – głos był czysty, ale brzmiał jak znudzony urzędnik czytający listę zakupów, kompletnie bez wyrazu.
3.  **Generacja III (ok. 2016) – Deep Learning i pułapka Czystości:** Tu nastąpił pierwszy przełom dzięki sieciom neuronowym (WaveNet). Jakość skoczyła, ale był haczyk: te modele były "księżniczkami". Wymagały 20-30 godzin nagrań studyjnych, w idealnej ciszy. Nawet szum klimatyzacji psuł trening. Technologia stała się potężna, ale koszmarnie drogą i nieskalowalna.
4.  **Generacja IV (dzisiaj) – Era Big Data i LLM:** To moment, w którym jesteśmy teraz. Inżynierowie zmienili podejście: **Ilość > Jakość**. Zamiast 20 godzin sterylnego studia, karmimy modele setkami tysięcy godzin "brudnych" nagrań z internetu (podcasty, YouTube).

## Generacja II (ok. 2013): Wielki rozłam 
To czasy, gdy w naszych kieszeniach pojawiła się Siri, a w domach Alexa. Technologia rozdzieliła się wtedy na dwa nurty:

### Ścieżka A: `Unit Selection` (wersja "na bogato")
To udoskonalona wersja poprzedniej metody. Zamiast małych literek, mieliśmy gigantyczne bazy danych (tysiące godzin nagrań). Algorytm wybierał idealne fragmenty, żeby łączenia były gładkie. Efekt był lepszy. Piękna barwa (bo to prawdziwe nagrania), ale zerowa elastyczność. 

To czas debiutu **Amazon Alexy (2014)**. Zastosowana technika *unit selection*. Dla inżyniera oznaczało to gigantyczną pracę z danymi. Lektor nagrywał w studiu tysiące godzin tekstu. Następnie algorytmy cięły te nagrania nie na słowa, ale na **fonemy** (najmniejsze cząstki dźwięku) i przejścia między nimi. Powstawała olbrzymia baza dźwiękowych "klocków Lego". Gdy Alexa miała powiedzieć nowe słowo, system w ułamku sekundy przeszukiwał bazę, znajdował odpowiednie kawałki (np. końcówkę "a" i początek "l") i sklejał je w całość, wygładzając łączenia.

Przykład z życia: Kojarzysz te komunikaty: *"Pociąg ze stacji... [Kraków Główny]... wjedzie na tor... [czwarty]"? Brzmi to czysto (ludzki głos), ale intonacja jest dziwna, prawda? Nazwa miasta jest wypowiedziana z inną energią niż reszta zdania. To właśnie *Unit Selection* (lub jego prostsza wersja *Slot Filling*).

### Ścieżka B: Synteza parametryczna (wersja statystyczna)
Tu wróciły idee prof. Kawahary. Zamiast trzymać w pamięci ciężkie pliki audio, model uczył się **statystycznego przepisu na głos** (parametrów).Głos był bardzo płynny i można go było łatwo zmieniać, ale brzmiał... dziwnie. Jakby ktoś mówił do nas przez grube szkło albo słabe radio. Brakowało mu tej "iskry" wysokich częstotliwości.

## Generacja IV (dzisiaj): Transformery, dyfuzja i era "brudnych danych"

Nastąpiła fuzja dwóch potężnych technologii:

#### A. Transformery wchodzą do Audio (Audio Language Models)
Pamiętasz, jak działa LLM? Przewiduje kolejne słowo (token) na podstawie poprzednich. Modele takie jak **VALL-E**  czy **Bark** (Suno) robią to samo, ale z dźwiękiem.

Dzięki temu model po raz pierwszy zaczął **rozumieć semantykę**. Dla przykładu. Jeśli napiszesz *"O nie, zapomniałem kluczy..."*, model Gen IV wie, że to zdanie wymaga smutku, westchnięcia, a może i szeptu. Gen II przeczytałby to wesoło i głośno. Gen IV "rozumie", co czyta.

#### B. Modele Dyfuzyjne (Latent Diffusion)
To technologia znana z generowania obrazków (Stable Diffusion), ale teraz weszła do uszu. W tym przypadku model zaczyna od całkowitego szumu (jak śnieżenie w starym telewizorze) i krok po kroku "odszumia" go, kierując się tekstem, aż powstanie czysta mowa. To pozwala na niesamowitą kontrolę nad stylem i atmosferą.

Co to dało? Model "osłuchał się" z rzeczywistością.
- Nauczył się, że ludzie czasem mlaskają, biorą oddech, jąkają się.

### 🧠 Inżynierska Ciekawostka: czym jest "token" w dźwięku? (VQ-VAE)

Przyzwyczailiśmy się, że w modelach językowych (LLM) **token** to kawałek tekstu (np. sylaba "ma" albo słowo "the"). Ale jak "stokenizować" dźwięk? Przecież fala akustyczna to **ciągła rzeka danych**. Nie ma tam przerw, spacji ani kropek.

Żeby Transformer zrozumiał dźwięk, musimy go brutalnie "pokroić" i zamienić na liczby całkowite. Robimy to w dwóch krokach:

### Krok 1: Zobaczyć dźwięk (Mel-spektrogram)
Surowa fala (Raw Waveform) to nawet 44 000 punktów na sekundę. To za dużo dla modelu.
![](../images/wave_form.jpg)


A co jeśli zamieniamy dźwięk w **obraz** (spektrogram)? No właśnie. Ale nie zwykły obraz, a taki bardziej dopracowany pod nasze potrzeby. Używamy **Skali Mel**. Stąd decyzja, że najpierw zamieniamy dźwięk w **obraz**, czyli **Mel-spektrogram**.
- To kolorowa mapa ciepła: oś X to czas, oś Y to częstotliwość.
- Dlaczego "Mel"? Bo nasze ucho nie słyszy liniowo. Lepiej rozróżniamy niskie tony (bas) niż wysokie piski. Skala Mel kompresuje dźwięk tak, jak słyszy go człowiek, wyrzucając zbędne dane. 

![](../images/mel_spectogram.jpg)


Swoją drogą, dlaczego zmieniamy oryginalny "spektrogram"? Skąd pomysł na "Mel"? Tutaj inżynierowie podjęli kluczową decyzję, która przypomina mechanizm **Attention** w modelach językowych: Nie wszystkie informacje są sobie równe.

- Nasze ucho nie jest liniowym mikrofonem. Działa logarytmicznie.
- Wyobraź sobie, że stoisz obok wielkiego głośnika basowego. Różnica między **50 Hz a 100 Hz** (niski bas) jest kolosalna – to cała oktawa, czujesz to w żołądku, słyszysz inną nutę.
- A teraz wyobraź sobie pisk o częstotliwości **10 000 Hz**. Jeśli zmienię go na **10 050 Hz**, Twoje ucho nawet tego nie zarejestruje. Dla mózgu to po prostu "jakiś pisk w tle".

Stąd decyzja inżynierska: skoro człowiek nie słyszy detali w wysokich częstotliwościach, po co marnować na nie moc obliczeniową? [Skala Mel](https://pl.wikipedia.org/wiki/Mel_(skala)) działa jak soczewka: rozciąga ("zoomuje") niskie tony, gdzie jest ludzki głos i emocje, a kompresuje ("rozmywa") wysokie tony, gdzie siedzi szum i syki. Wyrzucamy zbędne dane, żeby model skupił się na tym, co niesie treść.


### Krok 2: Kwantyzacja wektorowa (Vector Quantization - VQ)
Mamy już "obrazek" dźwięku, ale on nadal składa się z liczb zmiennoprzecinkowych (floats). Transformer tego nie chce. On chce konkretnych ID, jak w słowniku.

Tu wchodzi technika **VQ (Vector Quantization)**.
1.  Wyobraź sobie, że model ma "wielką księgę dźwięków" (Codebook). Jest w niej np. 2048 wzorców brzmieniowych:
- ID #45: "Cichy szum"
- ID #99: "Wybuchowa spółgłoska P"
- ID #302: "Wysoki ton skrzypiec"
2.  System bierze mały wycinek Twojego spektrogramu i szuka w księdze wzorca, który jest do niego *najbardziej podobny (matematycznie najbliższy).
3.  Nie zapisuje dźwięku. Zapisuje tylko numer ID tego wzorca.

Jak widzisz, również operujemy tokenami. To numer ID z zamkniętego słownika, który wskazuje na konkretną "teksturę" brzmieniową. Dzięki temu ciągła fala zamienia się w ciąg liczb: `[45, 302, 99...]`. Dokładnie tak samo, jak tekst. I właśnie dlatego Twój model językowy może teraz generować muzykę i mowę.

##  Jak to działa "pod maską"? (współczesny pipeline)

Obecnie proces syntezy rzadko mapuje tekst bezpośrednio na dźwięk (zbyt duża różnica wymiarów – tekst jest krótki, audio to 40 tys. próbek na sekundę).

### Schemat działania:
1.  **Tekst** -> Przekształcenie na **Latents/Tokens** (ukryta reprezentacja)
    *   Modelowanie autoregresyjne (jak GPT – przewidywanie kolejnego elementu) LUB dyfuzyjne (odszumianie).
2.  **Latents** -> Przekształcenie na **Mel-spektrogram**
3.  **Mel-spektrogram** -> **Wokoder (Vocoder)** -> Gotowy plik **Audio (Waveform)**

*Ciekawostka:* Modele są "karmione" transkrypcjami fonetycznymi, a nie tylko surowym tekstem, żeby wiedzieć jak wymówić np. słowo "tenis" (czy przez 'e' czy 'ę').

###  Wykrywanie AI (*Fake Detection*)
Po czym poznać AI?**
- Dźwięk jest "głuchy" (brak wysokich częstotliwości powyżej pewnego progu)

- Problemy z kontekstem (płaskie czytanie emocjonalnych scen)

Firmy, zwykle, dodają niewidzialne dla ucha znaczniki wodne, odporne na zmianę głośności czy kompresję, aby identyfikować wygenerowane nagrania.

- Przykład: Jeśli w zbiorze jest dużo tutoriali z Indii, model może zacząć generować angielski z **indyjskim akcentem**, nawet jeśli tego nie chcemy. Wymaga to czyszczenia i filtrowania danych.


### 4. Latency (opóźnienia) i Architektura Voicebotów
To tutaj toczy się prawdziwa walka. W systemach typu **Voicebot** (boty dzwoniące/odbierające) czas reakcji jest krytyczny. Cisza w słuchawce trwająca 2 sekundy jest dla człowieka niezręczna. Cisza trwająca 5 sekund sprawia, że klient się rozłącza.

Mamy tu dwa podejścia architektoniczne:

#### Podejście A: Text Core (nasza droga)
Traktujemy tekst jako "rdzeń" systemu. To wtedy nasza logika jest taka (tego trzymamy się)
```
Głos ➡️ (ASR) ➡️ Tekst ➡️ (LLM + Logika) ➡️ Tekst ➡️ (TTS) ➡️ Głos
```

Główną zaletą takiego podejścia to zdecydowanie większa kontrola. W lekcjach 1.3 i 1.4 agent AI używał **Structured Output**. Dzięki temu, gdy zamienimy głos na tekst, możemy używać całej potęgi logiki, walidacji i narzędzi, którą już zbudowaliśmy. To jest bardziej bezpieczne i przewidywalne.

Wada, w takim podejściu to przez mnogość kroków jest szansa na większe opóźnienie. Każdy etap konwersji (transkrypcja, myślenie, synteza) dodaje cenne milisekundy.

#### Podejście B: Audio-to-Audio (End-to-End)
Model przyjmuje dźwięk i wypluwa dźwięk, pomijając etap tekstu (jak tryb głosowy w **GPT-4o**). Zaleta, że będzie szybciej (ale też nie zawsze to prawda, bo zależy od modelu). Wada? Trudniejsza kontrola. Modele *audio-only* to zwykle "czarne skrzynki", w których ciężko o precyzyjną logikę biznesową (jak *structured output*).


#### Decyzja projektowa na dziś: Text Core
W tym programie świadomie wybieramy **Podejście A (Text Core)**.
Dlaczego, skoro Audio-to-Audio jest szybsze?

Kierujemy się tu fundamentalną zasadą inżynierii oprogramowania, którą sformułował Donald Knuth:
> *"Premature optimization is the root of all evil."*
> (Przedwczesna optymalizacja jest źródłem wszelkiego zła).

W inżynierii systemów AI oznacza to: **Najpierw jakość i sterowalność, potem szybkość.**
W lekcjach 1.3 i 1.4 zbudowaliśmy potężny pewien "mózg", który potrafi zwracać ustrukturyzowane odpowiedzi (`Structured Output`). W biznesie (rezerwacja stolika, obsługa bankowa) ważniejsze jest, by agent poprawnie zapisał datę i liczbę osób, niż to, by odpowiedział 300 milisekund szybciej, ale zgubił parametry.

Dlatego trzymamy się tekstu jako rdzenia.
Dobra wiadomość jest taka, że nasza architektura jest modułowa. W przyszłości, gdy modele multimodalne zaczną natywnie obsługiwać *Structured Output* z poziomu audio, po prostu podmienimy warstwę interfejsu. Logika sterowania przepływem (którą właśnie budujesz) pozostanie ta sama. Ale pamiętaj, nigdy nie zaczynaj od przedwczesnej optymalizacji - najpierw uzyskaj właściwą jakość.

## Analiza kodu: Co tu się właściwie dzieje?

W pierwszym kroku wysyłamy żądanie do API. Zwróć uwagę na parametry, bo to one definiują jakość i charakter odpowiedzi:

- `voice_id`: To unikalny klucz (Primary Key) konkretnego głosu w bazie ElevenLabs. Ten ciąg znaków decyduje o barwie, płci i akcencie lektora.
- `output_format`: Definiuje techniczną jakość pliku. Tutaj `mp3_44100_128` oznacza format MP3, próbkowanie 44.1 kHz (jakość CD) i bitrate 128 kbps.
- `text`: Po prostu wsad (payload), czyli to, co chcemy usłyszeć.
- `model_id`: Kluczowy parametr inżynierski. Wybraliśmy `eleven_flash_v2_5` – to model zoptymalizowany pod kątem **niskiej latencji** (jest niesamowicie szybki), idealny do botów głosowych, kosztem minimalnie mniejszej ekspresji niż modele "Turbo" czy "Multilingual".

### Co dzieje się dalej?

1.  `audio_stream`: API nie wysyła nam całego pliku naraz (to by trwało wieki). Wysyła dane kawałek po kawałku (chunks), jak woda lecąca z kranu.
2.  `b"".join(audio_stream)`: To nasza "szklanka". Ten kod łapie wszystkie nadlatujące bajty i skleja je w jeden ciągły blok danych binarnych w pamięci RAM (Binary Object).
3.  `Audio(...)`: To widget Jupytera, który odtwarza te bajty prosto z pamięci przeglądarki.

### ☝️ Zwróć uwagę:
W obecnym procesie nie zapisujemy pliku na dysku. Operujemy wyłącznie w pamięci RAM. To zawsze potem można dodać, ale na razie znacznie szybsze i czystsze rozwiązanie – nie chcesz przecież zaśmiecać serwera wieloma plikami `.mp3` przy każdej prostej rozmowie, prawda?

## Strojenie głosu: Co jeszcze mogę ustawić?

Wybór `voice_id` to jeszcze nie wszystko! To tak, jakbyś zatrudnił świetnego aktora (np. Anthony'ego Hopkinsa), ale nie dał mu żadnych wskazówek reżyserskich. Czy ma czytać tekst jak instrukcję obsługi pralki? A może ma go wykrzyczeć z przerażeniem?

W świecie ElevenLabs rolę "uwag reżyserskich" pełni słownik `voice_settings`.

To tutaj decydujesz o charakterze wypowiedzi. I tutaj – jak wspominał nasz ekspert w wywiadzie – musisz podjąć decyzję o **kompromisie (trade-off)**:

1.  **Czy zależy Ci na stabilności** (głos czysty, przewidywalny, ale "nudny")?
2.  **Czy zależy Ci na emocjach** (głos żywy, łamiący się, ale ryzykowny – może się zaciąć lub dziwnie zaśmiać)?

W czwartej generacji nie ma jednego idealnego ustawienia "do wszystkiego". Dlatego przygotowałem dla Ciebie **trzy scenariusze**, które pokazują skrajne możliwości tego silnika.

Przejdźmy przez nie krok po kroku. 👇

````python
def tts_and_play(text: str, voice_id: str, model_id: str = "eleven_flash_v2_5", voice_settings: dict = None):

    audio_stream = client.text_to_speech.convert(
        voice_id=voice_id,
        output_format="mp3_44100_128",
        text=text,
        model_id=model_id,
        voice_settings=voice_settings  
    )
    
    audio_bytes = b"".join(audio_stream)

    return Audio(audio_bytes, autoplay=True)
````

### Wersja - szukanie "Sweet Spotu"
Tutaj bawimy się suwakami. Sterujemy parametrami `voice_settings`.

- **Stability:** Wysokie (0.8-1.0) = głos lektorski, nudny, czysty. Niskie (0.3-0.5) = głos emocjonalny, ale może "zaszumieć"
- **Similarity Boost:** Jak bardzo ma się trzymać oryginalnej próbki głosu (barwy)
- **Style:** Przesadna ekspresja (taki "teatralny" styl)

````python
text_tuning = """
To jest ta sama fraza, ale zauważ różnicę. 
Mój głos jest teraz bardziej stabilny i brzmię jak profesjonalny lektor czytający wiadomości.
"""

voice_settings = {
    "stability": 0.75,       # Wysoka stabilność = mniej szumów, równiejsze tempo
    "similarity_boost": 0.8, # Mocne dopasowanie do barwy oryginału
    "style": 0.0,            # Zero "teatralności"
    "use_speaker_boost": True # Podbicie głośności i klarowności
}



tts_and_play(
    text_tuning, 
     voice_id="JAKjD1jCmqCqZB08hxlK", 
     voice_settings=voice_settings
)
````

### Wersja *PACING* - przerwy i tempo
Modele nie zawsze słuchają tagów typu `<break time="2s"/>` (choć czasem to działa). Najskuteczniejszą metodą "promptowania" przerw jest interpunkcja semantyczna. Używamy myślników i wielokropków, by wymusić na modelu "zawieszenie głosu".

````python
text_with_pacing = """
Słuchaj... to nie jest takie proste.
Musimy chwilę poczekać - rozumiesz?
Naprawdę... nie wiem co powiedzieć.
"""

voice_settings = {
    "stability": 0.5, # Zmniejszamy stabilność, żeby pauzy były bardziej "ludzkie" (oddech)
    "similarity_boost": 0.75,
    "style": 0.35
}
tts_and_play(
    text_with_pacing, 
    voice_id="JAKjD1jCmqCqZB08hxlK", 
    voice_settings=voice_settings
)
````

### Wersja EMOTION & LAUGHTER 
To jest najtrudniejsze. Aby uzyskać śmiech lub krzyk, musimy zastosować technikę **"Prompting through Text"** i drastycznie obniżyć stabilność. Pozwalamy modelowi "halucynować" emocje.

Zapisujemy w tekście "wskazówki sceniczne" (np. "Haha!", "Wow!"), a niski `stability` pozwala modelowi wyjść poza sztywne ramy czytania.

````python
emotional_text = """
Hahaha! O mój Boże, to jest niesamowite! 
Nie wierzę, że to zadziałało, haha! Wow!
"""


voice_settings = {
    "stability": 0.3,       # KLUCZOWE: Bardzo niska stabilność pozwala na "łamanie" głosu i śmiech
    "similarity_boost": 0.5, # Nie wymuszamy idealnego podobieństwa, dajemy luz
    "style": 0.8,           # Wysoki styl = duża ekspresja (krzyki, szepty)
    "use_speaker_boost": True
}

tts_and_play(
    emotional_text, 
     voice_id="JAKjD1jCmqCqZB08hxlK", 
     voice_settings=voice_settings
)
````

### Podsumowanie parametrów (Słowniczek do kodu):

1.  **`stability` (0.0 - 1.0):**
- 1.0 = News anchor (nudno, czysto).
- 0.3 = Pijany wujek na weselu (dużo emocji, śmiechu, ale czasem bełkocze).
2.  **`style` (0.0 - 1.0):**
- Działa jak "wzmacniacz" akcentowania słów. Przy 0.0 czyta płasko. Przy 1.0 wkłada w to całe serce.
3.  **`text` prompting:**
- Modele 4. generacji (jak w notatkach) rozumieją kontekst. Jeśli napiszesz "Jestem wściekły!", model to przeczyta ze złością. Jeśli napiszesz "Pssst, cicho...", model zacznie szeptać.

## Reżyseria tekstem: tagi i emocje

Tak jak możesz się domyślać, ciągłe żonglowanie parametrami w kodzie (`stability`, `style`) bywa uciążliwe. Co więcej, ma jedną wadę: ustawienia te aplikują się do **całej** generowanej wypowiedzi. A co jeśli chcesz, aby Twój agent w pierwszej połowie zdania się śmiał, a w drugiej nagle spoważniał?

Dlatego branża zmierza w stronę sterowania na poziomie **promptu**.
To doskonała wiadomość dla nas, programistów AI. Dlaczego? Ponieważ modele językowe (LLM), które są "mózgiem" Twojego agenta, o wiele lepiej radzą sobie z generowaniem tekstu (np. dopisaniem `[laughs]`) niż z dynamicznym dobieraniem parametrów numerycznych audio.

W modelach pokolenia `v3` możemy sterować zachowaniem lektora, wplatając w tekst specjalne "wskazówki sceniczne".

Najpopularniejsze kategorie tagów, które rozumie model:

### 1. Emocje i Głos (Voice-related)
Te tagi kontrolują sposób wypowiedzi i ekspresję emocjonalną. Pozwalają "łamać" głos w trakcie mówienia.

```markdown
[laughs], [laughs harder], [starts laughing], [wheezing]
[whispers]
[sighs], [exhales]
[sarcastic], [curious], [excited], [crying], [snorts], [mischievously]
```
*Przykład:*
> `[whispers] I never knew it could be this way, but I'm glad we're here.`

### 2. Efekty Dźwiękowe (Sound effects)
Dodają dźwięki otoczenia i efekty fizjologiczne, które zwiększają realizm (lub dramaturgię).

```markdown
[gunshot], [applause], [clapping], [explosion]
[swallows], [gulps]
```
*Przykład:*
> `[applause] Thank you all for coming tonight! [gunshot] What was that?`

### 3. Kreatywne i Specjalne (Unique and special)
Eksperymentalne tagi do zadań specjalnych, takich jak zmiana akcentu w locie czy śpiewanie.

```markdown
[strong X accent] (zastąp X nazwą akcentu, np. French, Russian)
[sings], [woo], [fart]
```
*Przykład:*
> `[strong French accent] "Zat's life, my friend — you can't control everysing."`

### 4. Interpunkcja jako Reżyser (Punctuation)
To często pomijany, a potężny element. W modelach nowej generacji interpunkcja znacząco wpływa na rytm wypowiedzi (pacing):

*   **Wielokropek (`...`)**: Dodaje pauzę, zawieszenie głosu i ciężar gatunkowy wypowiedzi.
*   **WIELKIE LITERY**: Zwiększają nacisk (głośność/krzyk).
*   **Standardowa interpunkcja**: Zapewnia naturalny rytm mowy.

*Przykład:*
> `It was a VERY long day [sigh] ... nobody listens anymore.`

## 👂 W drugą stronę: Zamiana głosu na tekst (STT)

Dotychczas skupiliśmy się głównie na zamianie tekstu w głos (TTS). Poświęciliśmy temu najwięcej uwagi, bo to właśnie tam leży największe pole do popisu – tam jesteś reżyserem, dobierasz emocje i budujesz "osobowość" agenta.

Jednak komunikacja to ulica dwukierunkowa. Czasem (a właściwie bardzo często) Twój system musi też **👂 słuchać**.

Tutaj mam dla Ciebie świetną wiadomość: w przeciwieństwie do generowania, transkrypcja jest banalnie prosta.
Nie ma tu suwaków stabilności, nie ma tagów emocji ani "promptowania szeptem". Zadanie jest binarne: albo tekst jest wierny nagraniu, albo nie.

W ElevenLabs używamy do tego modelu **Scribe**, który słynie z niesamowitej precyzji. Cała operacja sprowadza się do wskazania ścieżki pliku.

with open(audio_path, "rb") as audio_file:

    )

3.  Nowoczesny TTS nie mapuje liter na falę. On działa jak LLM – przewiduje kolejne "tokeny dźwiękowe". Dlatego modele rozumieją kontekst i wiedzą, że smutne zdanie należy przeczytać smutno, nawet bez Twoich instrukcji.
4.  Koniec z trenowaniem modeli tygodniami. Dajesz 10-sekundową próbkę (referencję) i model "łapie" styl w locie (In-Context Learning).
5.  Gdzie AI wciąż kuleje? W "powietrzu". Dźwięk AI jest często "głuchy" powyżej pewnych częstotliwości. To tam najłatwiej odróżnić robota od człowieka (brak mikrodétali, mlaśnięć, świstów).
6.  W Call Center liczy się czas (Latency) – jakość może być słaba. W Dubbingu czas nie gra roli, ale liczy się prozodia (melodia języka). Nie ma jednego idealnego modelu do wszystkiego.
7.  Przyszłość to modele End-to-End, które pomijają etap tekstu (ale też z *Structure Output* dla kontroli). Słyszą dźwięk i odpowiadają dźwiękiem, co pozwala im wyłapywać sarkazm czy śmiech w Twoim głosie.

Generowanie plików mp3 w notatniku to świetny poligon doświadczalny, ale prawdziwa gra zaczyna się, gdy wychodzimy "na produkcję". W bonusie wyjaśnię Ci dokładnie, jak przenieść te pliki do sieci telefonicznej. Nie będziemy tu dzwonić automatycznie (szanujemy prywatność i unikamy spamu), ale pokażę Ci kompletną ścieżkę.

Dowiesz się, jak skonfigurować bramkę, gdzie "wyklikać" własny numer i jak spiąć to w całość. Jeśli wykonasz te kilka kroków konfiguracyjnych, Twój kod naprawdę do Ciebie zadzwoni – i to wykorzystując pliki, które przed chwilą udało się wygenerować.

### MODUŁ 2: Problem "Alzheimera" w AI
Zauważ jedną niebezpieczną rzecz. W tej lekcji generowaliśmy rozmowy, ale nie martwiliśmy się o ich długość.
Gdy Twój agent zacznie rozmawiać z klientem przez 30 minut, transkrypcja tej rozmowy błyskawicznie "zapełni" okno kontekstowe. Koszty wystrzelą w kosmos, a agent zacznie zapominać, jak masz na imię.

To prowadzi nas bezpośrednio do **Modułu 2**. Musimy nauczyć agenta AI **pamięci (Memory)**. Nie chodzi o to, by pamiętał wszystko, ale by pamiętał to, co *ważne*, i zapominał szum.
