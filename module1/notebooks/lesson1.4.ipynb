### Siła jasno zdefiniowanego celu

I tu chcę zwrócić Twoją szczególną uwagę na słowo **"cel"**, bo ono jest absolutnie kluczowe.

Zauważ, że jak na razie świadomie skupiamy się na rozwiązywaniu problemów, gdzie cel jest wprost wskazany i precyzyjnie zdefiniowany: zarezerwuj stolik na konkretnych warunkach, zamów taksówkę z punktu A do B z maksymalnym czasem oczekiwania. W branży nazywamy to podejściem **"task-oriented"** (zorientowanym na zadanie) lub **"mission-centred"** (skoncentrowanym na misji).

Swoją drogą, to moja gorąca rekomendacja, jeśli zamierzasz gdziekolwiek wdrażać AI: na pierwszy ogień bierz przede wszystkim zadania, które mają jasno określony cel.

Dlaczego to takie ważne? Bo jeśli nie masz celu i agent AI ma "po prostu coś zrobić", wchodzimy na bardzo grząski grunt. Do takich otwartych, niezdefiniowanych zadań należy podchodzić z dużą ostrożnością. Znacznie trudniej jest wtedy zaprojektować system, a jeszcze trudniej ocenić, czy on w ogóle działa poprawnie.

My ten temat będziemy jeszcze wałkować, kiedy w kolejnych modułach zaczniemy budować swoje własne leaderboardy do ewaluacji agentów. Ale już teraz podpowiadam, żeby było to dla Ciebie oczywiste: zaczynamy od solidnych, "task-oriented" fundamentów, bo to najpewniejsza droga do zbudowania systemów, które działają i za które można wziąć odpowiedzialność.

## Analiza nowego problemu biznesowego

Zanim napiszemy linijkę kodu, musimy zrozumieć świat, w którym nasz agent będzie operował. To już znasz, prawda? Nasz nowy cel to **AI Taxi Dispatcher 🚕**.

Na scenie mamy podobnych, ale jednak innych aktorów:
1.  **Klient:** Osoba, która potrzebuje transportu
2.  **Agent AI:** Nasz system, który wykonuje telefon
3.  **Dyspozytor:** Człowiek po drugiej stronie, pracownik korporacji taksówkarskiej

Proces na pozór jest prosty: zadzwoń i zamów taksówkę. Ale diabeł, jak zawsze, tkwi w szczegółach. Jakie informacje są absolutnie kluczowe, aby zadanie mogło się powieść?

- **Adres odbioru i cel podróży:** Gdzie i dokąd jedziemy?
- **Czas odbioru:** Czy taksówka ma być "na teraz", czy na konkretną godzinę?
- **Liczba pasażerów i bagażu:** Standardowy samochód czy może kombi?
- **Wymagania specjalne:** Czy potrzebny jest fotelik dla dziecka? Czy można przewieźć psa? Czy płatność będzie kartą?
- **Dane kontaktowe:** Imię lub numer telefonu dla kierowcy.
- **Status zamówienia:** Czy jest `PENDING`, `CONFIRMED`, `DRIVER_ASSIGNED`, a może `FAILED`?

Zwróć uwagę, jak subtelnie, ale jednak, zmienia się zestaw danych w porównaniu do rezerwacji stolika. To jest właśnie esencja analizy domeny.

## Odkrywanie złożoności przez dialogi

Podobnie jak ostatnio, zacznijmy od "happy path", czyli scenariusza idealnego.

1.  **Agent:** "Dzień dobry, chciałbym zamówić taksówkę na teraz, z adresu Złota 44 na lotnisko Chopina."
2.  **Dyspozytor:** "Oczywiście. Będzie za około 7 minut. Podać dane kierowcy?"
3.  **Agent:** "Tak, poproszę."
4.  **Dyspozytor:** "Skoda Superb, numer rejestracyjny WX 12345, kierowca Piotr. Wszystko się zgadza?"
5.  **Agent:** "Tak, dziękuję bardzo."

Proste, prawda? Ale co, jeśli rzeczywistość okaże się bardziej skomplikowana? Zastanówmy się nad kilkoma scenariuszami, które od razu pokazują nam, gdzie czają się problemy.

#### Scenariusz A: Negocjacja czasu oczekiwania

```
Agent: "Dzień dobry, potrzebuję taksówki na teraz z Dworca Centralnego."
Dyspozytor: "Dzień dobry. Niestety, w tej okolicy mamy teraz duży ruch. Najbliższy samochód może być za 25 minut."
Agent: ???
```
I tu pojawia się kluczowe pytanie biznesowe: jaki jest maksymalny akceptowalny czas oczekiwania? Czy agent ma mandat, by zaakceptować 25 minut? A może jego granicą jest 15 minut i w takiej sytuacji powinien podziękować i spróbować zadzwonić do innej korporacji?

#### Scenariusz B: Obsługa prośby specjalnej

```
Agent: "...i bardzo ważne, będę podróżować z pięcioletnim dzieckiem, więc niezbędny będzie fotelik."
Dyspozytor: "Rozumiem. Muszę sprawdzić, czy któryś z naszych kierowców w okolicy ma na wyposażeniu fotelik. Proszę o chwilę cierpliwości."
Agent: ???
```
To jest odpowiednik pytania o danie bezglutenowe. Proces wchodzi w stan oczekiwania. Agent musi umieć poczekać i odpowiednio zinterpretować odpowiedź zwrotną.

#### Scenariusz C: Pytanie o nietypowe warunki

```
Agent: "Chciałbym zamówić kurs na godzinę 16:00."
Dyspozytor: "Dobrze, ale czy mogę zapytać o formę płatności? Kierowca nie posiada terminala, akceptuje tylko gotówkę."
Agent: ???
```
Nasz system musi być przygotowany na to, że druga strona narzuci nieoczekiwany warunek. Czy agent wie, jaką formę płatności preferuje klient? Czy może to zaakceptować, czy musi przerwać proces i skonsultować się z użytkownikiem?

Widzisz, co się dzieje? Proces myślowy jest identyczny jak w lekcji 1.3, ale konkretne problemy są zupełnie inne. To jest właśnie siła tej metody.

## Wczytujemy biblioteki

````python
from openai import OpenAI
from dotenv import load_dotenv
import instructor
from pydantic import BaseModel, Field
from typing import Optional, List
from enum import Enum
import json
````

## Konfigurujemy klienta i `call_api`

````python
load_dotenv()
````

````python
client = instructor.patch(
    OpenAI(base_url = "https://llm-api.dataworkshop.eu"), 
    mode=instructor.Mode.MD_JSON
)


def call_api(messages, response_model, model="qwen/qwen3-30b-a3b-thinking-2507"):
    response = client.chat.completions.create(
        model=model,  
        messages=messages,
        response_model=response_model,
        temperature=0.0
    )
    
    return response
````

## Adaptacja struktur danych do nowej domeny

Zanim poprosimy LLM-a o wygenerowanie dialogów, musimy wykonać jeden, niezwykle ważny krok. Pamiętasz nasze modele Pydantic z poprzedniej lekcji? Służyły nam jako precyzyjny "kontrakt", który definiował, jakiej odpowiedzi oczekujemy.

Ten kontrakt był jednak skrojony na miarę rozmowy z restauracją. Mieliśmy tam `DialogueSpeaker` z opcjami `AI_AGENT` i `RESTAURANT`. W naszym nowym świecie taksówek, rozmówcą nie jest już pracownik restauracji, a **dyspozytor**.

To pozornie drobna zmiana, ale ma fundamentalne znaczenie. Nasze modele muszą idealnie odzwierciedlać rzeczywistość, w której operuje agent. Dlatego naszym pierwszym zadaniem technicznym będzie aktualizacja tych struktur.

Zwróć uwagę, jak zmieniamy `Enum` `DialogueSpeaker`, aby pasował do nowego kontekstu. Reszta struktur pozostaje bez zmian, co pokazuje, jak dobrze zaprojektowany, generyczny model może być reużywalny.

Pomogę Ci z tym. Poniżej znajdziesz gotowy do użycia blok kodu ze zaktualizowanymi strukturami: `DialogueSpeaker`, `DialogueTurn`, `DialogueScenario` oraz `DialogueGenerationResponse`.

````python
class DialogueSpeaker(str, Enum):
    AI_AGENT = "AI Agent"
    TAXI_DISPATCHER = "Dyspozytor"


class DialogueTurn(BaseModel):
    speaker: DialogueSpeaker = Field(..., description="Osoba mówiąca w danej turze dialogu.")
    content: str = Field(..., description="Dokładna treść wypowiedzi tej osoby.")

class DialogueScenario(BaseModel):
    title: str = Field(..., description="Tytuł, który zwięźle opisuje główny problem poruszany w scenariuszu.")
    dialogue: List[DialogueTurn] = Field(..., description="Sekwencja kolejnych wypowiedzi składających się na pełen dialog.")

class DialogueGenerationResponse(BaseModel):
    scenarios: List[DialogueScenario] = Field(..., description="Lista wygenerowanych scenariuszy dialogów.")
````
