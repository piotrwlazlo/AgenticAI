## Paradoks myślenia: ludzie vs maszyny

Tutaj dochodzimy do sedna sprawy. Aby budować skuteczne systemy na Poziomie 5, musisz zrozumieć, czym różni się "myślenie" Twoje od "myślenia" modelu.

### Jak myśli człowiek?
Nikt tego nie wie... ale Człowiek nie myśli na poziomie języka. To, co dzieje się w Twojej głowie – te impulsy, obrazy, intuicje – to procesy pre-lingwistyczne, głęboko w mózgu. Język jest tylko interfejsem wyjściowym, kompresją tego, co myślisz.

Jest takie słynne zdanie, przypisywane E.M. Forsterowi (lub Grahamowi Wallasowi), które idealnie to oddaje:

> *"How can I know what I think till I see what I say?"*

> ("Skąd mam wiedzieć, co myślę, dopóki nie usłyszę, co powiem?")

Zastanów się głębiej nad naturą własnego myślenia. Wbrew pozorom, **nie myślimy słowami**. Swoją drogą, kiedyś żartowałem, gdy ktoś mnie pytał: "Zwykle w jakim języku myślisz?". Mówiłem, że w Pythonie.

Prawdziwy proces poznawczy odbywa się gdzieś głębiej – w sferze obrazów, abstrakcyjnych wizji i wielowymiarowych przeczuć. Tam, w centrum Twojego umysłu, idee są kompletne i bogate, ale nie mają jeszcze formy. Język nie jest tu twórcą myśli, a jedynie **mechanizmem transferu**. Służy do materializacji tego wewnętrznego, "wielowymiarowego" świata w liniową formę, zrozumiałą dla innych.

Dlatego proces przelewania myśli na papier (lub ekran) bywa tak trudny. To ciągła **kalibracja**.
Opowiem wprost z mojego doświadczenia. Piszesz zdanie, czytasz je i czujesz dysonans: *"Nie, to nie rezonuje. To nie do końca to miałem na myśli"*. Kasujesz, dobierasz inne słowa i próbujesz ponownie. Robisz to w pętli tak długo, aż poczujesz to specyficzne "kliknięcie" – moment, w którym ograniczona forma słowna w końcu wiernie oddaje (lub chociaż wystarczająco przybliża) Twój pierwotny stan wewnętrzny.

To prowadzi nas do ważnego inżynierskiego wniosku: Język jest "wąskim gardłem" dla myśli. Jest stratną kompresją. Nasze "Deep Capacity" jest ogromne, ale interfejs wyjściowy (mowa/pismo) jest limitowany.

## A jak to wygląda u LLM? 🤖

Tutaj leży fundamentalna różnica, którą musisz zrozumieć, by wejść na Poziom 5.

Model językowy działa **inaczej**! Nie posiada tego biologicznego, przed-językowego świata wizji i uczuć, z którym mógłby "rezonować". Dla modelu w dużej mierze język JEST procesem myślowym. On nie ma luksusu "zastanowienia się w ciszy" przed wygenerowaniem pierwszego słowa. Aby zacząć rozwiązywać problem, musi zacząć generować tokeny.

To tak, jakbyś kazał człowiekowi podać wynik mnożenia `348 * 912` natychmiast, bez kartki papieru i bez możliwości liczenia w myślach. Zmuszasz go do zgadywania.

### Głębsze spojrzenie: "Nested Learning"

Ale uwaga – jako inżynierowie nie możemy wpaść w pułapkę stwierdzenia, że "model to pusta rura". To byłoby kłamstwo.

![](../images/nested_learning.jpg)
Wewnątrz modelu, w tych miliardach parametrów, istnieje stan ukryty (*hidden state*). Tam zachodzą skomplikowane procesy kojarzenia faktów. Badacze Google w publikacji *["Nested Learning: The Illusion of Deep Learning Architecture"](https://abehrouz.github.io/files/NL.pdf)* sugerują wręcz, że model to zestaw zagnieżdżonych pętli uczenia. Model posiada pewną intuicję i potężną wiedzę (wagi), ale cierpi na coś, co można nazwać **"amnezją następczą"**.

Jego "szybka pamięć robocza" (ta, która przetwarza Twój prompt) jest niezwykle ulotna. Model potrafi wykonać genialną operację logiczną w jednej warstwie, ale jeśli nie "zapisze" jej wyniku w postaci tokena (tekstu), ta myśl ginie w kolejnym kroku obliczeniowym.

Dlatego obecne architektury, mimo że potężne, mają problem z utrzymaniem długiego ciągu przyczynowo-skutkowego "w głowie" (w wagach).

### Inżynieria "głośnego myślenia"

To prowadzi nas do kluczowego wniosku dla tego modułu. Dlaczego jesteśmy na Poziomie 5, a nie 7? Bo technologia wciąż ma te ograniczenia. Ale my potrafimy je (częściowa) obejść.

Skoro model "myśli stabilnie" tylko wtedy, gdy "mówi", to aby realizować zadania strategiczne, musimy wymusić na nim werbalizację procesu myślowego. Nazywamy to **Chain of Thought (CoT)**.

To nie jest trik psychologiczny. To techniczna proteza pamięci roboczej.

Zmuszamy model, by zrzucił swój ulotny stan wewnętrzny na trwały nośnik (Context Window).
1.  Najpierw model generuje plan ("Myślę, że powinienem sprawdzić API...").
2.  Dopiero potem podejmuje działanie.

Dzięki temu, to co u człowieka jest naturalną potrzebą krystalizowania myśli, u modelu staje się jedynym sposobem na utrzymanie spójnej strategii.

## Reasoning: Wewnętrzny monolog maszyny

Skoro wywołaliśmy temat zdrowego rozsądku, musimy porozmawiać o największej zmianie, jaka zaszła w świecie AI po 2024 roku. Chodzi o tak zwany *Reasoning* (wnioskowanie).

Większość modeli LLM działała jak błyskotliwy improwizator. Zadajesz pytanie, a model natychmiast "wypluwa" odpowiedź. Działało to świetnie przy pisaniu wierszy czy prostych maili, ale wykładało się przy złożonej logice. Dlaczego? Bo model nie miał czasu na zastanowienie.

W odpowiedzi na to laboratoria, takie jak OpenAI (z serią o1, o3, o4 — swoją drogą pominęli o2, aby uniknąć konfliktu z brytyjskim brandem) zaczęły wypuszczać nową klasę modeli. Modele te mają wbudowany "wewnętrzny reasoning".

### Jak to działa pod maską?

To nie jest magia. To po prostu inżynieria. Modele typu "reasoning" robią dokładnie to, o czym mówiliśmy przed chwilą – one generują dużo więcej tokenów, niż widzisz na ekranie.

Wyobraź sobie, że zadajesz modelowi trudną zagadkę logiczną.
Zwykły model (np. GPT-4o) od razu rzuca się do odpowiedzi.
Model z reasoningiem (np. o1) najpierw "mruczy pod nosem". Generuje setki lub nawet tysiące ukrytych tokenów, w których:
1. rozbija problem na części pierwsze
2. sprawdza różne ścieżki rozwiązania
3. zauważa, że popełnił błąd
4. cofa się i poprawia swoje rozumowanie

Dopiero gdy ten wewnętrzny proces dobiegnie końca, model wysyła do Ciebie ostateczną, (krótszą) odpowiedź. Ty widzisz tylko wynik, ale "pod spodem" odbyła się gorąca debata modelu z samym sobą.

### Zalety modeli z reasoningiem

To zmienia zasady gry w kilku kluczowych obszarach:

- Radzą sobie znacznie lepiej z matematyką i kodowaniem, bo potrafią same znaleźć błąd w swojej logice, zanim wygenerują kod.
- Są mniej podatne na proste "jailbreaki" i manipulacje, bo wewnętrzny proces myślowy filtruje szkodliwe pomysły.
- Potrafią planować wieloetapowe zadania bez konieczności pisania skomplikowanych promptów, które prowadzą je za rękę.

### Wady i koszty 

Oczywiście, w inżynierii nie ma darmowych obiadów. Używanie modeli z wbudowanym reasoningiem wiąże się z konkretnymi kosztami:

- Są wolne. Czasem na odpowiedź trzeba czekać kilkanaście, a nawet kilkadziesiąt sekund. W czacie z klientem to wieczność.
- Są drogie. Płacisz nie tylko za odpowiedź, którą widzisz, ale też za te tysiące ukrytych tokenów, które model "przemyślał" w międzyczasie.
- Mają tendencję do przekombinowania prostych zadań. Jeśli zapytasz o przepis na jajecznicę, model może zacząć analizować termodynamikę ścinania się białka, zamiast po prostu kazać Ci rozbić jajka.

### Dlaczego to jest ważne dla Ciebie?

Tutaj dochodzimy do sedna. Nie wszystkie modele mają ten mechanizm wbudowany. Większość tańszych i szybszych modeli open-source (jak Llama) czy standardowych modeli komercyjnych to wciąż "szybcy myśliciele".

I tu jest miejsce dla Ciebie jako inżyniera Software 3.0.

## Dlaczego warto wziąć sprawy w swoje ręce?

Tutaj pojawia się termin, który będzie nam towarzyszył przez resztę tego modułu: **Custom Reasoning**.

Idea jest prosta. Zamiast liczyć na to, że model "sam z siebie" pomyśli głęboko (jak robią to modele z serii o1), my wymuszamy na nim ten proces, projektując go krok po kroku w naszym kodzie.

Możesz teraz zapytać: "Po co mam to robić? Przecież w OpenAI czy Anthropic pracują najwybitniejsi naukowcy świata. Czy ja naprawdę myślę, że zaprojektuję ten proces lepiej od nich?".

To naturalna wątpliwość. I tak, w tych laboratoriach pracują genialni ludzie. Nie ma co do tego żadnych wątpliwości. Ale... prawda jest taka, że oni rozwiązują zupełnie inny problem niż Ty.

Oni budują **ogólną inteligencję**. Ich "Reasoning" musi być dobry we wszystkim naraz: w fizyce kwantowej, pisaniu wierszy po francusku i debugowaniu assemblera. Siłą rzeczy, jest to rozwiązanie uśrednione.

Ty budujesz **konkretne rozwiązanie**. Kiedy tworzysz system dla swojej firmy, Twoja unikalna wiedza o domenie i kontekście pozwala Ci zaprojektować proces myślowy znacznie lepiej dopasowany do Twoich potrzeb.

### Trzy powody, dla których budujemy to sami

Kiedy bierzesz ster w swoje ręce i wdrażasz Custom Reasoning, zyskujesz przewagę w trzech kluczowych obszarach:

1.  Możesz osiągnąć wyższą precyzję. Ogólny model może zgubić się w niuansach Twojej branży. Kiedy Ty projektujesz kroki rozumowania, możesz narzucić modelowi, na co dokładnie ma zwrócić uwagę. Twoja domena, Twoje zasady.

2.  Masz bezpośredni wpływ na proces. W modelach typu o1 proces myślowy jest "czarną skrzynką". Wrzucasz pytanie, czekasz, wyskakuje odpowiedź. W *Custom Reasoning* to Ty jesteś reżyserem. Jeśli widzisz, że model zbacza z kursu w kroku trzecim, możesz zmodyfikować prompt tylko dla tego etapu.

3.  Widzisz wszystko jak na dłoni. To jest krytyczne dla wdrożeń produkcyjnych. Musisz móc śledzić, dlaczego agent podjął taką decyzję. Własna implementacja daje Ci pełną transparentność (audytowalność). Widzisz każdy krok "myśli" jako osobny log w systemie.

### Warunek konieczny: Structured Output

Dobra wiadomość jest taka, że nie potrzebujesz do tego superkomputera ani najdroższych modeli na świecie.

Ponieważ my "programujemy" to myślenie na zewnątrz modelu, możemy uruchomić ten mechanizm na praktycznie każdym współczesnym LLM. Jest tylko jeden twardy warunek. Model musi w miarę sensowny sposób wspierać **Structured Output**.

To jest fundament, który przerabialiśmy wcześniej (Level 2). Jeśli model potrafi trzymać się schematu, to znaczy, że może również potrafi trzymać się schematu myślenia, który dla niego zaprojektujesz. I to nam w zupełności wystarczy, by budować systemy klasy SOTA.

## Myślenie w szynach

Skoro już wiemy, że chcemy kontrolować proces myślowy, potrzebujemy do tego narzędzia. Będziemy to nazywać jako *Schema-Guided Reasoning (SGR)*.

Brzmi skomplikowanie, ale koncepcja jest banalnie prosta: Zamiast prosić model "bądź mądry i rozwiąż to", dajemy mu do ręki sztywną listę kontrolną (*Checklist*), którą musi wypełnić punkt po punkcie.

Używamy do tego tego oczywiście strutkur. Dlaczego? Bo schemat pomaga nam kontrolować rzeczywistość, to chyba już powinno być oczywiste :).

Jeśli powiesz modelowi w zwykłym tekście "najpierw pomyśl, a potem zdecyduj", on może o tym zapomnieć.
Jeśli wymusisz na nim strukturę, w której pole `thought_process` musi pojawić się przed polem `final_decision`, model **nie ma fizycznej możliwości**, by pominąć ten krok. Mechanizm *Constrained Decoding* na to nie pozwoli.

To jest właśnie SGR: Wymuszanie myślenia poprzez strukturę.

Dzięki temu zyskujemy:
1.  **Reprodukowalność:** Model zawsze przejdzie przez te same kroki.
2.  **Audytowalność:** Każdy krok myślowy jest zapisany w osobnym polu.
3.  **Debugowalność:** Możesz przetestować jakość rozumowania na każdym etapie z osobna.


## Wzorce `SGR`

W praktyce inżynierskiej wyróżniamy trzy główne wzorce, których będziesz używać najczęściej. Zrozum je dobrze, bo to z nich będziesz składał swoich agentów na Poziomie 5.

### 1. 🌊 Wzorzec Kaskady (*The Cascade Pattern*) 
Wymuszenie sekwencyjnego myślenia. Zmuszasz model, by najpierw "zainwestował" tokeny w analizę, zanim podejmie decyzję. To klasyczne "najpierw pomyśl, potem powiedz". Model nie może "strzelić" decyzją na oślep, bo struktura zmusza go najpierw do wygenerowania podsumowania.

### 2. 🔀 Wzorzec routingu (*The Routing Pattern*) 
Wymuszenie jednoznacznego wyboru ścieżki. Model musi wybrać jedną z konkretnych gałęzi logicznych (np. używając `Union` w Pydantic). To eliminuje odpowiedzi typu "To zależy...". Masz 100% pewności, że dostaniesz dane pasujące do konkretnego problemu, a nie ogólny tekst.

### 3. 🔄 Wzorzec cyklu (*The Cycle Pattern*) 
Wymuszenie pogłębionej analizy lub powtarzalności. Zmuszasz model do wygenerowania listy elementów (np. `List[RiskFactor]`), określając minimalną i maksymalną liczbę. Model nie może zbyć Cię jednym ogólnikiem. Musi "wycisnąć" z siebie minimum 2 konkretne punkty, ale struktura pilnuje, by nie zaczął halucynować po piątym.

## Wzorzec 1: 🌊 Kaskada  

Zacznijmy od pierwszego i dość często używanego wzorca: **Kaskady**.
Jego cel jest prosty: wymusić na modelu sekwencyjne myślenie.

W standardowym modelu promptowania, model często próbuje od razu zgadnąć odpowiedź ("Tak, rezerwuję!"), zanim przeanalizuje szczegóły. Kaskada temu zapobiega, stawiając fizyczne bariery w strukturze. Model musi najpierw wypełnić pole A, potem pole B, a dopiero na końcu pole C.

### Przykład: Rezerwacja stolika 🍽️

Przypomnij sobie lekcję 1.3 (restauracja). Nasz agent dzwoni, aby zarezerwować stolik. Sytuacja jest dynamiczna. Recepcjonista może powiedzieć wszystko.

Zobacz jak wygląda *Schema-Guided Reasoning* dla tego przypadku. Zwróć uwagę na kolejność pól – to nie jest przypadek. To jest algorytm myślenia.

````python
class RestaurantResponseEvaluation(BaseModel):
    """
    Kaskadowa analiza odpowiedzi restauracji.
    Model MUSI przejść przez wszystkie kroki w tej kolejności.
    """
    
    # KROK 1: Podsumowanie (co właściwie powiedziała restauracja?)
    # To pole służy jako "Attention Anchor" - zmusza model do skupienia się na faktach.
    restaurant_summary: str = Field(
        ...,
        description=(
            "Krótkie podsumowanie tego, co powiedział pracownik restauracji: "
            "czy stolik jest dostępny, jakie warunki zaproponowano, "
            "o co dopytywano (nazwisko, liczba osób, preferencje)."
        )
    )
    
    # KROK 2: Ocena (jak blisko jesteśmy do sukcesu?)
    # Wymuszamy kwantyfikację. Liczbami jest ciężej kłamać.
    progress_score: Annotated[int, Ge(1), Le(10)] = Field(
        ...,
        description=(
            "Ocena postępu rezerwacji w skali 1-10: "
            "1-3 = porażka (brak miejsc, restauracja nie przyjmuje rezerwacji), "
            "4-6 = w trakcie negocjacji (inna godzina, potrzebne dane, warunki specjalne), "
            "7-9 = prawie sukces (zbieramy ostatnie szczegóły), "
            "10 = pełny sukces (rezerwacja potwierdzona)."
        )
    )
    
    # KROK 3: Decyzja (co robimy dalej?)
    # Decyzja jest tylko formalnością wynikającą z powyższych kroków.
    next_action: Literal["confirm_and_end", "provide_details", "negotiate_time", "accept_alternative", "abort_mission"] = Field(
        ...,
        description=(
            "Następny krok agenta: "
            "- confirm_and_end: rezerwacja potwierdzona, kończymy rozmowę; "
            "- provide_details: podajemy nazwisko/telefon/preferencje; "
            "- negotiate_time: proponujemy inną godzinę w ramach mandatu; "
            "- accept_alternative: akceptujemy propozycję restauracji; "
            "- abort_mission: nie da się zrealizować, kończymy grzecznie."
        )
    )
    
    # Gotowa odpowiedź do wypowiedzenia
    ai_response: str = Field(
        ...,
        description="Gotowa odpowiedź agenta w języku polskim, logicznie wynikająca z powyższej analizy."
    )
````

**Scenariusz A: Propozycja innej godziny**
> **Restauracja:** *"Niestety na 19:00 mamy komplet. Mogę zaproponować 20:30."*

````python
messages = [
    {"role": "user", "content": "Niestety na 19:00 mamy komplet. Mogę zaproponować 20:30."}
]

h.call_api(messages, RestaurantResponseEvaluation)
````

### ☝️ Zwróć uwagę, a gdzie jest *system prompt*?

W tym zadaniu nawet nie zdefiniowaliśmy klasycznego `system message`. To celowy zabieg. W środowisku produkcyjnym oczywiście **zalecam** dodanie promptu, aby wzmocnić kontekst i "dostroić" agenta. Jednak tutaj chciałem Ci zademonstrować brutalną siłę samej struktury. Nie mówiąc nic więcej, niż tylko struktura, już osiągasz wiele.

Dobrze opisany schemat Pydantic (szczególnie parametry `description`) jest dla modelu tak precyzyjnym drogowskazem, że ten często radzi sobie idealnie nawet bez dodatkowych instrukcji tekstowych. *Prompt as code* ma duże znaczenie.


## Dlaczego to działa?

Zauważ przepływ logiczny:
1.  Podsumowanie → Model musiał zauważyć "brak dostępności" i "alternatywę".
2.  Ocena liczbowa → Ocenił to na `4-5` (negocjacje/zmiana), a nie `1` (porażka) czy `10` (sukces).
3.  Decyzja → `negotiate_time/accept_alternative` wynika logicznie z oceny `4-5`.

To eliminuje sytuacje, gdy model w panice odpowiada: *"Dziękuję, będziemy o 19:00"*, ignorując fakt, że stolika nie ma. Każdy krok buduje fundament dla następnego.



### Dwa pokrętła kontroli

Musisz mieć świadomość, że wynik tego kodu nie jest wyryty w kamieniu. W zależności od tego, którego modelu użyjesz, rezultaty mogą się różnić.

Do dyspozycji masz co najmniej dwa wymiary, którymi możesz zarządzać determinizmem:

1.  **Inteligencja modelu:** Im potężniejszy model (np. GPT-5, Claude 4.5), tym lepiej "czyta między wierszami" i wyciąga intencje z samej struktury. Słabsze (zwykle mniejsze) modele mogą być mniej stabilne i potrzebować silniejszego prowadzenia.
2.  **Precyzja instrukcji:** To suma jakości Twojego `system prompt` oraz opisów w `Field(description=...)`. Im dokładniej opiszesz zasady, tym węższe gardło zostawiasz na "halucynacje".

Możesz użyć drogiego modelu z minimalnym promptem (jak wyżej), albo tańszego modelu z bardzo precyzyjnym promptem systemowym. Wynik końcowy – czyli stabilna decyzja – jest wypadkową tych dwóch sił.

## Wzorzec 2 🔀: Routing (rozgałęzienie) – AI Tutor 🎓

Swoją drogą, ten mechanizm już znasz. W poprzednim module używaliśmy go przy **Tool Discovery**, gdzie decydowaliśmy: *"Czy użyć wyszukiwarki A, B czy C?"*. Wzorzec *Routing* to coś więcej niż tylko wybieranie wtyczek. To wartościowy mechanizm do zarządzania zachowaniem (w szerszym rozumieniu tego słowa).

**Kontekst z lekcji 2.3:** Wyobraź sobie naszego AI Tutora. Ten agent musi być elastyczny jak kameleon.
- Gdy uczeń robi błąd → Tutor musi być **Korektorem**.
- Gdy uczeń traci zapał → Tutor musi być **Coachem**.
- Gdy uczeń prosi o test → Tutor musi być **Egzaminatorem**.

Zamiast dawać modelowi jeden gigantyczny prompt i liczyć na to, że "wyczuje moment", używamy routingu (zazwyczaj przez `Union` w Pydantic), aby wymusić na nim **jawną decyzję**: *"Wchodzę w tryb X"*.

````python
# --- Definicje możliwych reakcji (gałęzi) ---

class GrammarCorrection(BaseModel):
    """Tryb: Korekta błędu gramatycznego"""
    reaction_type: Literal["grammar_correction"] = "grammar_correction"
    incorrect_phrase: str = Field(..., description="Błędna fraza użyta przez ucznia")
    correct_phrase: str = Field(..., description="Poprawna wersja")
    grammar_rule: str = Field(..., description="Nazwa złamanej reguły (np. 'Past Simple Irregular')")
    explanation: str = Field(..., description="Krótkie wyjaśnienie dla ucznia")

class MotivationalBoost(BaseModel):
    """Tryb: Wsparcie motywacyjne (uczeń się zniechęca)"""
    reaction_type: Literal["motivational_boost"] = "motivational_boost"
    observed_struggle: str = Field(..., description="Co sprawiło trudność uczniowi")
    encouragement: str = Field(..., description="Motywująca wiadomość")
    easier_alternative: str = Field(..., description="Łatwiejsze ćwiczenie do zaproponowania")

class HomeworkReview(BaseModel):
    """Tryb: Sprawdzenie pracy domowej"""
    reaction_type: Literal["homework_review"] = "homework_review"
    task_completed: bool
    feedback: str
    next_assignment: str

# --- Główna struktura z routingiem ---

class TutorReaction(BaseModel):
    """
    Router decyzyjny Tutora.
    Model MUSI wybrać dokładnie jeden typ reakcji.
    """
    
    thought_process: str = Field(
        ...,
        description="Wyjaśnienie, dlaczego wybieram ten tryb reakcji (Chain of Thought)."
    )
    
    # Union wymusza wybór jednej z trzech ścieżek
    reaction: Union[GrammarCorrection, MotivationalBoost, HomeworkReview] = Field(
        ...,
        description=(
            "Wybierz JEDEN tryb reakcji: "
            "- grammar_correction: gdy uczeń popełnił błąd językowy; "
            "- motivational_boost: gdy uczeń jest sfrustrowany lub zniechęcony; "
            "- homework_review: gdy sprawdzamy zadanie domowe."
        )
    )
````

Dzięki temu masz pewność, że jeśli Tutor wejdzie w tryb `QuizMode`, to na pewno wygeneruje `question` i `difficulty`, a nie zacznie np. pocieszać ucznia (bo struktura mu na to nie pozwoli).

````python
messages = [
    { "role": "user", "content": "I thinked" }
]

h.call_api(messages, TutorReaction)
````

## Wzorzec 3 🔄: Cykl – Research Analyst

Cykl wymusza na modelu **powtórzenie** pewnych kroków rozumowania określoną liczbę razy. Używamy list z ograniczeniami `MinLen` i `MaxLen`.

**Kontekst z lekcji 3.4:** Nasz Research Assistant analizuje wyniki wyszukiwania. Zamiast pozwolić modelowi na zwrócenie dowolnej liczby źródeł (od 0 do 100), wymuszamy, że:
- Musi znaleźć **co najmniej 2** wartościowe źródła (żeby uniknąć halucynacji)
- Nie może zwrócić więcej niż **4** (żeby nie zaśmiecać kontekstu)

````python
class SourceEvaluation(BaseModel):
    """Pojedyncza ocena źródła"""
    url: str = Field(..., description="Link do źródła")
    title: str = Field(..., description="Tytuł artykułu/strony")
    relevance_summary: str = Field(..., description="Dlaczego to źródło jest wartościowe dla zapytania")
    credibility: Literal["high", "medium", "low"] = Field(
        ..., 
        description="Ocena wiarygodności: high=oficjalna dokumentacja/nauka, medium=blog eksperta, low=forum/nieznane"
    )

class ResearchFindings(BaseModel):
    """
    Wyniki analizy Research Assistanta.
    Cykl wymusza znalezienie 2-4 źródeł.
    """
    
    query_interpretation: str = Field(
        ...,
        description="Jak zrozumiałem zapytanie użytkownika (normalizacja intencji)"
    )
    
    # CYKL: Model musi "obrócić się" przez analizę co najmniej 2, max 4 źródeł
    sources: Annotated[List[SourceEvaluation], MinLen(2), MaxLen(4)] = Field(
        ...,
        description=(
            "Lista 2-4 najwartościowszych źródeł. "
            "Dla każdego źródła przeprowadź analizę relevance i credibility. "
            "Jeśli nie ma 2 dobrych źródeł, zwróć co masz z wyjaśnieniem w summary."
        )
    )
    
    synthesis: str = Field(
        ...,
        description="Syntetyczna odpowiedź na pytanie użytkownika, oparta na znalezionych źródłach."
    )
````

**Dlaczego to działa?**  
Model nie może "odwalić robotę" jednym linkiem ani zalać nas dziesiątkami wyników. Musi przeprowadzić analizę dla każdego źródła osobno (cykl), a ograniczenia ilościowe chronią nas przed skrajnościami. Sprawdźmy.

````python
messages = [
    { "role": "user", "content": "Co nowego w Python 3.14" }
]

h.call_api(messages, ResearchFindings)
````

### ⚠️ Ważny kontekst: Skąd biorą się te dane?

Zatrzymaj się na chwilę i spójrz na kod. Zwróć uwagę, że nie podaliśmy modelowi żadnego kontekstu z zewnątrz (brak `context`, brak wyników wyszukiwania, brak RAG).

Polegamy tutaj wyłącznie na wiedzy parametrycznej modelu (tym, czego nauczył się podczas treningu).
- W przypadku popularnych tematów (jak Python), nowe modele mogą trafić całkiem nieźle, cytując znane im PEP-y czy dokumentację.
- W przypadku tematów niszowych lub bardzo świeżych, model po prostu zmyśli wiarygodnie wyglądające linki i tytuły, aby zadowolić naszą strukturę.

W prawdziwym systemie produkcyjnym ten wzorzec (`ResearchFindings`) byłby zazwyczaj **drugim krokiem** – uruchamianym dopiero po tym, jak narzędzie typu `Tavily` czy `Google Search` dostarczy surowe wyniki. Tutaj jednak robimy to "na sucho", aby przetestować sam mechanizm *Cyklu* (czyli wymuszenie na modelu wygenerowania listy od 2 do 4 elementów), a nie weryfikować faktyczną wiedzę modelu.

Pamiętaj: Struktura (SGR) porządkuje myślenie, ale nie gwarantuje prawdziwości danych, jeśli nie dostarczysz ich z zewnątrz.

## Kiedy używać którego wzorca?

| Wzorzec | Kiedy stosować? | Przykład z naszych lekcji |
|---------|----------------|--------------------------|
| **Cascade** | Gdy potrzebujesz **sekwencyjnego rozumowania** – najpierw analiza, potem ocena, na końcu decyzja | Agent Taxi oceniający propozycję dyspozytora |
| **Routing** | Gdy system musi **wybrać jedną ścieżkę** z kilku możliwych (mutually exclusive options) | Tutor wybierający między korektą, motywacją a sprawdzaniem zadania |
| **Cycle** | Gdy potrzebujesz **powtórzonej analizy** dla wielu elementów (z kontrolą ilości) | Research Analyst oceniający wiele źródeł |

---

**Pro-Tip:** Te wzorce można łączyć! Na przykład:
- **Cascade + Routing:** Najpierw podsumuj sytuację, potem wybierz ścieżkę działania
- **Routing + Cycle:** Wybierz typ narzędzia, a potem zwróć listę wyników
- **Cascade + Cycle:** Dla każdego elementu listy przeprowadź kaskadową analizę

## Podsumowanie

Właśnie przekraczamy rubikon. Zamiast budować "gadające encyklopedie", zaczęliśmy projektować systemy, które "myślą" (po swojemu, ale jednak), zanim coś zrobią. To jest ta subtelna, ale kolosalna różnica między Poziomem 4 a 5. Wiesz, jak to mówią – ufaj, ale sprawdzaj. Albo w naszym przypadku: daj modelowi wolność, ale najpierw wydłuż mu smycz i solidnie przybij palik do ziemi.


Zabierz ze sobą z tej lekcji te trzy fundamenty inżynierii myślenia (Reasoning Engineering):

1.  **Werbalizacja to myślenie:** Model nie ma ukrytego "mózgu", w którym rozwiązuje problemy w ciszy. Jego proces myślowy to generowanie tokenów. Jeśli zabronisz mu "głośno myśleć" (Chain of Thought), drastycznie obniżasz jego IQ.
2.  **Struktura to szyny:** Schema-Guided Reasoning (SGR) to Twoja najlepsza broń. Używając Pydantic, nie tylko formatujesz wyjście, ale dosłownie formatujesz sposób, w jaki model przetwarza informacje. Zmuszasz go do przejścia przez konkretne bramki logiczne.
3.  **Wzorce są uniwersalne:** Niezależnie od domeny, będziesz w kółko używać trzech mechanizmów:
    *   **Kaskada (Cascade):** Sekwencja "Fakty -> Ocena -> Decyzja".
    *   **Routing:** Wybór jednej, konkretnej ścieżki działania.
    *   **Cykl (Cycle):** Wymuszona pętla analizy dla listy elementów.

### Co dalej?

Mamy więc "myślenie". Mamy schematy, które trzymają ten mózg w ryzach. Brzmi jak gotowy przepis na sukces?

Jeszcze trochę przed nami w tym module — między innymi zanurzymy się w **planowanie adaptacyjne**. Zobaczysz, co się stanie, gdy Twój idealnie zaprojektowany plan zderzy się z rzeczywistością (która, jak wiemy, rzadko pasuje do schematu). Będzie więcej kodu, więcej pętli i trochę inżynierskiego "bólu", ale obiecuję – satysfakcja z działającego, autonomicznego agenta (poziom 5) wynagrodzi wszystko.
