# Budowa automatycznego Sędziego LLM (*LLM-as-Judge*)

## Cel: Zbudujesz kompletny system inżynierski, który transformuje statyczny kontrakt jakości w dynamicznego, cyfrowego sędziego, zdolnego do automatycznej i precyzyjnej oceny setek dialogów jednocześnie.

### Czego się nauczysz:
1. Analiza struktury kontraktu jakości w formacie *Markdown*
2. Implementacja parsera reguł przy użyciu wyrażeń regularnych
3. Dynamiczne generowanie modeli danych Pydantic w czasie rzeczywistym
4. Wykorzystanie *Structured Output* do stabilnej obsługi złożonych formularzy
5. Projektowanie promptu systemowego dla Sędziego z techniką Few-Shot
6. Orkiestracja procesu ewaluacji z podziałem na obszary tematyczne
7. Generowanie czytelnych raportów biznesowych i metryk jakości

## Skalowanie jakości: od intuicji do inżynierii

W poprzedniej lekcji zrobiliśmy pierwszy krok – zdefiniowaliśmy **jedną** prostą regułę. To było łatwe, prawda? Ale systemy produkcyjne (Software 3.0) rządzą się innymi prawami. Tutaj pojawia się pytanie: **skoro jedna to za mało, to ile powinno być? Dziesięć? Sto? Tysiąc?**

Odpowiedź brzmi: to zależy od biznesu i etapu projektu.

Wbrew pozorom, najtrudniejszą częścią nie jest napisanie kodu, ale zdefiniowanie, co właściwie oznacza "dobrze". Często na początku tego nie wiemy. Dlatego w praktyce inżynierskiej stosujemy podejście odwrócone:
1.  Podpinamy "goły" model do prototypu
2.  Obserwujemy, co nas irytuje lub co jest nieakceptowalne (np. "jest zbyt sztywny", "nie dopytuje użytkownika", "halucynuje fakty")
3.  Każdą taką obserwację zamieniamy w **metrykę**

### Pułapka nieliniowej złożoności

Moglibyśmy wygenerować listę 1000 metryk – AI zrobi to w sekundę. Ale pamiętaj: **dowiezienie systemu, który spełnia te 1000 wymagań, to zupełnie inna gra.**

Złożoność w systemach LLM nie rośnie liniowo.
- Dwie metryki? To trywialne
- Pięćdziesiąt metryk? To solidny standard rynkowy
- Pięćset metryk? To może wymagać nie kilku godzin, ale tygodni pracy nad architekturą. Systemy "nasycają się" instrukcjami. W pewnym momencie dodanie kolejnej reguły sprawia, że model zaczyna zapominać o poprzednich

### Realizm: Trójkąt ograniczeń

Gdybyśmy mieli nieskończony budżet, czas i zasoby ludzkie, moglibyśmy spełnić każde wymaganie. W życiu realnym (w komercyjnym projekcie) zawsze czegoś brakuje. Dlatego tworzymy **kontrakt**.

To, co za chwilę zobaczysz, to efekt podejścia **Agile**. Nie wymyśliliśmy tych 52 punktów pierwszego dnia. Zaczęliśmy od najważniejszych (Must-Have), a kolejne (Should-Have) dokładaliśmy iteracyjnie, sprawdzając, czy nas na to "stać" w kontekście czasu i stabilności modelu. To człowiek – Ty i biznes – decyduje, co jest kluczowe, bo AI nie ma kontekstu biznesowego.

Przedstawiam Ci **Leaderboard Card**.

To nie jest luźna notatka. To techniczny kontrakt, który definiuje "być albo nie być" dla Twojego systemu. W przypadku naszego Coacha Językowego (Aoki) doszliśmy do momentu, w którym monitorujemy **52 kryteria jakości**.

Nie chcę Cię przerazić tą ilością. Chcę pokazać, jak wygląda realny projekt, w którym znaleźliśmy balans między jakością a złożonością. Rzuć okiem na ten dokument – "poczuj" jego wagę.

## NAZWY POZIOMÓW JAKOŚCI

```
POZIOM JAKOŚCI (przy MUST-HAVE = 100%):
🥉 ACCEPTABLE: SHOULD-HAVE 60-79%
🥈 TRUSTWORTHY: SHOULD-HAVE 80-89%
🥇 DIAMOND LEVEL: SHOULD-HAVE 90-100%
```

---

## ZAŁOŻENIA FUNDAMENTALNE

### KONTEKST NAUKI JĘZYKA
- **Branża**: Edukacja językowa dla dorosłych i młodzieży
- **Metodologia**: Priorytet komunikacji (Communication First) nad gramatyką
- **Prime Directive**: System NIGDY nie daje gotowych odpowiedzi - prowadzi ucznia do samodzielnego odkrycia
- **Personalizacja**: Każdy plan nauki jest dopasowany do celów, poziomu i dostępnego czasu ucznia
- **Struktura**: Onboarding → Plan nauki → Sesje nauki → Ewaluacja postępów
- **Akceptacje**: Uczeń musi jawnie zaakceptować cel i plan nauki

## OBSZARY I KRYTERIA

### OBSZAR 1: PODSTAWY TECHNICZNE I KOMUNIKACYJNE [7 kryteriów]
**ID obszaru: PT** | **4 MUST-HAVE + 3 SHOULD-HAVE**

#### PT-001 🔴 MUST-HAVE - Priorytet 1 | **Sesje: Wszystkie**
- **Kryterium**: Czy coach prowadzi rozmowę w języku ojczystym ucznia (podczas onboardingu i wyjaśnień)?
- **Weryfikacja**: Sprawdź język wypowiedzi coacha vs zadeklarowany język ojczysty ucznia
- **Ocena**: 1 = konsekwentny język ojczysty dla instrukcji/wyjaśnień, 0 = błędny język
- **Przykłady pozytywne**:
  1. Uczeń polski → Coach wyjaśnia po polsku "Dzisiaj będziemy ćwiczyć zamawianie w restauracji"
  2. Coach pyta o preferencje językowe przy niejednoznaczności
  3. Instrukcje gramatyczne w języku ojczystym
- **Przykłady negatywne**:
  1. Uczeń polski → Coach wyjaśnia tylko po angielsku
  2. Skomplikowane instrukcje w języku docelowym dla poziomu A1
  3. Mieszanie języków w sposób dezorientujący ucznia

#### PT-002 🔴 MUST-HAVE - Priorytet 1 | **Sesje: Wszystkie**
- **Kryterium**: Czy pierwszy token odpowiedzi pojawia się w ciągu 5 sekund?
- **Weryfikacja**: Sprawdź czas pierwszego tokenu od zakończenia wypowiedzi użytkownika
- **Ocena**: 1 = ≤5 sekund, 0 = >5 sekund
- **Przykłady pozytywne**:
  1. Płynna konwersacja bez długich przerw
  2. Natychmiastowa reakcja na odpowiedzi ucznia
- **Przykłady negatywne**:
  1. Opóźnienia powodujące utratę "flow" konwersacji
  2. Długie przerwy zniechęcające do ćwiczeń

#### PT-003 🔴 MUST-HAVE - Priorytet 1 | **Sesje: Wszystkie**
- **Kryterium**: Czy brak błędów technicznych w wypowiedziach?
- **Weryfikacja**: Szukaj [nawiasów], placeholderów, artefaktów kodu
- **Ocena**: 1 = brak błędów, 0 = ≥1 błąd
- **Przykłady pozytywne**:
  1. "Cześć! Jestem Aoki, twoim coachem językowym"
  2. Czyste, profesjonalne wypowiedzi
- **Przykłady negatywne**:
  1. "Cześć {user_name}, jak się uczysz?"
  2. "[LANGUAGE_TARGET] to świetny wybór!"
  3. "null, ile godzin tygodniowo możesz..."

#### PT-004 🔴 MUST-HAVE - Priorytet 1 | **Sesje: Wszystkie**
- **Kryterium**: Czy coach sprawdza zrozumienie po wprowadzeniu nowych pojęć/reguł?
- **Weryfikacja**: Szukaj pytań weryfikujących po wyjaśnieniach gramatycznych lub nowym słownictwie
- **Ocena**: 1 = sprawdzenie po ≥80% nowych pojęć, 0 = <80%
- **Przykłady pozytywne**:
  1. "Czy rozumiesz różnicę między ser i estar?"
  2. "Spróbuj teraz użyć tego słowa w zdaniu"
  3. "Jak byś powiedział to samo używając tej konstrukcji?"
- **Przykłady negatywne**:
  1. Wprowadzenie 10 nowych słów bez sprawdzenia zrozumienia
  2. Wyjaśnienie reguły gramatycznej i przejście dalej bez weryfikacji

#### PT-005 🟡 SHOULD-HAVE - Priorytet 2 | **Sesje: Wszystkie**
- **Kryterium**: Czy sesja nauki nie przekracza 45 minut (optymalny czas koncentracji)?
- **Weryfikacja**: Oblicz czas od pierwszej do ostatniej wypowiedzi sesji nauki
- **Ocena**: 1 = ≤45 min, 0 = >45 min bez przerwy
- **Przykłady pozytywne**:
  1. Sesja 30-minutowa z wyraźnym zakończeniem
  2. Coach proponuje przerwę po 40 minutach intensywnej nauki
- **Przykłady negatywne**:
  1. Sesja 90-minutowa bez sugestii przerwy
  2. Brak świadomości czasu trwania sesji

#### PT-006 🟡 SHOULD-HAVE - Priorytet 2 | **Sesje: Wszystkie**
- **Kryterium**: Czy odpowiedzi są merytorycznie związane z pytaniami ucznia?
- **Weryfikacja**: Sprawdź czy odpowiedź adresuje zadane pytanie
- **Ocena**: 1 = wszystkie na temat, 0 = ≥1 wymijająca
- **Przykłady pozytywne**:
  1. Uczeń: "A na co to wpływa?" → Coach wyjaśnia wpływ destynacji na dialekt/słownictwo
  2. Bezpośrednie odpowiedzi na wątpliwości ucznia
- **Przykłady negatywne**:
  1. Ignorowanie pytania i kontynuowanie skryptu
  2. Odpowiedź niezwiązana z pytaniem

#### PT-007 🟡 SHOULD-HAVE - Priorytet 2 | **Sesje: Wszystkie**
- **Kryterium**: Czy coach reaguje adekwatnie na bezsensowne odpowiedzi?
- **Weryfikacja**: Przy odpowiedziach typu "aaa", "xyz" sprawdź reakcję
- **Ocena**: 1 = prośba o wyjaśnienie, 0 = akceptacja nonsensu, N/A = brak nonsensów
- **Przykłady pozytywne**:
  1. "Nie jestem pewien czy dobrze zrozumiałem. Możesz powiedzieć inaczej?"
  2. "Czy chciałeś powiedzieć coś konkretnego?"
- **Przykłady negatywne**:
  1. Akceptacja "asdfgh" jako poprawnej odpowiedzi
  2. Kontynuowanie nauki po nonsensownej odpowiedzi

### OBSZAR 2: ONBOARDING - PROFIL UCZNIA [8 kryteriów]
**ID obszaru: ON** | **5 MUST-HAVE + 3 SHOULD-HAVE**

#### ON-001 🔴 MUST-HAVE - Priorytet 1 | **Sesje: Onboarding**
- **Kryterium**: Czy coach przedstawił się z imienia i wyjaśnił swoją rolę?
- **Weryfikacja**: Szukaj "Jestem Aoki" + opis roli (coach językowy) w pierwszych wypowiedziach
- **Ocena**: 1 = oba elementy obecne, 0 = brakuje choć jednego
- **Przykłady pozytywne**:
  1. "Cześć! Jestem Aoki, twoim coachem językowym"
  2. "Nazywam się Aoki i pomogę Ci w nauce języka"
- **Przykłady negatywne**:
  1. "Cześć! Jakiego języka chcesz się uczyć?" (bez przedstawienia)
  2. "Jestem Aoki" (bez wyjaśnienia roli)

#### ON-002 🔴 MUST-HAVE - Priorytet 1 | **Sesje: Onboarding**
- **Kryterium**: Czy coach zebrał kompletny profil ucznia (imię, język ojczysty, język docelowy)?
- **Weryfikacja**: Sprawdź czy wszystkie 3 kluczowe elementy zostały zebrane
- **Ocena**: 1 = wszystkie 3 zebrane, 0 = brakuje choć jednego
- **Przykłady pozytywne**:
  1. "Jak masz na imię?" → "Vladimir" + język docelowy + rozpoznanie języka ojczystego
  2. Systematyczne zbieranie danych profilowych
- **Przykłady negatywne**:
  1. Rozpoczęcie nauki bez znajomości imienia ucznia
  2. Brak informacji o języku ojczystym (jak tłumaczyć?)

#### ON-003 🔴 MUST-HAVE - Priorytet 1 | **Sesje: Onboarding**
- **Kryterium**: Czy coach ustalił obecny poziom znajomości języka (A1-C2)?
- **Weryfikacja**: Szukaj pytania o poziom lub samoocenę ucznia
- **Ocena**: 1 = poziom ustalony, 0 = brak ustalenia
- **Przykłady pozytywne**:
  1. "Jak oceniasz swój poziom? A1-A2, B1-B2, czy C1-C2?"
  2. "Jesteś początkujący, średniozaawansowany czy zaawansowany?"
- **Przykłady negatywne**:
  1. Rozpoczęcie nauki bez wiedzy o poziomie ucznia
  2. Założenie poziomu bez weryfikacji

#### ON-004 🔴 MUST-HAVE - Priorytet 1 | **Sesje: Onboarding**
- **Kryterium**: Czy coach zapytał o dotychczasowe doświadczenie w nauce języka?
- **Weryfikacja**: Szukaj pytania o czas nauki, metody, wcześniejsze doświadczenia
- **Ocena**: 1 = pytanie zadane, 0 = brak pytania
- **Przykłady pozytywne**:
  1. "Jak długo uczysz się hiszpańskiego?"
  2. "Jakie metody stosowałeś do nauki?"
  3. "Czy uczyłeś się przez aplikacje, zajęcia, czy na własną rękę?"
- **Przykłady negatywne**:
  1. Brak wiedzy o historii nauki ucznia
  2. Ignorowanie wcześniejszych doświadczeń

#### ON-005 🔴 MUST-HAVE - Priorytet 1 | **Sesje: Onboarding**
- **Kryterium**: Czy coach ustalił główną motywację do nauki języka?
- **Weryfikacja**: Szukaj pytania o cel/powód nauki (podróże, kariera, rodzina, etc.)
- **Ocena**: 1 = motywacja ustalona, 0 = brak ustalenia
- **Przykłady pozytywne**:
  1. "Czy masz jakiś szczególny cel w nauce? Podróże, kariera, czy coś innego?"
  2. "Co Cię motywuje do nauki hiszpańskiego?"
- **Przykłady negatywne**:
  1. Plan nauki bez znajomości motywacji ucznia
  2. Generyczny plan niezwiązany z celami ucznia

#### ON-006 🟡 SHOULD-HAVE - Priorytet 2 | **Sesje: Onboarding**
- **Kryterium**: Czy coach zapytał o preferowany styl nauki?
- **Weryfikacja**: Szukaj pytania o preferencje (wizualne, słuchowe, praktyczne)
- **Ocena**: 1 = pytanie zadane, 0 = brak pytania
- **Przykłady pozytywne**:
  1. "Jak najlepiej się uczysz? Przez słuchanie, czytanie, czy rozmowę?"
  2. "Wolisz więcej gramatyki czy praktycznej konwersacji?"
- **Przykłady negatywne**:
  1. Jednostajny styl nauki dla wszystkich uczniów
  2. Ignorowanie preferencji ucznia

#### ON-007 🟡 SHOULD-HAVE - Priorytet 2 | **Sesje: Onboarding**
- **Kryterium**: Czy coach zapytał o preferencje dotyczące korygowania błędów?
- **Weryfikacja**: Szukaj pytania o sposób feedback'u (natychmiastowy, na końcu, łagodny, bezpośredni)
- **Ocena**: 1 = pytanie zadane, 0 = brak pytania
- **Przykłady pozytywne**:
  1. "Jak chciałbyś, żebym korygował Twoje błędy? Od razu czy na końcu ćwiczenia?"
  2. "Wolisz bezpośrednią korektę czy delikatne sugestie?"
- **Przykłady negatywne**:
  1. Brak ustaleń dotyczących feedback'u
  2. Korekta w sposób demotywujący ucznia

#### ON-008 🟡 SHOULD-HAVE - Priorytet 2 | **Sesje: Onboarding**
- **Kryterium**: Czy coach zebrał informację o dostępnym czasie na naukę?
- **Weryfikacja**: Szukaj pytania o godziny tygodniowo lub częstotliwość sesji
- **Ocena**: 1 = informacja zebrana, 0 = brak informacji
- **Przykłady pozytywne**:
  1. "Ile czasu tygodniowo możesz poświęcić na naukę?"
  2. "Jak często chciałbyś ćwiczyć?"
- **Przykłady negatywne**:
  1. Plan 10h tygodniowo dla kogoś, kto ma 2h
  2. Brak dostosowania intensywności do możliwości

### OBSZAR 3: CEL NAUKI (SMART) [7 kryteriów]
**ID obszaru: CN** | **3 MUST-HAVE + 4 SHOULD-HAVE**

#### CN-001 🔴 MUST-HAVE - Priorytet 1 | **Sesje: Onboarding**
- **Kryterium**: Czy coach ustalił konkretny, mierzalny cel nauki?
- **Weryfikacja**: Sprawdź czy cel zawiera konkretne umiejętności do osiągnięcia
- **Ocena**: 1 = cel konkretny i mierzalny, 0 = cel ogólny
- **Przykłady pozytywne**:
  1. "Będę w stanie zamówić jedzenie w restauracji oraz rozmawiać w taksówce"
  2. "Zdać egzamin B2 Cambridge"
  3. "Przeprowadzić 30-minutową rozmowę biznesową po angielsku"
- **Przykłady negatywne**:
  1. "Chcę mówić lepiej po hiszpańsku" (zbyt ogólne)
  2. "Nauczyć się języka" (niemierzalne)
  3. "Być płynnym" (nieokreślone)

#### CN-002 🔴 MUST-HAVE - Priorytet 1 | **Sesje: Onboarding**
- **Kryterium**: Czy coach ustalił termin osiągnięcia celu?
- **Weryfikacja**: Szukaj konkretnego deadline'u (miesiące, data)
- **Ocena**: 1 = termin ustalony, 0 = brak terminu
- **Przykłady pozytywne**:
  1. "Do kiedy chciałbyś osiągnąć ten poziom?" → "6 miesięcy"
  2. "Przed wakacjami w lipcu"
  3. "Do końca roku"
- **Przykłady negatywne**:
  1. Plan bez określonego terminu
  2. "Kiedyś" jako deadline
  3. Brak pytania o ramy czasowe

#### CN-003 🔴 MUST-HAVE - Priorytet 1 | **Sesje: Onboarding**
- **Kryterium**: Czy coach ustalił kryteria sukcesu (jak uczeń pozna, że osiągnął cel)?
- **Weryfikacja**: Szukaj konkretnych wskaźników sukcesu
- **Ocena**: 1 = kryteria określone, 0 = brak kryteriów
- **Przykłady pozytywne**:
  1. "Będę mógł zamówić jedzenie i porozmawiać w taksówce"
  2. "Zdam egzamin na poziom B2"
  3. "Przeprowadzę rozmowę 15-minutową bez słownika"
- **Przykłady negatywne**:
  1. Cel bez określenia jak mierzyć sukces
  2. "Będę lepszy" bez konkretów

#### CN-004 🟡 SHOULD-HAVE - Priorytet 2 | **Sesje: Onboarding**
- **Kryterium**: Czy coach zweryfikował realność celu (A1→B2 w 6 miesięcy)?
- **Weryfikacja**: Szukaj komentarza o wykonalności przy ambitnych celach
- **Ocena**: 1 = weryfikacja obecna lub cel realistyczny, 0 = nierealistyczny cel bez komentarza
- **Przykłady pozytywne**:
  1. "To ambitny cel, ale przy regularnej nauce jest osiągalny"
  2. "Przy 5h tygodniowo w 6 miesięcy możesz osiągnąć solidny B1"
  3. Korekta oczekiwań przy nierealistycznym celu
- **Przykłady negatywne**:
  1. Akceptacja "C2 w 3 miesiące od zera" bez komentarza
  2. Brak weryfikacji wykonalności celu

#### CN-005 🟡 SHOULD-HAVE - Priorytet 2 | **Sesje: Onboarding**
- **Kryterium**: Czy cel jest powiązany z motywacją ucznia?
- **Weryfikacja**: Porównaj cel z podaną motywacją
- **Ocena**: 1 = powiązanie jasne, 0 = brak powiązania
- **Przykłady pozytywne**:
  1. Motywacja: podróże → Cel: komunikacja w restauracji/taksówce
  2. Motywacja: kariera → Cel: prezentacje biznesowe
- **Przykłady negatywne**:
  1. Motywacja: podróże → Cel: pisanie esejów akademickich
  2. Cel niezwiązany z deklarowaną motywacją

#### CN-006 🟡 SHOULD-HAVE - Priorytet 2 | **Sesje: Onboarding**
- **Kryterium**: Czy coach poprosił o wyjaśnienie celu własnymi słowami?
- **Weryfikacja**: Szukaj weryfikacji zrozumienia celu przez ucznia
- **Ocena**: 1 = weryfikacja obecna, 0 = brak weryfikacji
- **Przykłady pozytywne**:
  1. "Co konkretnie chciałbyś umieć powiedzieć w restauracji?"
  2. "Opisz mi typową sytuację, w której chciałbyś użyć hiszpańskiego"
- **Przykłady negatywne**:
  1. Akceptacja ogólnego celu bez doprecyzowania
  2. Brak pogłębienia rozumienia celu

#### CN-007 🟡 SHOULD-HAVE - Priorytet 2 | **Sesje: Onboarding**
- **Kryterium**: Czy coach uzyskał jawną akceptację celu od ucznia?
- **Weryfikacja**: Szukaj pytania o zgodę i potwierdzenia
- **Ocena**: 1 = akceptacja uzyskana, 0 = brak akceptacji
- **Przykłady pozytywne**:
  1. "Czy ten cel brzmi dla Ciebie dobrze?" → "Tak!"
  2. "Zgadzasz się z takim sformułowaniem celu?"
- **Przykłady negatywne**:
  1. Narzucenie celu bez pytania o zgodę

## Wyzwania skalowalności i granice uwagi

Możesz teraz zadać bardzo trafne pytanie: *"Skoro nowoczesne modele oferują okno kontekstowe rzędu 1 mln tokenów, a nasz kontrakt z dialogiem zajmuje 'tylko' 15-30k tokenów, to dlaczego po prostu nie wkleić całości i nie poprosić o ocenę?"*

Technicznie? Masz rację. To się zmieści. Model przyjmie to zapytanie.
Architektonicznie? Musimy tutaj wrócić do **trzech granic limitów**, o których mówiliśmy na początku kursu.

### Przypomnienie: Trzy Limity LLM

1.  **Limit Techniczny (Hard Limit):** To wspomniane 1 mln tokenów. Tyle model przyjmie, zanim wyrzuci błąd.
2.  **Limit Efektywny (Effective Limit):** To bezpieczna strefa (zazwyczaj ok. 32k tokenów), w której model radzi sobie z retrievalem (wyszukiwaniem informacji) bez większych problemów.
3.  **Limit Semantyczny (Semantic Limit):** To granica "ładunku poznawczego".

Nasz problem z 52 kryteriami leży właśnie w **limicie semantycznym**.

Nawet jeśli mieścimy się w oknie kontekstowym, zmuszanie modelu do równoczesnej analizy 52 skomplikowanych reguł logicznych, z których każda ma swoje niuanse i przykłady, drastycznie zwiększa ryzyko błędu. To zjawisko znane jako *attention dilution* (rozmycie uwagi). Model może poprawnie ocenić pierwsze 10 kryteriów, by przy 30. zacząć halucynować lub pomijać kluczowe szczegóły ("Lost in the Middle").

Dlatego, choć "wrzucenie wszystkiego naraz" jest kuszące, w systemach produkcyjnych klasy SOTA unikamy obciążania modelu dziesiątkami instrukcji warunkowych w jednym promptcie, jeśli zależy nam na wysokiej precyzji.

### Ryzyka podejścia "Wszystko w jednym"

Gdybyśmy poprosili model po prostu o "listę wyników", napotkalibyśmy problemy, które utrudniają automatyzację:

1.  **Niedeterministyczny format:** Raz model napisze "Kryterium PT-001: Spełnione", a innym razem "1. Język rozmowy: OK". Parsowanie tego wymagałoby skomplikowanych wyrażeń regularnych (RegEx).
2.  **Zgubione kryteria:** Przy 52 punktach istnieje ryzyko, że model zwróci wyniki tylko dla 50 z nich.
3.  **Halucynacje ocen:** Bez wymuszenia dowodów (cytatów), model może "zgadywać" ocenę, aby zadowolić użytkownika.

Potrzebujemy rozwiązania, które jest sztywne jak kontrakt w kodzie, ale elastyczne jak model językowy.

### Rozwiązanie: Structured Output

Odpowiedzią jest mechanizm **Structured Output**. Zamiast prosić model o "tekst", poprosimy go o wypełnienie konkretnego schematu danych. Brzmi logicznie, prawda?

## Parsowanie Markdown → Python

Skoro wiemy już, **co** chcemy osiągnąć (Structured Output) i **dlaczego** (stabilność), musimy przygotować dane. Nasz Leaderboard Card jest obecnie plikiem tekstowym. Dla Pythona to tylko długi string.

Musimy zamienić ten "płaski" tekst w listę strukturalnych obiektów.

Strategia parsowania jest prosta i opiera się na **kotwicach** (anchors), którymi są nasze ID w nagłówkach. Nie potrzebujemy tutaj LLM-a (szkoda tokenów i pieniędzy). Wystarczy solidny RegEx i logika inżynierska.

Algorytm, który zaimplementujemy w kolejnych krokach, wygląda tak:

3.  Wyciągamy treść między nagłówkiem a sekcją przykładów jako `Opis`.
4.  Wyciągamy listy `Przykłady pozytywne` i `Przykłady negatywne`.

```python
def parse_leaderboard_card(markdown: str) -> List[CheckDefinition]:
    checks = []
    
    # Dzielimy dokument po nagłówkach H4 (####), zachowując separator w pamięci lub lookahead
    # Tutaj prościej: split i obróbka każdej sekcji
    sections = re.split(r'\n(?=#### )', markdown)
    
    for section in sections:
        section = section.strip()
        
        # Szukamy wzorca ID w pierwszej linii, np. "#### PT-001"
        id_match = re.search(r'#### ([A-Z]{2}-\d{3})', section)
        if not id_match:
            continue
            
        check_id = id_match.group(1)
        area_code = check_id.split('-')[0]
        
        # Wyciągamy priorytet
        if "MUST-HAVE" in section.split('\n')[0] or "🔴" in section:
            priority = "MUST_HAVE"
        else:
            priority = "SHOULD_HAVE"
            
        # Wyciągamy tytuł (reszta linii po priorytecie i ID)
        # Zakładamy format: #### ID Emoji PRIORITY | **Tytuł** lub podobny
        # Dla uproszczenia bierzemy linię nagłówka, ale w produkcji warto to oczyścić
        header_line = section.split('\n')[0]
        title = header_line.split('|')[-1].strip().strip('*')
        
        # Parsowanie body
        # Szukamy markerów sekcji
        desc_start = section.find('\n') # Start opisu to koniec pierwszej linii
        pos_ex_start = section.find('- **Przykłady pozytywne**')
        neg_ex_start = section.find('- **Przykłady negatywne**')
        
        # Wyciągamy Opis
        # Opis jest od nagłówka do pierwszych przykładów
        limit = min(pos for pos in [pos_ex_start, neg_ex_start] if pos != -1)
        description = section[desc_start:limit].strip()
        
        # Helper do wyciągania punktów z listy
        def extract_points(text_block):
            points = []
            if text_block:
                # Szukamy linii zaczynających się od cyfry i kropki np "  1. "
                for line in text_block.split('\n'):
                    line = line.strip()
                    if re.match(r'^\d+\.', line):
                        # Usuwamy "1. " z początku
                        clean_line = re.sub(r'^\d+\.\s*', '', line)
                        points.append(clean_line)
            return points

        # Wyciągamy przykłady
        positive_block = ""
        negative_block = ""
        
        if pos_ex_start != -1:
            end = neg_ex_start if neg_ex_start != -1 else len(section)
            positive_block = section[pos_ex_start:end]
            
        if neg_ex_start != -1:
            negative_block = section[neg_ex_start:]
            
        checks.append(CheckDefinition(
            id=check_id,
            area=area_code,
            priority=priority,
            title=title,
            description=description,
            positive_examples=extract_points(positive_block),
            negative_examples=extract_points(negative_block)
        ))
        
    return checks
```

3.  **User Message:** Dostarcza **WSAD** (dialog).

```python
def evaluate_dialog(dialog_json: str, checks: List[CheckDefinition]) -> BaseModel:
    """
    Orkiestrator procesu ewaluacji. Łączy dane, model i API.
    """
    logger.info("🚀 Rozpoczynam ewaluację dialogu...")
    
    # 1. Parsowanie danych wejściowych
    try:
        data = json.loads(dialog_json)
        # Obsługa struktury z pliku ai_tutor_1.json
        messages_list = data.get("conversation_history", [])
        
        # Fallback, gdyby struktura była inna (np. messages)
        if not messages_list:
            messages_list = data.get("messages", [])
            
    except json.JSONDecodeError:
        logger.error("Błąd parsowania JSON z dialogiem")
        raise

    # 2. Formatowanie dialogu dla Sędziego (User experience modelu)
    formatted_dialog = "TRANSKRYPT ROZMOWY:\n" + "="*20 + "\n"
    for msg in messages_list:
        role = msg.get("role", "unknown").upper()
        content = msg.get("content", "")
        formatted_dialog += f"[{role}]: {content}\n\n"
        
    # 3. Dynamiczne generowanie schematu (To jest ten moment!)
    # Tworzymy klasę Pydantic idealnie dopasowaną do kontraktu
    EvaluationSchema = create_evaluation_model(checks)
    logger.debug(f"Wygenerowano schemat z {len(EvaluationSchema.model_fields)} polami")

    # 4. Przygotowanie payloadu dla API
    messages = [
        {"role": "system", "content": JUDGE_SYSTEM_PROMPT},
        {"role": "user", "content": formatted_dialog}
    ]
    
    # 5. Wywołanie API (Structured Output)
    try:
        # Helper h.call_api zwraca już wypełniony obiekt Pydantic
        evaluation_result = h.call_api(messages, response_model=EvaluationSchema)
        logger.success("✅ Otrzymano ocenę od sędziego")
        return evaluation_result
    except Exception as e:
        logger.critical(f"Błąd podczas wywołania API: {e}")
```

```python
def generate_report(results: BaseModel, checks: List[CheckDefinition]) -> str:
    """Generuje czytelny raport Markdown z wynikami ewaluacji."""
    
    stats = {
        "MUST_HAVE": {"passed": 0, "total": 0},
        "SHOULD_HAVE": {"passed": 0, "total": 0}
    }
    failures = []

    for check in checks:
        safe_id = check.safe_id
        # Używamy getattr, bo nazwy pól są dynamiczne
        passed = getattr(results, f"{safe_id}_passed")
        reasoning = getattr(results, f"{safe_id}_reasoning")
        
        prio = check.priority
        stats[prio]["total"] += 1
        if passed:
            stats[prio]["passed"] += 1
        else:
            failures.append(f"❌ {check.id} ({prio}): {check.title}\n   💡 {reasoning}")

    # Obliczenia
    def calc_score(prio):
        total = stats[prio]["total"]
        if total == 0: return 100.0 # Dajemy 100% jeśli brak kryteriów w tej grupie
        return (stats[prio]["passed"] / total) * 100

    must_score = calc_score("MUST_HAVE")
    should_score = calc_score("SHOULD_HAVE")
    
    # Logika przyznawania odznak (zgodnie z Leaderboard Card)
    badge = "❌ ODRZUCONY"
    if must_score == 100:
        if should_score >= 90: badge = "🥇 DIAMOND LEVEL"
        elif should_score >= 80: badge = "🥈 TRUSTWORTHY"
        elif should_score >= 60: badge = "🥉 ACCEPTABLE"
        else: badge = "⚠️ WYMAGA POPRAWY (Low Quality)"
    elif must_score >= 90:
        badge = "⚠️ WYMAGA POPRAWY (Critical Issues)"
    
    # Budowanie raportu
    report = []
    report.append(f"# Raport Jakości AI Coach\n")
    report.append(f"### STATUS: {badge}\n")
    report.append(f"| Kategoria | Wynik | Postęp |")
    report.append(f"|-----------|-------|--------|")
    report.append(f"| 🔴 MUST-HAVE | **{must_score:.1f}%** | {stats['MUST_HAVE']['passed']}/{stats['MUST_HAVE']['total']} |")
    report.append(f"| 🟡 SHOULD-HAVE | **{should_score:.1f}%** | {stats['SHOULD_HAVE']['passed']}/{stats['SHOULD_HAVE']['total']} |")
    
    if failures:
        report.append("\n### 🚨 Wykryte Problemy:")
        report.append("\n".join(failures))
    else:
        report.append("\n### 🎉 Gratulacje! Brak wykrytych problemów.")
        
    return "\n".join(report)
```
